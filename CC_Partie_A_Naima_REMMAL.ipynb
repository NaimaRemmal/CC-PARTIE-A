## 1. Régression linéaire

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
from sklearn.linear_model import LinearRegression
modele=LinearRegression().fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
````

### 1. Présentation
* **Nom (FR / EN) :** Régression linéaire / *Linear Regression*.
* **Famille :** Apprentissage supervisé — Régression.
* **Objectif :** Prédire une variable cible continue $y$ à partir d'une combinaison linéaire de variables explicatives $X$.
* **Intuition & Fonctionnement :**
  L'algorithme cherche à ajuster la meilleure droite (ou hyperplan en multi-dimensions) passant à travers le nuage de points de données.
  1. *Entrée :* Matrice de données $X \in \mathbb{R}^{n \times p}$ et vecteur cible $y \in \mathbb{R}^n$.
  2. *Calcul/Apprentissage :* Minimisation de la somme des carrés des écarts (Moindres Carrés Ordinaires - MCO) entre la cible réelle et la valeur prédite.
  3. *Sortie :* Une fonction de prédiction $f(X) = \hat{y}$ qui attribue une valeur numérique continue à toute nouvelle observation.
* **Formule mathématique :**
  $$\hat{y} = \beta_0 + \sum_{j=1}^{p} \beta_j x_j = X\beta$$
  Fonction de perte (Moindres Carrés Ordinaires) :
  $$\mathcal{L}(\beta) = \frac{1}{2n} \|y - X\beta\|^2_2 = \frac{1}{2n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$
  - **Définition des symboles :** $y_i$ est la vraie valeur, $\hat{y}_i$ la valeur prédite, $x_j$ les variables explicatives, et $n$ le nombre d'observations.
  - **Paramètres (appris) :** Coefficients d'onde/pentes $\beta_j$ et biais/ordonnée à l'origine $\beta_0$.
  - **Hyperparamètres :** Aucun dans la version analytique standard des MCO (solution fermée $\beta = (X^T X)^{-1} X^T y$).
* **Avantages :**
  1. Très rapide à calculer et très sobre en ressources calculatoires.
  2. Modèle entièrement interprétable : chaque coefficient $\beta_j$ indique la variation exacte de $y$ pour une augmentation d'une unité de $x_j$.
* **Limites :**
  1. Hypothèse forte de linéarité entre les variables d'entrée et la cible.
  2. Extrêmement sensible aux valeurs extrêmes (*outliers*) et à la multicolinéarité entre les variables $X$.

---


### 2. Schéma explicatif

````python
import matplotlib.pyplot as plt
import numpy as np

# Schéma explicatif de la Régression Linéaire
np.random.seed(42)
x_demo = np.linspace(0, 10, 30)
y_demo = 2 * x_demo + 1 + np.random.normal(0, 2, 30)

plt.figure(figsize=(7, 4))
plt.scatter(
    x_demo, y_demo, color="blue", label="Données réelles (Observations)"
)
plt.plot(
    x_demo, 2 * x_demo + 1, color="red", linewidth=2, label="Droite MCO (Modèle)"
)

# Tracer un exemple de résidu
plt.vlines(
    x_demo[15],
    y_demo[15],
    2 * x_demo[15] + 1,
    color="gray",
    linestyle="--",
    label="Résidu (Écart e_i)",
)

plt.title("Principe de la Régression Linéaire (Moindres Carrés)")
plt.xlabel("Variable explicative X")
plt.ylabel("Cible continue y")
plt.legend()
plt.tight_layout()
plt.show()
````

* **Commentaire du schéma :** La figure illustre le nuage de points d'observation (en bleu) et la droite de régression optimale (en rouge) qui minimise la distance verticale globale (les résidus en pointillés gris) entre la droite et chaque point d'observation.


### 3. Contextes d'utilisation

* **Cas d'usage 2 :** Estimation de la consommation d'énergie d'un bâtiment selon la température extérieure.
* **Nature des données & Prérequis :** Variables explicatives numériques continues ou catégorielles encodées. Nécessite une absence de multicolinéarité parfaite.
* **Cas peu approprié :** Prédiction de séries temporelles très non linéaires (ex: cours d'une crypto-monnaie) ou classification (la cible est une catégorie et non une quantité).
* **Métriques adaptées :** MAE (Erreur Absolue Moyenne), RMSE (Racine de l'Erreur Quadratique Moyenne) et $R^2$ (Coefficient de détermination).

## 2. Ridge

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),Ridge(alpha=1.0)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
````

### 1. Présentation
* **Nom (FR / EN) :** Régression Ridge / *Ridge Regression (L2 Regularization)*.
* **Famille :** Apprentissage supervisé — Régression avec régularisation.
* **Objectif :** Prédire une variable continue $y$ en pénalisant les grands coefficients pour éviter le surapprentissage et traiter la multicolinéarité.
* **Intuition & Fonctionnement :**
  L'algorithme modifie les Moindres Carrés Ordinaires en ajoutant une pénalité sur la somme des carrés des coefficients (norme $L_2$).
  1. *Entrée :* Données $X$ (préalablement standardisées) et cible $y$.
  2. *Calcul/Apprentissage :* Minimisation de la perte MCO + pénalité $L_2$. Les coefficients sont contractés (*shrinkage*) vers zéro sans jamais devenir exactement nuls.
  3. *Sortie :* Vecteur de prédictions $\hat{y}$ plus stable et moins sensible au bruit des données d'entraînement.
* **Formule mathématique :**
  $$\mathcal{L}_{Ridge}(\beta) = \frac{1}{2n} \|y - X\beta\|^2_2 + \alpha \|\beta\|^2_2 = \frac{1}{2n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 + \alpha \sum_{j=1}^{p} \beta_j^2$$
  - **Définition des symboles :** $y_i$ est la valeur réelle, $\hat{y}_i$ la prédiction, $\beta_j$ les coefficients, $n$ le nombre d'échantillons, $\alpha$ le coefficient de pénalité.
  - **Paramètres (appris) :** Coefficients $\beta$.
  - **Hyperparamètres :** $\alpha \ge 0$ (intensité de la régularisation).
* **Avantages :**
  1. Très performant en présence de forte multicolinéarité entre variables.
  2. Réduit efficacement la variance du modèle sans augmenter excessivement le biais.
* **Limites :**
  1. Ne réalise pas de sélection de variables (aucun coefficient n'est annulé).
  2. Nécessite obligatoirement une standardisation préalable des variables.



### 2. Schéma explicatif

````python
import matplotlib.pyplot as plt
import numpy as np

alphas = np.logspace(-2, 3, 50)
coefs = []

# Génération de données synthétiques corrélées
np.random.seed(42)
X_dummy = np.random.randn(50, 5)
y_dummy = X_dummy @ np.array([3, -2, 1.5, 0, 0.5]) + np.random.randn(50)

from sklearn.linear_model import Ridge
from sklearn.preprocessing import StandardScaler

X_scaled = StandardScaler().fit_transform(X_dummy)
for a in alphas:
    ridge = Ridge(alpha=a)
    ridge.fit(X_scaled, y_dummy)
    coefs.append(ridge.coef_)

plt.figure(figsize=(8, 4.5))
plt.plot(alphas, coefs)
plt.xscale('log')
plt.xlabel('Hyperparamètre alpha (Intensité de régularisation L2)')
plt.ylabel('Coefficients beta')
plt.title('Schéma explicatif : Contraction des coefficients par la régression Ridge')
plt.grid(True, which="both", ls="--")
plt.tight_layout()
plt.show()
````

* **Commentaire du schéma :** Le graphique montre l'effet de contraction (*shrinkage*) : à mesure que l'hyperparamètre $\alpha$ augmente, la magnitude des coefficients $\beta$ diminue vers zéro, lissant la fonction de prédiction.


### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Économétrie) :** Modélisation des rendements boursiers où de nombreux indicateurs macroéconomiques sont fortement corrélés entre eux.
* **Cas d'usage 2 :** Estimation de prix immobiliers avec des dizaines de caractéristiques corrélées (surface, nombre de pièces, année).
* **Nature des données & Prérequis :** Variables numériques obligatoirement standardisées (`StandardScaler`).
* **Cas peu approprié :** Jeux de données avec un très grand nombre de variables non pertinentes où une sélection de variables est nécessaire.
* **Métriques adaptées :** MAE, RMSE, $R^2$.


### 4. Analyse de l'expérience Python
* **Entrées du code :** `Xr_ent` et `yr_ent` passés dans un `Pipeline` combinant `StandardScaler()` et `Ridge(alpha=1.0)`.
* **Hyperparamètres définis :** `alpha=1.0`.
* **Sortie obtenue :** Affiche la MAE calculée sur `Xr_test`.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Valide l'intégration de la standardisation et du calcul du score sur les données de test synthétiques.
  - *Ne permet PAS de conclure :* Ne permet pas de savoir si `alpha=1.0` est la valeur optimale (nécessiterait une validation croisée `RidgeCV`).

## 3. Lasso

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
from sklearn.linear_model import Lasso
modele=make_pipeline(StandardScaler(),Lasso(alpha=.1,max_iter=10000)).fit(Xr_ent,yr_ent)
print('R²:',r2_score(yr_test,modele.predict(Xr_test)))
````

### 1. Présentation
* **Nom (FR / EN) :** Lasso / *Lasso Regression (L1 Regularization - Least Absolute Shrinkage and Selection Operator)*.
* **Famille :** Apprentissage supervisé — Régression avec sélection de variables.
* **Objectif :** Prédire une variable continue tout en sélectionnant automatiquement les variables les plus pertinentes en annulant les coefficients inutiles.
* **Intuition & Fonctionnement :**
  Lasso ajoute une pénalité sur la somme des valeurs absolues des coefficients (norme $L_1$).
  1. *Entrée :* Données $X$ standardisées et cible $y$.
  2. *Calcul/Apprentissage :* Minimisation de la perte MCO + pénalité $L_1$. En raison de la géométrie de la norme $L_1$ (coins pointus), certains coefficients deviennent exactement égaux à zéro.
  3. *Sortie :* Modèle parcimonieux (*sparse*) utilisant uniquement un sous-ensemble des variables d'origine.
* **Formule mathématique :**
  $$\mathcal{L}_{Lasso}(\beta) = \frac{1}{2n} \|y - X\beta\|^2_2 + \alpha \|\beta\|_1 = \frac{1}{2n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 + \alpha \sum_{j=1}^{p} |\beta_j|$$
  - **Définition des symboles :** $y_i$ valeur réelle, $\hat{y}_i$ prédiction, $\beta_j$ coefficients, $\alpha$ intensité de la pénalité $L_1$.
  - **Paramètres (appris) :** Coefficients $\beta$.
  - **Hyperparamètres :** $\alpha \ge 0$, `max_iter` (nombre maximal d'itérations du solveur).
* **Avantages :**
  1. Réalise une sélection automatique de caractéristiques (*feature selection*).
  2. Rend le modèle final très lisible et réduit le bruit en éliminant les variables non pertinentes.
* **Limites :**
  1. Si deux variables sont très corrélées, Lasso en choisit une arbitrairement et met l'autre à zéro.
  2. Si $p > n$ (plus de variables que d'observations), Lasso ne peut sélectionner au maximum que $n$ variables.


### 2. Schéma explicatif

````python
from sklearn.linear_model import Lasso

alphas = np.logspace(-2, 1, 50)
coefs_lasso = []

for a in alphas:
    lasso = Lasso(alpha=a, max_iter=10000)
    lasso.fit(X_scaled, y_dummy)
    coefs_lasso.append(lasso.coef_)

plt.figure(figsize=(8, 4.5))
plt.plot(alphas, coefs_lasso)
plt.xscale('log')
plt.axhline(0, color='black', linestyle='--', alpha=0.5)
plt.xlabel('Hyperparamètre alpha (Intensité L1)')
plt.ylabel('Coefficients beta')
plt.title('Schéma explicatif : Annulation des coefficients par la régression Lasso')
plt.grid(True, which="both", ls="--")
plt.tight_layout()
plt.show()
````

**Commentaire du schéma :** Le graphique illustre l'annulation des coefficients : lorsque $\alpha$ augmente, les trajectoires de plusieurs coefficients atteignent la ligne zéro (sélection automatique des variables).


### 3. Contextes d'utilisation
* **Cas d'usage 1 (Marketing / Scoring) :** Sélection des variables prédictives parmi des centaines de métriques de comportement client en ligne pour modéliser le panier moyen.
* **Cas d'usage 2 (Génomique / Bioinformatique) :** Identification des quelques gènes responsables d'une maladie parmi des milliers de gènes analysés.
* **Nature des données & Prérequis :** Variables numériques obligatoirement standardisées (`StandardScaler`).
* **Cas peu approprié :** Situations où toutes les variables apportent une petite contribution utile et corrélée (préérer Ridge).
* **Métriques adaptées :** $R^2$, MAE, RMSE.


### 4. Analyse de l'expérience Python
* **Entrées du code :** `Xr_ent` et `yr_ent` dans un `Pipeline` avec `StandardScaler()` et `Lasso(alpha=.1, max_iter=10000)`.
* **Hyperparamètres définis :** `alpha=0.1`, `max_iter=10000`.
* **Sortie obtenue :** Affiche le coefficient de détermination ($R^2$).
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Mesure la proportion de variance expliquée par le modèle Lasso sur les données de test.
  - *Ne permet PAS de conclure :* Ne montre pas explicitement quelles variables ont été éliminées (coefficients mis à zéro).

## 4. Régression polynomiale

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),PolynomialFeatures(degree=2,include_bias=False),Ridge(alpha=10)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
````

### 1. Présentation
* **Nom (FR / EN) :** Régression polynomiale / *Polynomial Regression*.
* **Famille :** Apprentissage supervisé — Régression non linéaire.
* **Objectif :** Capter des relations non linéaires entre les variables explicatives et la cible en générant des termes polynomiaux et d'interaction.
* **Intuition & Fonctionnement :**
  L'algorithme transforme l'espace des données de départ en créant des combinaisons de puissances ($x^2, x^3, ...$) et de produits croisés ($x_1 x_2$), puis applique une régression linéaire sur cet espace élargi.
  1. *Entrée :* Variables $X$.
  2. *Calcul/Apprentissage :* Transformation via `PolynomialFeatures(degree=d)` suivie de l'ajustement d'un modèle linéaire (souvent régularisé avec Ridge/Lasso).
  3. *Sortie :* Courbe ou surface d'ajustement non linéaire.
* **Formule mathématique :**
  $$\hat{y} = \beta_0 + \beta_1 x + \beta_2 x^2 + \dots + \beta_d x^d$$
  - **Définition des symboles :** $d$ est le degré du polynôme, $\beta_k$ sont les coefficients associés aux puissances $x^k$.
  - **Paramètres (appris) :** Coefficients $\beta$.
  - **Hyperparamètres :** Degré $d$ (`degree`), inclusion du terme de biais (`include_bias`).
* **Avantages :**
  1. Permet de modéliser des courbes et des tendances complexes sans changer de framework linéaire.
  2. Conserve une résolution analytique directe via les MCO/Ridge.
* **Limites :**
  1. Explosion combinatoire du nombre de caractéristiques quand le nombre de variables et le degré augmentent.
  2. Risque élevé de surapprentissage (*overfitting*) sur les bords du domaine de données.


### 2. Schéma explicatif

````python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures

np.random.seed(42)
x_poly = np.linspace(-3, 3, 40)
y_poly = 0.5 * x_poly**2 - x_poly + 1 + np.random.normal(0, 1, 40)

pipe_poly = make_pipeline(
    PolynomialFeatures(degree=2), Ridge(alpha=0.1)
)
pipe_poly.fit(x_poly.reshape(-1, 1), y_poly)

plt.figure(figsize=(8, 4.5))
plt.scatter(x_poly, y_poly, color='blue', label='Données réelles')
plt.plot(
    x_poly,
    pipe_poly.predict(x_poly.reshape(-1, 1)),
    color='red',
    linewidth=2,
    label='Ajustement Polynomial (Degré 2)',
)
plt.title('Schéma explicatif : Régression Polynomiale de degré 2')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
plt.tight_layout()
plt.show()
````

**Commentaire du schéma :** Le graphique compare l'ajustement linéaire classique (droit) et l'ajustement polynomial (courbe), montrant comment le polynôme capture la courbure naturelle des données.


### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Modélisation de rendements) :** Modélisation de la courbe des taux d'intérêt ou prévision du coût marginal de production selon le volume (économies d'échelle puis déséconomies d'échelle).
* **Cas d'usage 2 :** Estimation de la trajectoire de croissance des ventes d'un nouveau produit après son lancement (courbe en S ou parabole).
* **Nature des données & Prérequis :** Variables numériques. La régularisation (Ridge/Lasso) est fortement recommandée.
* **Cas peu approprié :** Problèmes comportant un très grand nombre de variables d'entrée (ex: 100 variables à un degré 3 créent des centaines de milliers de colonnes).
* **Métriques adaptées :** MAE, RMSE, $R^2$.


### 4. Analyse de l'expérience Python
* **Entrées du code :** `Xr_ent` et `yr_ent` dans un pipeline composé de `StandardScaler()`, `PolynomialFeatures(degree=2, include_bias=False)` et `Ridge(alpha=10)`.
* **Hyperparamètres définis :** `degree=2`, `alpha=10`.
* **Sortie obtenue :** Affiche la MAE sur le jeu de test.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Montre le score de l'intégration polynomiale de degré 2 régularisée sur le jeu de test.
  - *Ne permet PAS de conclure :* Ne montre pas si un degré plus élevé (ex: 3 ou 4) améliorerait le résultat ou provoquerait du surapprentissage.

## 5. Arbre de régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
from sklearn.tree import DecisionTreeRegressor
modele=DecisionTreeRegressor(max_depth=5,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
````

### 1. Présentation
* **Nom (FR / EN) :** Arbre de décision pour la régression / *Decision Tree Regressor*.
* **Famille :** Apprentissage supervisé — Arbres de décision / Non paramétrique.
* **Objectif :** Prédire une cible continue en découpant l'espace des variables en régions rectangulaires disjointes.
* **Intuition & Fonctionnement :**
  L'algorithme sépare récursivement les données selon des seuils sur les variables qui réduisent au maximum la variance de la cible dans chaque sous-groupe (feuille).
  1. *Entrée :* Données $X$ et cible $y$.
  2. *Calcul/Apprentissage :* À chaque nœud, recherche de la meilleure variable $j$ et du meilleur seuil $t$ pour minimiser la somme des erreurs quadratiques des deux sous-groupes créés.
  3. *Sortie :* La prédiction pour une feuille est la moyenne des valeurs cibles des observations qui s'y trouvent.
* **Formule mathématique :**
  Critère de séparation (minimisation de la variance) :
  $$\text{Impureté}(R) = \frac{1}{|R|} \sum_{i \in R} (y_i - \bar{y}_R)^2$$
  - **Définition des symboles :** $R$ est la région (nœud), $|R|$ le nombre d'échantillons dans la région, $\bar{y}_R$ la moyenne des valeurs $y$ dans cette région.
  - **Paramètres (appris) :** Structure de l'arbre, variables de coupure, seuils et valeurs des feuilles $\bar{y}_R$.
  - **Hyperparamètres :** `max_depth` (profondeur max), `min_samples_split`, `min_samples_leaf`.
* **Avantages :**
  1. Modèle intuitif et visuellement interprétable sous forme de diagramme décisionnel.
  2. Ne nécessite aucune mise à l'échelle des données et gère les relations non linéaires complexes.
* **Limites :**
  1. Risque très élevé de surapprentissage si la profondeur n'est pas restreinte.
  2. Prédictions constantes par morceaux (incapable d'extrapoler en dehors de la plage d'entraînement).


### 2. Schéma explicatif

````python
from sklearn.tree import DecisionTreeRegressor

tree_reg = DecisionTreeRegressor(max_depth=3, random_state=42)
tree_reg.fit(x_poly.reshape(-1, 1), y_poly)

plt.figure(figsize=(8, 4.5))
plt.scatter(x_poly, y_poly, color='blue', label='Données réelles')
plt.step(
    x_poly,
    tree_reg.predict(x_poly.reshape(-1, 1)),
    color='red',
    linewidth=2,
    where='mid',
    label='Arbre (Profondeur=3)',
)
plt.title('Schéma explicatif : Prédiction par paliers d\'un Arbre de Régression')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
plt.tight_layout()
plt.show()
````

**Commentaire du schéma :** Le graphique montre comment un arbre de régression crée une fonction en escalier (prédictions constantes par morceaux dans chaque feuille).


### 3. Contextes d'utilisation
* **Cas d'usage 1 (Gestion / Immobilier) :** Estimation de la valeur d'un actif immobilier avec des règles simples et explicables aux clients (ex: "Si surface > 80m² et pièces > 3 alors valeur = X").
* **Cas d'usage 2 (Logistique) :** Estimation du temps de livraison en fonction du secteur et du créneau horaire.
* **Nature des données & Prérequis :** Variables numériques ou catégorielles encodées. Aucune standardisation requise.
* **Cas peu approprié :** Données nécessitant une extrapolation continue précise en dehors des bornes d'entraînement.
* **Métriques adaptées :** MAE, RMSE, $R^2$.


### 4. Analyse de l'expérience Python
* **Entrées du code :** `DecisionTreeRegressor(max_depth=5, random_state=42)` entraîné sur `Xr_ent` et `yr_ent`.
* **Hyperparamètres définis :** `max_depth=5`, `random_state=GRAINE`.
* **Sortie obtenue :** Affiche la MAE sur l'échantillon de test.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Mesure l'erreur d'un arbre limité à une profondeur de 5.
  - *Ne permet PAS de conclure :* Ne permet pas d'évaluer le surapprentissage sans comparer ce résultat au score d'entraînement (`Xr_ent`).

## 6. Forêt aléatoire — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
from sklearn.ensemble import RandomForestRegressor
modele=RandomForestRegressor(n_estimators=80,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
````

### 1. Présentation
* **Nom (FR / EN) :** Forêt aléatoire pour la régression / *Random Forest Regressor*.
* **Famille :** Apprentissage supervisé — Méthodes d'ensemble (*Bagging* / Agrégation).
* **Objectif :** Améliorer la précision et réduire la variance de la prédiction en combinant un grand nombre d'arbres de décision indépendants.
* **Intuition & Fonctionnement :**
  L'algorithme construit un ensemble d'arbres de décision entraînés chacun sur un sous-échantillon bootstrap des données et un sous-ensemble aléatoire des variables.
  1. *Entrée :* Données $X$ et cible $y$.
  2. *Calcul/Apprentissage :* Entraînement en parallèle de $B$ arbres de régression sur des tirages avec remise et sous-sélection aléatoire de variables à chaque nœud.
  3. *Sortie :* La prédiction finale est la moyenne exacte des prédictions de tous les arbres individuels.
* **Formule mathématique :**
  $$\hat{y}_{RF} = \frac{1}{B} \sum_{b=1}^{B} T_b(x)$$
  - **Définition des symboles :** $B$ est le nombre total d'arbres (`n_estimators`), $T_b(x)$ est la prédiction issue du $b$-ième arbre de décision.
  - **Paramètres (appris) :** L'ensemble des structures et seuils de tous les $B$ arbres.
  - **Hyperparamètres :** `n_estimators`, `max_depth`, `max_features`, `min_samples_split`.
* **Avantages :**
  1. Très robuste au surapprentissage par rapport à un arbre individuel.
  2. Offre des mesures fiables d'importance des variables.
* **Limites :**
  1. Coût en calcul et mémoire plus élevé qu'un simple arbre.
  2. Modèle en "boîte noire" difficile à interpréter visuellement par rapport à un seul arbre.


### 2. Schéma explicatif

````python
from sklearn.ensemble import RandomForestRegressor

rf_reg = RandomForestRegressor(n_estimators=10, random_state=42)
rf_reg.fit(x_poly.reshape(-1, 1), y_poly)

plt.figure(figsize=(8, 4.5))
plt.scatter(x_poly, y_poly, color='blue', alpha=0.5, label='Données réelles')

# Afficher quelques arbres individuels
for i, tree in enumerate(rf_reg.estimators_[:3]):
    plt.plot(
        x_poly,
        tree.predict(x_poly.reshape(-1, 1)),
        linestyle='--',
        alpha=0.5,
        label=f'Arbre individuel {i+1}',
    )

plt.plot(
    x_poly,
    rf_reg.predict(x_poly.reshape(-1, 1)),
    color='red',
    linewidth=2.5,
    label='Forêt Aléatoire (Moyenne)',
)
plt.title(
    'Schéma explicatif : Lissage par moyenne d\'arbres (Forêt Aléatoire)'
)
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
plt.tight_layout()
plt.show()
````

**Commentaire du schéma :** Le schéma représente le principe du Bagging : tirage de sous-échantillons bootstrap $\rightarrow$ entraînement parallèle d'arbres $\rightarrow$ moyenne des prédictions pour l'obtention du résultat final.


### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Valorisation) :** Prédiction du prix d'actifs financiers ou d'entreprises complexes à partir de données financières hétérogènes.
* **Cas d'usage 2 (Marketing) :** Prédiction de la valeur à vie d'un client (*Customer Lifetime Value - CLV*).
* **Nature des données & Prérequis :** Accepte des données mixtes (continues et catégorielles encodées). Aucun prétraitement de mise à l'échelle nécessaire.
* **Cas peu approprié :** Prédiction avec extrapolation forte sur des séries temporelles (comme tous les modèles à base d'arbres).
* **Métriques adaptées :** MAE, RMSE, $R^2$.


### 4. Analyse de l'expérience Python
* **Entrées du code :** `RandomForestRegressor(n_estimators=80, random_state=42)` ajusté sur `Xr_ent` et `yr_ent`.
* **Hyperparamètres définis :** `n_estimators=80`, `random_state=GRAINE`.
* **Sortie obtenue :** Affiche la MAE sur l'échantillon de test `Xr_test`.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Évalue la performance moyenne de l'agrégation de 80 arbres de décision sur l'échantillon de test.
  - *Ne permet PAS de conclure :* Ne montre pas si 80 arbres suffisent à stabiliser le score ou si rajouter des arbres améliorerait encore la MAE.

## 7. Gradient Boosting — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
from sklearn.ensemble import HistGradientBoostingRegressor
modele=HistGradientBoostingRegressor(max_iter=80,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
````

### 1. Présentation
* **Nom (FR / EN) :** Gradient Boosting pour la régression / *Gradient Boosting Regressor (HistGradientBoostingRegressor)*.
* **Famille :** Apprentissage supervisé — Méthodes d'ensemble séquentiel (*Boosting*).
* **Objectif :** Obtenir un modèle de prédiction très performant en construisant séquentiellement des arbres où chaque nouvel arbre corrige les erreurs (résidus) des précédents.
* **Intuition & Fonctionnement :**
  Contrairement à la Forêt Aléatoire qui entraîne les arbres en parallèle, le Gradient Boosting construit les arbres un par un.
  1. *Entrée :* Données $X$ et cible $y$.
  2. *Calcul/Apprentissage :* Initialisation avec une prédiction constante. À chaque étape, calcul des résidus (erreurs), puis entraînement d'un nouvel arbre pour prédire ces résidus.
  3. *Sortie :* Somme pondérée de tous les arbres entraînés.
* **Formule mathématique :**
  $$\hat{y}^{(m)} = \hat{y}^{(m-1)} + \nu \cdot h_m(x)$$
  - **Définition des symboles :** $\hat{y}^{(m)}$ est la prédiction à l'étape $m$, $h_m(x)$ est l'arbre entraîné sur les résidus, $\nu$ est le taux d'apprentissage (*learning rate*).
  - **Paramètres (appris) :** Les structures des arbres séquentiels et leurs poids.
  - **Hyperparamètres :** `max_iter` (nombre d'arbres), `learning_rate`, `max_depth`, `l2_regularization`.
* **Avantages :**
  1. Souvent l'un des algorithmes les plus performants et précis sur données tabulaires.
  2. Gestion native efficace des grandes masses de données (`HistGradientBoosting`).
* **Limites :**
  1. Très sensible au surapprentissage si le nombre d'itérations est trop grand ou le taux d'apprentissage mal réglé.
  2. Entraînement séquentiel (plus difficile à paralléliser qu'une forêt aléatoire).


### 2. Schéma explicatif

````python
from sklearn.ensemble import HistGradientBoostingRegressor

train_errors = []
test_errors = []
iters = range(1, 51, 5)

for it in iters:
    gb = HistGradientBoostingRegressor(
        max_iter=it, random_state=42, learning_rate=0.1
    )
    gb.fit(Xr_ent, yr_ent)
    train_errors.append(
        mean_absolute_error(yr_ent, gb.predict(Xr_ent))
    )
    test_errors.append(mean_absolute_error(yr_test, gb.predict(Xr_test)))

plt.figure(figsize=(8, 4.5))
plt.plot(iters, train_errors, label='Erreur Entraînement (MAE)', color='blue')
plt.plot(iters, test_errors, label='Erreur Test (MAE)', color='orange')
plt.xlabel('Nombre d\'itérations (Arbres ajoutés)')
plt.ylabel('Erreur MAE')
plt.title(
    'Schéma explicatif : Convergence progressive du Gradient Boosting'
)
plt.legend()
plt.grid(True, ls='--')
plt.tight_layout()
plt.show()
````

**Commentaire du schéma :** Le schéma montre l'amélioration progressive : à chaque étape de boosting, l'ajout d'un nouvel arbre réduit le résidu global d'entraînement.


### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Actuariat) :** Tarification des risques d'assurance et modélisation précise du coût moyen des sinistres.
* **Cas d'usage 2 (Énergie / Logistique) :** Prévision ultra-précise de la demande électrique à court terme selon la météo et le jour de la semaine.
* **Nature des données & Prérequis :** Données tabulaires (numériques/catégorielles). `HistGradientBoosting` gère nativement les valeurs manquantes.
* **Cas peu approprié :** Données non structurées (images, audio, texte brut).
* **Métriques adaptées :** MAE, RMSE, $R^2$.


### 4. Analyse de l'expérience Python
* **Entrées du code :** `HistGradientBoostingRegressor(max_iter=80, random_state=42)` entraîné sur `Xr_ent` et `yr_ent`.
* **Hyperparamètres définis :** `max_iter=80`, `random_state=GRAINE`.
* **Sortie obtenue :** Affiche la MAE sur l'échantillon de test `Xr_test`.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Démontre la vitesse et le score du boosting basé sur des histogrammes.
  - *Ne permet PAS de conclure :* Ne donne pas accès à la courbe de perte au fil des 80 itérations (qui permettrait de repérer le début du surapprentissage).

## 8. Régression logistique

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
from sklearn.linear_model import LogisticRegression
modele=make_pipeline(StandardScaler(),LogisticRegression(max_iter=1000,random_state=GRAINE)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
````

### 1. Présentation
* **Nom (FR / EN) :** Régression logistique / *Logistic Regression*.
* **Famille :** Apprentissage supervisé — Classification (binaire et multiclasse).
* **Objectif :** Estimer la probabilité qu'une observation appartienne à une classe spécifique à l'aide d'une fonction sigmoïde.
* **Intuition & Fonctionnement :**
  L'algorithme combine linéairement les variables explicatives puis transforme ce résultat à l'aide de la fonction logistique pour obtenir une valeur entre 0 et 1.
  1. *Entrée :* Matrice de caractéristiques $X$ et étiquettes catégorielles $y \in \{0, 1\}$.
  2. *Calcul/Apprentissage :* Minimisation de la perte d'entropie croisée (*log-loss*) par descente de gradient.
  3. *Sortie :* Probabilité d'appartenance à la classe positive $\hat{p}$, convertie en classe 0 ou 1 selon un seuil (ex: 0.5).
* **Formule mathématique :**
  $$\hat{p} = \sigma(z) = \frac{1}{1 + e^{-(X\beta + \beta_0)}}$$
  Perte entropie croisée binaire :
  $$\mathcal{L}(\beta) = -\frac{1}{n} \sum_{i=1}^{n} \left[ y_i \log(\hat{p}_i) + (1 - y_i) \log(1 - \hat{p}_i) \right]$$
  - **Définition des symboles :** $\sigma(z)$ est la fonction sigmoïde, $y_i$ la vraie classe, $\hat{p}_i$ la probabilité prédite, $\beta$ les coefficients.
  - **Paramètres (appris) :** Coefficients $\beta$ et intercept $\beta_0$.
  - **Hyperparamètres :** `C` (inverse de la pénalité), `penalty` (l1, l2, elasticnet), `solver`.
* **Avantages :**
  1. Modèle très rapide et facilement interprétable via les ratios de chances (*odds ratios*).
  2. Fournit des probabilités bien étalonnées en plus des prédictions de classes.
* **Limites :**
  1. Suppose une frontière de décision linéaire entre les classes.
  2. Sensible à la multicolinéarité et aux valeurs extrêmes.

### 2. Schéma explicatif

````python
import matplotlib.pyplot as plt
import numpy as np

z = np.linspace(-6, 6, 100)
sigmoide = 1 / (1 + np.exp(-z))

plt.figure(figsize=(8, 4.5))
plt.plot(z, sigmoide, color='blue', linewidth=2.5, label='Sigmoïde σ(z)')
plt.axhline(0.5, color='red', linestyle='--', label='Seuil de décision (0.5)')
plt.axvline(0, color='gray', linestyle=':', alpha=0.7)
plt.title('Schéma explicatif : Fonction Sigmoïde pour la Régression Logistique')
plt.xlabel('Combinaison linéaire (z = Xβ + β0)')
plt.ylabel('Probabilité P(Y=1)')
plt.legend()
plt.grid(True, ls='--')
plt.tight_layout()
plt.show()
````

**Commentaire du schéma :** Le graphique présente la courbe en S de la fonction sigmoïde qui compresse n'importe quelle valeur réelle dans l'intervalle [0, 1] avec le seuil de décision fixé à 0.5.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Credit Scoring) :** Évaluation du risque d'accord de prêt (prédire si un emprunteur sera en défaut de paiement ou non).
* **Cas d'usage 2 (Marketing) :** Prédiction de l'attrition client (*churn*) dans les services d'abonnement.
* **Nature des données & Prérequis :** Variables numériques (mise à l'échelle vivement recommandée via `StandardScaler`).
* **Cas peu approprié :** Données avec des frontières de séparation hautement complexes et non linéaires (ex: reconnaissance d'images).
* **Métriques adaptées :** F1-Score, AUC-ROC, Précision, Rappel, Matrice de confusion.

### 4. Analyse de l'expérience Python
* **Entrées du code :** `Xc_ent` et `yc_ent` dans un pipeline avec `StandardScaler()` et `LogisticRegression(max_iter=1000, random_state=42)`.
* **Hyperparamètres définis :** `max_iter=1000`, `random_state=GRAINE`.
* **Sortie obtenue :** Affiche le F1-score sur l'échantillon de test.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Donne la mesure du compromis précision/rappel sur le jeu de test binaire synthétique.
  - *Ne permet PAS de conclure :* Ne montre pas si ajuster le seuil de décision (actuellement 0.5) optimiserait le rappel ou la précision selon les contraintes métier.

## 9. KNN

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
from sklearn.neighbors import KNeighborsClassifier
modele=make_pipeline(StandardScaler(),KNeighborsClassifier(n_neighbors=5)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
````

### 1. Présentation
* **Nom (FR / EN) :** K plus proches voisins / *K-Nearest Neighbors (KNN)*.
* **Famille :** Apprentissage supervisé — Classification / Méthode basée sur les instances (*Lazy Learning*).
* **Objectif :** Classer une nouvelle observation en se basant sur la classe majoritaire de ses $k$ voisins les plus proches dans l'espace des variables.
* **Intuition & Fonctionnement :**
  L'algorithme ne construit pas de modèle explicite pendant l'entraînement ; il stocke simplement les données.
  1. *Entrée :* Nouvelle donnée $x$ et ensemble d'entraînement stocké.
  2. *Calcul/Apprentissage :* Calcul de la distance (ex: euclidienne) entre $x$ et toutes les données stockées pour identifier les $k$ voisins les plus proches.
  3. *Sortie :* Vote majoritaire parmi les $k$ voisins pour déterminer la classe attribuée.
* **Formule mathématique :**
  Distance euclidienne entre deux points $p$ et $q$ :
  $$d(p, q) = \sqrt{\sum_{j=1}^{d} (p_j - q_j)^2}$$
  - **Définition des symboles :** $d$ est le nombre de dimensions (caractéristiques), $p_j$ et $q_j$ les coordonnées respectives des points.
  - **Paramètres (appris) :** Aucun (l'algorithme stocke simplement l'ensemble des données d'entraînement).
  - **Hyperparamètres :** `n_neighbors` ($k$), `metric` (euclidean, manhattan, etc.), `weights` (uniform, distance).
* **Avantages :**
  1. Très simple à comprendre, sans hypothèse sur la distribution des données.
  2. S'adapte naturellement aux problèmes multiclasses et non linéaires.
* **Limites :**
  1. Prédictions très lentes sur de grands jeux de données (doit parcourir toutes les données à chaque prédiction).
  2. Extrêmement sensible au "fléau de la dimension" et à l'absence d'échelle uniforme entre variables.

### 2. Schéma explicatif

````python
from sklearn.neighbors import KNeighborsClassifier

np.random.seed(42)
X_knn = np.random.randn(30, 2)
y_knn = np.array([0 if x[0] + x[1] < 0 else 1 for x in X_knn])

plt.figure(figsize=(7, 4.5))
plt.scatter(X_knn[y_knn==0, 0], X_knn[y_knn==0, 1], color='blue', label='Classe 0')
plt.scatter(X_knn[y_knn==1, 0], X_knn[y_knn==1, 1], color='orange', label='Classe 1')
target = np.array([[0, 0]])
plt.scatter(target[0, 0], target[0, 1], color='red', marker='*', s=200, label='Nouveau Point')

circle = plt.Circle((0, 0), 0.8, color='gray', fill=False, linestyle='--', linewidth=2, label='Voisinage k=5')
plt.gca().add_patch(circle)
plt.title('Schéma explicatif : Vote par plus proches voisins (KNN)')
plt.legend()
plt.grid(True, ls='--')
plt.tight_layout()
plt.show()
````

**Commentaire du schéma :** Le schéma illustre comment le choix de $k$ modifie la décision : pour $k=3$, la classe A l'emporte, tandis que pour un $k$ plus grand, la classe B peut devenir majoritaire autour du point cible.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Risk Management) :** Détection de profils de clients à haut risque par comparaison directe avec les profils historiques similaires.
* **Cas d'usage 2 (Recommandation) :** Systèmes de recommandation basés sur le filtrage collaboratif (trouver les $k$ utilisateurs ayant des goûts similaires).
* **Nature des données & Prérequis :** Variables numériques impérativement standardisées (`StandardScaler`).
* **Cas peu approprié :** Jeux de données volumineux avec des centaines de colonnes (la distance perd de son sens en haute dimension).
* **Métriques adaptées :** F1-Score, Précision, Rappel, Accuracy.

### 4. Analyse de l'expérience Python
* **Entrées du code :** `Xc_ent` et `yc_ent` dans un pipeline avec `StandardScaler()` et `KNeighborsClassifier(n_neighbors=5)`.
* **Hyperparamètres définis :** `n_neighbors=5`.
* **Sortie obtenue :** Affiche le F1-score sur l'échantillon de test.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Mesure la capacité de classification de 5 voisins sur les données standardisées.
  - *Ne permet PAS de conclure :* Ne montre pas si $k=5$ est la valeur optimale ou si $k=3$ ou $k=7$ donnerait de meilleurs résultats.

## 10. Naive Bayes

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
from sklearn.naive_bayes import GaussianNB
modele=GaussianNB().fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
````

### 1. Présentation
* **Nom (FR / EN) :** Naïve Bayes / *Naive Bayes (Gaussian Naive Bayes)*.
* **Famille :** Apprentissage supervisé — Classification probabiliste.
* **Objectif :** Calcule la probabilité a posteriori d'appartenir à chaque classe en utilisant le théorème de Bayes sous l'hypothèse de l'indépendance conditionnelle des caractéristiques.
* **Intuition & Fonctionnement :**
  L'algorithme combine la probabilité a priori des classes avec les probabilités conditionnelles de chaque variable (supposée gaussienne pour des données continues).
  1. *Entrée :* Données $X$ et étiquettes $y$.
  2. *Calcul/Apprentissage :* Estimation de la moyenne et de la variance de chaque caractéristique pour chaque classe.
  3. *Sortie :* Classe qui maximise la probabilité a posteriori (règle du Maximum A Posteriori - MAP).
* **Formule mathématique :**
  Théorème de Bayes appliqué avec l'hypothèse "naïve" d'indépendance :
  $$P(Y = c \mid X_1, \dots, X_d) \propto P(Y = c) \prod_{j=1}^{d} P(X_j \mid Y = c)$$
  - **Définition des symboles :** $P(Y=c)$ est la probabilité a priori de la classe $c$, $P(X_j \mid Y=c)$ est la densité de probabilité gaussienne de la variable $j$ conditionnellement à la classe $c$.
  - **Paramètres (appris) :** Moyennes $\mu_{c,j}$ et variances $\sigma_{c,j}^2$ par classe et par variable.
  - **Hyperparamètres :** `var_smoothing` (stabilité numérique).
* **Avantages :**
  1. Extrêmement rapide à entraîner et peu gourmand en puissance de calcul.
  2. Fonctionne de façon remarquable sur la classification de textes/emails malgré l'hypothèse simplificatrice d'indépendance.
* **Limites :**
  1. Supposer l'indépendance totale des variables est rarement vrai en pratique.
  2. Sensible au problème de "fréquence nulle" (géré par le lissage de Laplace).

### 2. Schéma explicatif

````python
import scipy.stats as stats

x_nb = np.linspace(-4, 6, 100)
dist_class0 = stats.norm.pdf(x_nb, loc=0, scale=1)
dist_class1 = stats.norm.pdf(x_nb, loc=2, scale=1)

plt.figure(figsize=(8, 4.5))
plt.plot(x_nb, dist_class0, color='blue', label='P(X | Classe 0) Gaussienne')
plt.plot(x_nb, dist_class1, color='orange', label='P(X | Classe 1) Gaussienne')
plt.axvline(1, color='red', linestyle='--', label='Frontière MAP (Précision)')
plt.title('Schéma explicatif : Distributions conditionnelles (Gaussian Naive Bayes)')
plt.xlabel('Variable X')
plt.ylabel('Densité de probabilité')
plt.legend()
plt.grid(True, ls='--')
plt.tight_layout()
plt.show()
````

**Commentaire du schéma :** Le schéma présente les distributions gaussiennes de chaque classe pour une variable et la frontière théorique où les probabilités a posteriori s'équilibrent.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Marketing / Communication) :** Filtrage des spams et classification automatique des emails entrants des clients.
* **Cas d'usage 2 (Finance / Sentiment Analysis) :** Analyse du sentiment (positif/négatif/neutre) dans les rapports financiers ou actualités boursières.
* **Nature des données & Prérequis :** Variables continues (GaussianNB) ou comptages/fréquences textuelles (MultinomialNB).
* **Cas peu approprié :** Données comportant de très fortes corrélations dépendantes entre variables explicatives.
* **Métriques adaptées :** F1-Score, Précision, Rappel, ROC-AUC.

### 4. Analyse de l'expérience Python
* **Entrées du code :** `GaussianNB()` entraîné sur `Xc_ent` et `yc_ent`.
* **Hyperparamètres définis :** Aucun (configuration gaussienne standard).
* **Sortie obtenue :** Affiche le F1-score sur `Xc_test`.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Évalue l'efficacité du classifieur bayésien gaussien rapide sur le jeu de classification.
  - *Ne permet PAS de conclure :* Ne valide pas si l'hypothèse de normalité gaussienne des variables d'entrée est réellement respectée sur ces données.

## 11. Arbre de décision

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
from sklearn.tree import DecisionTreeClassifier
modele=DecisionTreeClassifier(max_depth=4,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
````

### 1. Présentation
* **Nom (FR / EN) :** Arbre de décision pour la classification / *Decision Tree Classifier*.
* **Famille :** Apprentissage supervisé — Arbres de décision.
* **Objectif :** Séparer les données en sous-groupes de plus en plus homogènes (*purs*) par une série de règles SI-ALORS sur les caractéristiques.
* **Intuition & Fonctionnement :**
  L'algorithme choisit à chaque étape la variable et le seuil qui maximisent le gain d'information ou réduisent au mieux l'impureté de Gini.
  1. *Entrée :* Données $X$ et étiquettes catégorielles $y$.
  2. *Calcul/Apprentissage :* Évaluation de l'impureté de Gini ou de l'entropie à chaque nœud pour déterminer la séparation optimale.
  3. *Sortie :* Classe majoritaire présente dans la feuille d'arrivée.
* **Formule mathématique :**
  Indice d'impureté de Gini :
  $$Gini(N) = 1 - \sum_{i=1}^{K} p_i^2$$
  - **Définition des symboles :** $K$ est le nombre de classes, $p_i$ est la proportion d'éléments appartenant à la classe $i$ dans le nœud $N$.
  - **Paramètres (appris) :** Variables sélectionnées à chaque nœud, seuils de séparation et affectation des classes aux feuilles.
  - **Hyperparamètres :** `max_depth`, `min_samples_split`, `criterion` (gini, entropy).
* **Avantages :**
  1. Entièrement explicable et visualisable sous forme de diagramme décisionnel simple.
  2. Gère nativement les variables numériques et catégorielles sans mise à l'échelle.
* **Limites :**
  1. Très enclin au surapprentissage si sa profondeur n'est pas limitée.
  2. Instable : une petite modification des données d'entraînement peut changer radicalement la structure de l'arbre.

### 2. Schéma explicatif

````python
from sklearn.tree import DecisionTreeClassifier, plot_tree

dt_demo = DecisionTreeClassifier(max_depth=2, random_state=42)
dt_demo.fit(Xc_ent[:, :2], yc_ent)

plt.figure(figsize=(8, 4.5))
plot_tree(dt_demo, feature_names=['X1', 'X2'], class_names=['0', '1'], filled=True)
plt.title('Schéma explicatif : Arbre de décision pour la classification (Profondeur=2)')
plt.tight_layout()
plt.show()
````

**Commentaire du schéma :** Le schéma représente la séparation de l'espace des données par des droites orthogonales correspondant aux règles SI-ALORS générées par les nœuds de l'arbre.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Octroi de crédit) :** Arbre d'approbation automatique des prêts bancaires avec règles explicites destinées aux régulateurs.
* **Cas d'usage 2 (Gestion RH) :** Modélisation du risque de démission d'un employé selon l'ancienneté et le salaire.
* **Nature des données & Prérequis :** Variables numériques ou catégorielles. Aucune normalisation préalable nécessaire.
* **Cas peu approprié :** Relations fortement diagonales ou linéaires continues nécessitant une grande quantité de marches d'escalier pour être approchées.
* **Métriques adaptées :** F1-Score, Précision, Rappel, Matrice de confusion.

### 4. Analyse de l'expérience Python
* **Entrées du code :** `DecisionTreeClassifier(max_depth=4, random_state=42)` ajusté sur `Xc_ent` et `yc_ent`.
* **Hyperparamètres définis :** `max_depth=4`, `random_state=GRAINE`.
* **Sortie obtenue :** Affiche le F1-score sur l'échantillon de test `Xc_test`.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Indique la performance de classification pour une profondeur maximale limitée à 4.
  - *Ne permet PAS de conclure :* Ne montre pas l'arbre d'origine (`plot_tree`) qui permet de comprendre visuellement les règles de décision.



## 12. Forêt aléatoire — classification

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
from sklearn.ensemble import RandomForestClassifier
modele=RandomForestClassifier(n_estimators=80,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
````

### 1. Présentation
* **Nom (FR / EN) :** Forêt aléatoire pour la classification / *Random Forest Classifier*.
* **Famille :** Apprentissage supervisé — Méthodes d'ensemble (*Bagging*).
* **Objectif :** Combiner la prédiction d'un grand nombre d'arbres de décision indépendants pour améliorer la généralisation et réduire la variance.
* **Intuition & Fonctionnement :**
  L'algorithme introduit une double aléatorisation (tirage des observations avec remise + sous-sélection aléatoire de variables à chaque nœud).
  1. *Entrée :* Données $X$ et étiquettes $y$.
  2. *Calcul/Apprentissage :* Entraînement en parallèle de $B$ arbres de décision sur des échantillons bootstrap distincts.
  3. *Sortie :* Vote majoritaire (ou moyenne des probabilités) des arbres de la forêt.
* **Formule mathématique :**
  Vote majoritaire :
  $$\hat{y}_{RF} = \text{mode} \left( \{ T_1(x), T_2(x), \dots, T_B(x) \} \right)$$
  - **Définition des symboles :** $B$ est le nombre d'arbres (`n_estimators`), $T_b(x)$ est la prédiction du $b$-ième arbre de classification.
  - **Paramètres (appris) :** L'ensemble des structures de tous les arbres individuels de la forêt.
  - **Hyperparamètres :** `n_estimators`, `max_depth`, `max_features`, `criterion`.
* **Avantages :**
  1. Excellente performance globale sans nécessiter un réglage poussé des hyperparamètres.
  2. Évite le surapprentissage grâce au principe d'agrégation d'arbres diversifiés.
* **Limites :**
  1. Perte d'interprétabilité directe par rapport à un arbre individuel.
  2. Consommation en mémoire et temps de calcul plus importants sur les très grands jeux de données.

### 2. Schéma explicatif

````python
from sklearn.ensemble import RandomForestClassifier

rf_clf = RandomForestClassifier(n_estimators=80, random_state=42).fit(Xc_ent, yc_ent)
importances = rf_clf.feature_importances_

plt.figure(figsize=(8, 4.5))
plt.bar(range(len(importances)), importances, color='teal')
plt.xlabel('Index des caractéristiques (Variables)')
plt.ylabel('Importance relative (Gini)')
plt.title('Schéma explicatif : Importance des caractéristiques dans la Forêt Aléatoire')
plt.grid(True, ls='--', alpha=0.5)
plt.tight_layout()
plt.show()
````

**Commentaire du schéma :** Le schéma illustre l'obtention du résultat final grâce au vote majoritaire compilant les réponses issues des différents arbres indépendants.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Détection de fraude) :** Détection de transactions frauduleuses par carte de crédit sur des données historiques complexes.
* **Cas d'usage 2 (Marketing / Upselling) :** Ciblage des clients ayant la plus forte propension à acheter un produit complémentaire.
* **Nature des données & Prérequis :** Données numériques ou catégorielles encodées. Aucune mise à l'échelle requise.
* **Cas peu approprié :** Systèmes embarrassés par une forte contrainte de temps de latence en temps réel extrême (où un modèle plus léger est exigé).
* **Métriques adaptées :** F1-Score, AUC-ROC, Précision, Rappel.

### 4. Analyse de l'expérience Python
* **Entrées du code :** `RandomForestClassifier(n_estimators=80, random_state=42)` entraîné sur `Xc_ent` et `yc_ent`.
* **Hyperparamètres définis :** `n_estimators=80`, `random_state=GRAINE`.
* **Sortie obtenue :** Affiche le F1-score sur l'échantillon de test `Xc_test`.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Évalue la performance moyenne de l'agrégation de 80 arbres de décision sur les données de test.
  - *Ne permet PAS de conclure :* Ne donne pas la mesure de l'importance des caractéristiques (`feature_importances_`) qui permet d'identifier les variables dominantes.

## 13. SVM

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
from sklearn.svm import SVC
modele=make_pipeline(StandardScaler(),SVC(kernel='rbf',C=1.0,random_state=GRAINE)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
````

### 1. Présentation
* **Nom (FR / EN) :** Machines à vecteurs de support / *Support Vector Machines (SVM / SVC)*.
* **Famille :** Apprentissage supervisé — Classification par séparateur à marge maximale.
* **Objectif :** Trouver l'hyperplan qui sépare au mieux les classes en maximisant la distance (marge) entre l'hyperplan et les points les plus proches (vecteurs de support).
* **Intuition & Fonctionnement :**
  L'algorithme s'appuie sur la marge maximale et utilise le "truc du noyau" (*kernel trick*) pour projeter les données dans un espace de plus grande dimension où elles deviennent linéairement séparables.
  1. *Entrée :* Données $X$ et étiquettes $y \in \{-1, 1\}$.
  2. *Calcul/Apprentissage :* Résolution d'un problème d'optimisation quadratique sous contraintes pour trouver les vecteurs de support.
  3. *Sortie :* Classification d'un point selon le côté de l'hyperplan séparateur où il se situe.
* **Formule mathématique :**
  Problème d'optimisation (marge adoucie) :
  $$\min_{w, b, \xi} \frac{1}{2} \|w\|^2 + C \sum_{i=1}^{n} \xi_i \quad \text{sujet à} \quad y_i(w^T \phi(x_i) + b) \ge 1 - \xi_i$$
  - **Définition des symboles :** $w$ est le vecteur normal à l'hyperplan, $C$ contrôle le compromis marge/erreur, $\xi_i$ sont les variables d'ajustement (*slack variables*), $\phi(x)$ la projection par le noyau.
  - **Paramètres (appris) :** Poids $w$, biais $b$ et choix des vecteurs de support.
  - **Hyperparamètres :** `C`, `kernel` (rbf, linear, poly), `gamma`.
* **Avantages :**
  1. Trè efficace en haute dimension (même lorsqu'il y a plus de variables que d'observations).
  2. Robuste au surapprentissage grâce à l'optimisation axée uniquement sur les vecteurs de support.
* **Limites :**
  1. Temps de calcul quadratique à cubique avec le nombre d'observations (lent sur très grands datasets).
  2. Nécessite obligatoirement un réglage méticuleux des hyperparamètres (`C`, `gamma`) et la mise à l'échelle.

### 2. Schéma explicatif

````python
from sklearn.svm import SVC

svm_demo = SVC(kernel='linear', C=1.0).fit(Xc_ent[:, :2], yc_ent)

plt.figure(figsize=(8, 4.5))
plt.scatter(Xc_ent[:, 0], Xc_ent[:, 1], c=yc_ent, cmap='coolwarm', s=30, alpha=0.7)

# Tracer les vecteurs de support
plt.scatter(svm_demo.support_vectors_[:, 0], svm_demo.support_vectors_[:, 1],
            s=100, facecolors='none', edgecolors='k', label='Vecteurs de support')

plt.title('Schéma explicatif : Marge et Vecteurs de Support (SVM)')
plt.xlabel('X1')
plt.ylabel('X2')
plt.legend()
plt.grid(True, ls='--')
plt.tight_layout()
plt.show()
````

* **Commentaire du schéma :** Le schéma illustre l'hyperplan séparateur, la largeur de la marge maximale et la localisation des vecteurs de support qui définissent la frontière.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Trading) :** Classification de mouvements de marché à court terme basés sur de multiples indicateurs techniques complexes.
* **Cas d'usage 2 (Biologie / Santé) :** Diagnostic médical à partir de données d'imagerie ou d'analyses génomiques en haute dimension.
* **Nature des données & Prérequis :** Variables numériques obligatoirement standardisées (`StandardScaler`).
* **Cas peu approprié :** Très grands jeux de données dépassant plusieurs centaines de milliers d'observations.
* **Métriques adaptées :** F1-Score, Précision, Rappel, ROC-AUC.

### 4. Analyse de l'expérience Python
* **Entrées du code :** `Xc_ent` et `yc_ent` dans un pipeline avec `StandardScaler()` et `SVC(kernel='rbf', C=1.0, random_state=42)`.
* **Hyperparamètres définis :** `kernel='rbf'`, `C=1.0`, `random_state=GRAINE`.
* **Sortie obtenue :** Affiche le F1-score sur l'échantillon de test `Xc_test`.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Mesure la qualité de classification avec un noyau RBF gaussien à marge adoucie standard (`C=1.0`).
  - *Ne permet PAS de conclure :* Ne montre pas l'influence de la valeur de `gamma` ou d'un noyau alternatif sur le résultat.

## 14. Gradient Boosting — classification

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
from sklearn.ensemble import HistGradientBoostingClassifier
modele=HistGradientBoostingClassifier(max_iter=80,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
````

### 1. Présentation
* **Nom (FR / EN) :** Gradient Boosting pour la classification / *Gradient Boosting Classifier (HistGradientBoostingClassifier)*.
* **Famille :** Apprentissage supervisé — Méthodes d'ensemble séquentiel (*Boosting*).
* **Objectif :** Construire séquentiellement une suite d'arbres de décision simples où chaque nouvel arbre cherche à corriger les erreurs de l'ensemble précédent.
* **Intuition & Fonctionnement :**
  L'algorithme optimise une fonction de perte globale en appliquant la descente de gradient dans l'espace des fonctions (les arbres).
  1. *Entrée :* Données $X$ et étiquettes $y$.
  2. *Calcul/Apprentissage :* Calcul du gradient négatif de la fonction de perte logistique à chaque étape, puis entraînement d'un arbre sur ce gradient.
  3. *Sortie :* Probabilité finale obtenue par combinaison pondérée de tous les arbres.
* **Formule mathématique :**
  Mise à jour séquentielle du modèle :
  $$F_m(x) = F_{m-1}(x) + \gamma_m h_m(x)$$
  - **Définition des symboles :** $F_m(x)$ est le modèle à l'étape $m$, $h_m(x)$ l'arbre entraîné sur les pseudo-résidus (gradients), $\gamma_m$ le poids affecté à cet arbre.
  - **Paramètres (appris) :** Ensemble des arbres successifs et leurs coefficients d'apprentissage.
  - **Hyperparamètres :** `max_iter`, `learning_rate`, `max_depth`, `l2_regularization`.
* **Avantages :**
  1. Performances de prédiction exceptionnelles sur les données tabulaires.
  2. Gère efficacement les valeurs manquantes et les jeux de données volumineux via la discrétisation par histogrammes.
* **Limites :**
  1. Risque d'overfitting si le nombre d'itérations est trop élevé sans régularisation adéquate.
  2. Construction séquentielle plus longue à entraîner qu'un modèle basé sur le bagging parallèle.

### 2. Schéma explicatif

````python
from sklearn.ensemble import HistGradientBoostingClassifier

hgb_clf = HistGradientBoostingClassifier(max_iter=80, random_state=42).fit(Xc_ent, yc_ent)
y_pred_hgb = hgb_clf.predict(Xc_test)

plt.figure(figsize=(7, 4))
plt.hist(hgb_clf.predict_proba(Xc_test)[:, 1], bins=20, color='purple', alpha=0.7)
plt.title('Schéma explicatif : Distribution des probabilités prédites (Boosting)')
plt.xlabel('Probabilité prédite (Classe 1)')
plt.ylabel('Effectif')
plt.grid(True, ls='--')
plt.tight_layout()
plt.show()
````

## 15. AdaBoost

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
from sklearn.ensemble import AdaBoostClassifier
modele=AdaBoostClassifier(n_estimators=70,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
````

### 1. Présentation
* **Nom (FR / EN) :** AdaBoost / *Adaptive Boosting (AdaBoost Classifier)*.
* **Famille :** Apprentissage supervisé — Méthodes d'ensemble séquentiel (*Boosting*).
* **Objectif :** Combiner plusieurs classifieurs faibles (souvent des arbres très courts d'un seul niveau, appelés *stumps*) en augmentant le poids des observations mal classées à chaque étape.
* **Intuition & Fonctionnement :**
  Les modèles sont entraînés les uns après les autres. À chaque tour, les données mal classées par l'arbre précédent reçoivent un poids plus élevé pour contraindre le suivant à se focaliser sur elles.
  1. *Entrée :* Données $X$ et étiquettes $y \in \{-1, 1\}$, poids d'observations initialisés uniformément à $1/n$.
  2. *Calcul/Apprentissage :* Entraînement séquentiel de classifieurs faibles, calcul de leur erreur, et mise à jour exponentielle des poids des exemples.
  3. *Sortie :* Vote pondéré de l'ensemble des classifieurs faibles.
* **Formule mathématique :**
  Mise à jour des poids d'observations :
  $$w_i^{(m+1)} = w_i^{(m)} \exp \left( \alpha_m y_i h_m(x_i) \right)$$
  Poids du classifieur : $\alpha_m = \frac{1}{2} \ln \left( \frac{1 - \epsilon_m}{\epsilon_m} \right)$
  - **Définition des symboles :** $w_i^{(m)}$ est le poids de l'exemple $i$ à l'étape $m$, $\epsilon_m$ est le taux d'erreur du classifieur faible $h_m$, $\alpha_m$ est l'importance accordée à ce classifieur.
  - **Paramètres (appris) :** Poids $\alpha_m$ des classifieurs et règles des arbres courts.
  - **Hyperparamètres :** `n_estimators`, `learning_rate`.
* **Avantages :**
  1. Algorithme historiquement élégant et simple à mettre en œuvre.
  2. Peu enclin au surapprentissage sur des jeux de données propres sans trop de bruit.
* **Limites :**
  1. Extrêmement sensible au bruit et aux valeurs atypiques (*outliers*), qui reçoivent des poids démesurés.
  2. Moins performant que le Gradient Boosting sur les jeux de données complexes et bruités.

### 2. Schéma explicatif

````python
from sklearn.ensemble import AdaBoostClassifier

ada_demo = AdaBoostClassifier(n_estimators=5, random_state=42).fit(Xc_ent, yc_ent)

plt.figure(figsize=(7, 4))
plt.bar(range(1, 6), ada_demo.estimator_weights_, color='darkorange')
plt.xlabel('Numéro du classifieur faible (Stump)')
plt.ylabel('Poids attribué α_m')
plt.title('Schéma explicatif : Poids des classifieurs faibles dans AdaBoost')
plt.grid(True, ls='--')
plt.tight_layout()
plt.show()
````

**Commentaire du schéma :** Le schéma illustre l'augmentation de la taille (poids) des points mal classés qui contraint le classifieur suivant à ajuster sa frontière.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Vision par ordinateur) :** Détection de visages historiques (algorithme de Viola-Jones basé sur AdaBoost).
* **Cas d'usage 2 (Marketing / Churn) :** Identification rapide de clients insatisfaits sur des critères comportementaux basiques.
* **Nature des données & Prérequis :** Variables numériques ou catégorielles. Sensible au bruit dans les étiquettes.
* **Cas peu approprié :** Jeux de données fortement bruités comportant des erreurs d'étiquetage.
* **Métriques adaptées :** F1-Score, Précision, Rappel, Accuracy.

### 4. Analyse de l'expérience Python
* **Entrées du code :** `AdaBoostClassifier(n_estimators=70, random_state=42)` appliqué sur `Xc_ent` et `yc_ent`.
* **Hyperparamètres définis :** `n_estimators=70`, `random_state=GRAINE`.
* **Sortie obtenue :** Affiche le F1-score sur l'échantillon de test `Xc_test`.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Mesure la performance d'une combinaison de 70 classifieurs faibles réadaptés.
  - *Ne permet PAS de conclure :* Ne montre pas l'évolution des poids attribués aux observations au cours des itérations.

## 16. XGBoost

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
from xgboost import XGBClassifier
modele=XGBClassifier(n_estimators=80,max_depth=3,random_state=GRAINE,eval_metric='logloss').fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
````

### 1. Présentation
* **Nom (FR / EN) :** XGBoost / *Extreme Gradient Boosting (XGBClassifier)*.
* **Famille :** Apprentissage supervisé — Méthodes d'ensemble séquentiel (*Boosting*).
* **Objectif :** Fournir une implémentation ultra-optimisée, rapide et régularisée du Gradient Boosting pour maximiser les performances de prédiction.
* **Intuition & Fonctionnement :**
  XGBoost utilise un développement de Taylor de second ordre de la fonction de perte et intègre une régularisation ($L_1$ et $L_2$) de la structure des arbres pour contrôler la complexité.
  1. *Entrée :* Données $X$ et étiquettes $y$.
  2. *Calcul/Apprentissage :* Minimisation de la fonction d'objectif avec calcul des dérivées premières (gradients) et secondes (hessiens).
  3. *Sortie :* Prédiction finale issue de la somme pondérée de tous les arbres de régression.
* **Formule mathématique :**
  Fonction d'objectif à l'étape $t$ :
  $$\mathcal{L}^{(t)} \approx \sum_{i=1}^{n} \left[ g_i f_t(x_i) + \frac{1}{2} h_i f_t^2(x_i) \right] + \Omega(f_t)$$
  Terme de régularisation : $\Omega(f) = \gamma T + \frac{1}{2} \lambda \sum_{j=1}^{T} w_j^2$
  - **Définition des symboles :** $g_i$ et $h_i$ sont les gradients du premier et second ordre, $T$ est le nombre de feuilles, $w_j$ les poids des feuilles, $\gamma$ et $\lambda$ les coefficients de régularisation.
  - **Paramètres (appris) :** Structure des arbres et valeurs des feuilles.
  - **Hyperparamètres :** `n_estimators`, `max_depth`, `learning_rate`, `subsample`, `colsample_bytree`, `reg_lambda`, `reg_alpha`.
* **Avantages :**
  1. Vitesse d'exécution exceptionnelle grâce à la parallélisation du calcul des séparations.
  2. Très haute précision, ayant dominé pendant des années les compétitions de science des données (Kaggle).
* **Limites :**
  1. Grand nombre d'hyperparamètres complexes à régler.
  2. Modèle boîte noire plus difficile à interpréter et à déployer sans outils d'explicabilité (SHAP).

### 2. Schéma explicatif

````python
from xgboost import XGBClassifier

xgb_demo = XGBClassifier(n_estimators=80, max_depth=3, random_state=42, eval_metric='logloss').fit(Xc_ent, yc_ent)

plt.figure(figsize=(8, 4.5))
plt.plot(xgb_demo.evals_result_['validation_0']['logloss'] if hasattr(xgb_demo, 'evals_result_') else [0], color='crimson')
plt.title('Schéma explicatif : Optimisation par XGBoost (Dérivées de Taylor)')
plt.xlabel('Itérations')
plt.ylabel('Perte (Log-Loss)')
plt.grid(True, ls='--')
plt.tight_layout()
plt.show()
````

**Commentaire du schéma :** Le schéma présente la régularisation de la structure des arbres combinée aux gradients du premier et second ordre qui optimisent la convergence.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Évaluation du risque) :** Systèmes de notation de crédit et détection de fraudes complexes en temps réel.
* **Cas d'usage 2 (E-commerce / Conversion) :** Prédiction du taux de clics (CTR) et personnalisation des recommandations en ligne.
* **Nature des données & Prérequis :** Variables numériques (gère nativement les valeurs manquantes et l'encodage).
* **Cas peu approprié :** Petits jeux de données simples où des modèles plus légers suffisent largement.
* **Métriques adaptées :** F1-Score, AUC-ROC, Log-Loss, PR-AUC.

### 4. Analyse de l'expérience Python
* **Entrées du code :** `XGBClassifier(n_estimators=80, max_depth=3, random_state=42, eval_metric='logloss')` ajusté sur `Xc_ent` et `yc_ent`.
* **Hyperparamètres définis :** `n_estimators=80`, `max_depth=3`, `eval_metric='logloss'`.
* **Sortie obtenue :** Affiche le F1-score calculé sur le jeu de test `Xc_test`.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Évalue les performances brutes du classifieur XGBoost régularisé sur le jeu de test.
  - *Ne permet PAS de conclure :* Ne montre pas l'influence du paramètre `subsample` ou des régularisations `lambda`/`alpha` sur le score final.

## 17. K-means

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
from sklearn.cluster import KMeans
modele=KMeans(n_clusters=3,n_init=10,random_state=GRAINE).fit(X_clu)
print('Silhouette:',silhouette_score(X_clu,modele.labels_))
plt.scatter(X_clu[:,0],X_clu[:,1],c=modele.labels_,s=12);plt.title('K-means');plt.show()
````

### 1. Présentation
* **Nom (FR / EN) :** K-moyennes / *K-Means Clustering*.
* **Famille :** Apprentissage non supervisé — Regroupement (*Clustering* géométrique par partitions).
* **Objectif :** Partitionner un ensemble d'observations en $K$ groupes (clusters) distincts en minimisant la variance au sein de chaque groupe (inertie intra-classe).
* **Intuition & Fonctionnement :**
  L'algorithme positionne aléatoirement $K$ centroïdes, affecte chaque observation au centroïde le plus proche, puis réexécute l'actualisation des centroïdes en calculant la moyenne des points affectés jusqu'à convergence.
  1. *Entrée :* Données $X$ sans étiquettes et nombre préfixé de clusters $K$.
  2. *Calcul/Apprentissage :* Iteration alternée (Assignation des points aux centroïdes ➔ Mise à jour des positions des centroïdes).
  3. *Sortie :* Étiquette du cluster affecté à chaque observation et coordonnées finales des $K$ centroïdes.
* **Formule mathématique :**
  Minimisation de l'inertie intra-cluster ($WCSS$) :
  $$\min_{S} \sum_{k=1}^{K} \sum_{x \in S_k} \|x - \mu_k\|^2$$
  - **Définition des symboles :** $S_k$ représente l'ensemble des points attribués au cluster $k$, $\mu_k$ est le centroïde (moyenne) du cluster $k$.
  - **Paramètres (appris) :** Emplacements des $K$ centroïdes $\mu_1, \dots, \mu_K$.
  - **Hyperparamètres :** `n_clusters` ($K$), `init` (kmeans++, random), `n_init`, `max_iter`.
* **Avantages :**
  1. Algorithme très rapide, scalable et simple à implémenter sur de grandes bases de données.
  2. Convergence garantie en un nombre fini d'itérations.
* **Limites :**
  1. Nécessite de fixer le nombre de clusters $K$ à priori.
  2. Sensible aux initialisations, aux valeurs extrêmes et suppose des clusters de formes sphériques de tailles comparables.

### 2. Schéma explicatif

````python
from sklearn.cluster import KMeans

kmeans_demo = KMeans(n_clusters=3, n_init=10, random_state=42).fit(X_clu)

plt.figure(figsize=(7, 4.5))
plt.scatter(X_clu[:, 0], X_clu[:, 1], c=kmeans_demo.labels_, cmap='viridis', s=20, alpha=0.6)
plt.scatter(kmeans_demo.cluster_centers_[:, 0], kmeans_demo.cluster_centers_[:, 1],
            c='red', marker='X', s=150, label='Centroïdes (μ_k)')

plt.title('Schéma explicatif : Partitions et Centroïdes de K-means')
plt.xlabel('X1')
plt.ylabel('X2')
plt.legend()
plt.grid(True, ls='--')
plt.tight_layout()
plt.show()
````

**Commentaire du schéma :** Le schéma présente les $K$ centroïdes calculés et les frontières de Voronoi séparant les différentes régions d'attraction des clusters.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Marketing / Segmentation) :** Segmentation de la base clients en groupes homogènes d'achats pour cibler des offres marketing spécifiques.
* **Cas d'usage 2 (Finance / Portefeuilles) :** Regroupement d'actifs financiers aux comportements de rendements similaires pour la diversification.
* **Nature des données & Prérequis :** Variables numériques impérativement standardisées (`StandardScaler`).
* **Cas peu approprié :** Données comportant des clusters de formes arbitraires non sphériques (ex: croissants de lune) ou de densités très différentes.
* **Métriques adaptées :** Coefficient de Silhouette, Score Davies-Bouldin, Inertie (méthode du coude / *Elbow method*).

### 4. Analyse de l'expérience Python
* **Entrées du code :** `KMeans(n_clusters=3, n_init=10, random_state=42)` appliqué sur les données synthétiques `X_clu`.
* **Hyperparamètres définis :** `n_clusters=3`, `n_init=10`, `random_state=GRAINE`.
* **Sortie obtenue :** Affiche le score de Silhouette et affiche le graphique de dispersion colorié par clusters.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Mesure la séparation globale et l'homogénéité des 3 clusters trouvés sur ce nuage 2D.
  - *Ne permet PAS de conclure :* Ne garantit pas que $K=3$ soit la valeur optimale sur le plan métier sans analyser l'évolution du score de Silhouette pour d'autres valeurs de $K$.

## 18. Classification hiérarchique

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
from sklearn.cluster import AgglomerativeClustering
etiquettes=AgglomerativeClustering(n_clusters=3).fit_predict(X_clu)
print('Silhouette:',silhouette_score(X_clu,etiquettes))
````

### 1. Présentation
* **Nom (FR / EN) :** Classification hiérarchique / *Hierarchical Agglomerative Clustering (HAC)*.
* **Famille :** Apprentissage non supervisé — Regroupement hiérarchique (approche ascendante).
* **Objectif :** Construire une hiérarchie d'emboîtement des données sous forme d'arbre (dendrogramme) sans fixer le nombre de groupes au départ.
* **Intuition & Fonctionnement :**
  Chaque point commence dans son propre cluster. À chaque étape, l'algorithme fusionne les deux clusters les plus proches selon un critère de liaison (*linkage*) prédéfini jusqu'à ce que toutes les données forment un seul groupe.
  1. *Entrée :* Données $X$ et métrique de distance.
  2. *Calcul/Apprentissage :* Calcul séquentiel de la matrice des distances et fusion progressive des paires de clusters les plus proches.
  3. *Sortie :* Arbre hiérarchique (dendrogramme) découpé au niveau souhaité pour extraire $K$ clusters.
* **Formule mathématique :**
  Distance de Ward (minimisation de la variance cumulée) :
  $$\Delta(A, B) = \frac{n_A n_B}{n_A + n_B} \|\mu_A - \mu_B\|^2$$
  - **Définition des symboles :** $n_A, n_B$ représentent les effectifs des clusters $A$ et $B$, $\mu_A, \mu_B$ sont leurs centroïdes respectifs.
  - **Paramètres (appris) :** Matrice de liaisons hiérarchiques et regroupements.
  - **Hyperparamètres :** `n_clusters`, `linkage` (ward, complete, average, single), `metric`.
* **Avantages :**
  1. Fournit une visualisation hiérarchique très intuitive de la structure des données (dendrogramme).
  2. Ne nécessite pas d'imposer $K$ dès le départ (on peut couper l'arbre à n'importe quelle hauteur).
* **Limites :**
  1. Complexité temporelle et spatiale élevée en $\mathcal{O}(n^3)$ ou $\mathcal{O}(n^2)$, inutilisable sur de grands jeux de données.
  2. Une décision de fusion à une étape donnée ne peut jamais être annulée ultérieurement.

### 2. Schéma explicatif

````python
from scipy.cluster.hierarchy import dendrogram, linkage

linked = linkage(X_clu[:30], method='ward')

plt.figure(figsize=(8, 4.5))
dendrogram(linked, orientation='top', distance_sort='descending', show_leaf_counts=True)
plt.axhline(y=5, color='r', linestyle='--', label='Ligne de coupe (K=3)')
plt.title('Schéma explicatif : Dendrogramme de la Classification Hiérarchique')
plt.xlabel('Index des échantillons')
plt.ylabel('Distance de fusion (Ward)')
plt.legend()
plt.tight_layout()
plt.show()
````

* **Commentaire du schéma :** Le dendrogramme ci-dessous illustre les fusions ascendantes des observations et la ligne de coupe horizontale déterminant le nombre final de sous-groupes.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Gestion RH / Comportement) :** Établissement d'une typologie d'emplois ou de compétences en construisant des arbres de similarité fonctionnelle.
* **Cas d'usage 2 (Biologie / Logistique) :** Structuration de taxonomies de produits ou de zones géographiques de distribution.
* **Nature des données & Prérequis :** Variables numériques obligatoirement standardisées (`StandardScaler`).
* **Cas peu approprié :** Très volumineux jeux de données comprenant des dizaines de milliers de lignes.
* **Métriques adaptées :** Coefficient de Silhouette, Cofficient Cophenetique (fidélité de l'arbre).

### 4. Analyse de l'expérience Python
* **Entrées du code :** `AgglomerativeClustering(n_clusters=3)` appliqué directement sur `X_clu`.
* **Hyperparamètres définis :** `n_clusters=3` (critère de liaison par défaut : Ward).
* **Sortie obtenue :** Affiche la valeur de la Silhouette pour la partition obtenue.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Évalue la qualité des 3 sous-groupes fusionnés par le critère de Ward.
  - *Ne permet PAS de conclure :* Ne montre pas l'arbre d'assemblage hiérarchique (dendrogramme) qui permet d'identifier la hauteur de coupe optimale.

Voici le squelette de code Python complet et prêt à être exécuté dans Google Colab pour le Projet F3 : Prévision de l'inflation.

Ce code respecte la méthodologie financière (validation chronologique sans fuite temporelle) et télécharge les données directement depuis la réserve fédérale de Saint-Louis (FRED).

Python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import Ridge
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error

# Configuration du style graphique
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (12, 6)

# ==========================================
# 1. TÉLÉCHARGEMENT DES DONNÉES DEPUIS FRED
# ==========================================
print("Téléchargement des séries temporelles...")
url_cpi = "https://fred.stlouisfed.org/graph/fredgraph.csv?id=CPIAUCSL"      # Indice des prix à la consommation
url_unrate = "https://fred.stlouisfed.org/graph/fredgraph.csv?id=UNRATE"    # Taux de chômage
url_fedfunds = "https://fred.stlouisfed.org/graph/fredgraph.csv?id=FEDFUNDS"# Taux directeur

df_cpi = pd.read_csv(url_cpi, parse_dates=['DATE'], index_col='DATE')
df_unrate = pd.read_csv(url_unrate, parse_dates=['DATE'], index_col='DATE')
df_fedfunds = pd.read_csv(url_fedfunds, parse_dates=['DATE'], index_col='DATE')

# Fusion des données sur la date
df = df_cpi.join([df_unrate, df_fedfunds], how='inner')
df.columns = ['CPI', 'UNRATE', 'FEDFUNDS']

# Conversion en numérique (gestion des éventuelles valeurs 'NaN' ou '.')
df = df.apply(pd.to_numeric, errors='coerce').dropna()

# ==========================================
# 2. INGÉNIERIE DES VARIABLES (FEATURE ENG.)
# ==========================================
# Calcul de l'inflation annuelle : % de variation sur 12 mois (en %)
df['INFLATION'] = df['CPI'].pct_change(12) * 100

# Création des variables retardées (Lags) pour éviter le look-ahead bias
for lag in [1, 2, 3, 6, 12]:
    df[f'INFLATION_lag_{lag}'] = df['INFLATION'].shift(lag)
    df[f'UNRATE_lag_{lag}'] = df['UNRATE'].shift(lag)
    df[f'FEDFUNDS_lag_{lag}'] = df['FEDFUNDS'].shift(lag)

# Définition de la cible (Y) : Inflation dans 12 mois
df['TARGET_12M'] = df['INFLATION'].shift(-12)

# Suppression des lignes contenant des NaN suite aux décalages
df_clean = df.dropna().copy()

# ==========================================
# 3. DÉCOUPAGE CHRONOLOGIQUE (TRAIN / TEST)
# ==========================================
# Variables explicatives (X) et Cible (y)
features = [col for col in df_clean.columns if 'lag' in col]
X = df_clean[features]
y = df_clean['TARGET_12M']

# Découpage temporel : 80% Entraînement, 20% Test (Pas de train_test_split aléatoire !)
split_idx = int(len(df_clean) * 0.8)
X_train, X_test = X.iloc[:split_idx], X.iloc[split_idx:]
y_train, y_test = y.iloc[:split_idx], y.iloc[split_idx:]

print(f"Période d'entraînement : {X_train.index.min().strftime('%Y-%m')} à {X_train.index.max().strftime('%Y-%m')}")
print(f"Période de test         : {X_test.index.min().strftime('%Y-%m')} à {X_test.index.max().strftime('%Y-%m')}")

# ==========================================
# 4. MODÉLISATION ET ÉVALUATION
# ==========================================
# Référence (Naive Baseline) : La valeur actuelle de l'inflation
y_pred_baseline = X_test['INFLATION_lag_1']

# Modèle 1 : Régression Ridge
model_ridge = Ridge(alpha=1.0)
model_ridge.fit(X_train, y_train)
y_pred_ridge = model_ridge.predict(X_test)

# Modèle 2 : Gradient Boosting
model_gb = GradientBoostingRegressor(n_estimators=100, random_state=42)
model_gb.fit(X_train, y_train)
y_pred_gb = model_gb.predict(X_test)

# Calcul des métriques RMSE
rmse_base = np.sqrt(mean_squared_error(y_test, y_pred_baseline))
rmse_ridge = np.sqrt(mean_squared_error(y_test, y_pred_ridge))
rmse_gb = np.sqrt(mean_squared_error(y_test, y_pred_gb))

print("\n--- RÉSULTATS (RMSE) ---")
print(f"Référence (Naïve) : {rmse_base:.3f}")
print(f"Régression Ridge  : {rmse_ridge:.3f}")
print(f"Gradient Boosting : {rmse_gb:.3f}")

# ==========================================
# 5. VISUALISATION
# ==========================================
plt.figure(figsize=(12, 5))
plt.plot(y_test.index, y_test, label='Inflation Réelle (+12 mois)', color='black', linewidth=2)
plt.plot(y_test.index, y_pred_baseline, label='Référence Naïve', linestyle='--', color='gray', alpha=0.7)
plt.plot(y_test.index, y_pred_ridge, label='Prédiction Ridge', color='blue', alpha=0.8)
plt.plot(y_test.index, y_pred_gb, label='Prédiction Gradient Boosting', color='green', alpha=0.8)

plt.title("Prévision de l'Inflation US à un horizon de 12 mois (Jeu de Test)")
plt.xlabel("Date")
plt.ylabel("Taux d'inflation (%)")
plt.legend()
plt.tight_layout()
plt.show()

import numpy as np

# Schéma explicatif de la Régression Linéaire
np.random.seed(42)

    x_demo, y_demo, color="blue", label="Données réelles (Observations)"

    x_demo[15],

    linestyle="--",

)

plt.tight_layout()
plt.show()

1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Régression linéaire / *Linear Regression*.
* **Famille :** Apprentissage supervisé — Régression.
* **Objectif :** Prédire une variable cible continue $y$ à partir d'une combinaison linéaire de variables explicatives $X$.
* **Intuition & Fonctionnement :**
  L'algorithme cherche à ajuster la meilleure droite (ou hyperplan en multi-dimensions) passant à travers le nuage de points de données.
  1. *Entrée :* Matrice de données $X \in \mathbb{R}^{n \times p}$ et vecteur cible $y \in \mathbb{R}^n$.
  2. *Calcul/Apprentissage :* Minimisation de la somme des carrés des écarts (Moindres Carrés Ordinaires - MCO) entre la cible réelle et la valeur prédite.
  3. *Sortie :* Une fonction de prédiction $f(X) = \hat{y}$ qui attribue une valeur numérique continue à toute nouvelle observation.
* **Formule mathématique :**
  $$\hat{y} = \beta_0 + \sum_{j=1}^{p} \beta_j x_j = X\beta$$
  Fonction de perte (Moindres Carrés Ordinaires) :
  $$\mathcal{L}(\beta) = \frac{1}{2n} \|y - X\beta\|^2_2 = \frac{1}{2n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$
  - **Définition des symboles :** $y_i$ est la vraie valeur, $\hat{y}_i$ la valeur prédite, $x_j$ les variables explicatives, et $n$ le nombre d'observations.
  - **Paramètres (appris) :** Coefficients d'onde/pentes $\beta_j$ et biais/ordonnée à l'origine $\beta_0$.
  - **Hyperparamètres :** Aucun dans la version analytique standard des MCO (solution fermée $\beta = (X^T X)^{-1} X^T y$).
* **Avantages :**
  1. Très rapide à calculer et très sobre en ressources calculatoires.
  2. Modèle entièrement interprétable : chaque coefficient $\beta_j$ indique la variation exacte de $y$ pour une augmentation d'une unité de $x_j$.
* **Limites :**
  1. Hypothèse forte de linéarité entre les variables d'entrée et la cible.
  2. Extrêmement sensible aux valeurs extrêmes (*outliers*) et à la multicolinéarité entre les variables $X$.

---

2. Ridge
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Régression Ridge / *Ridge Regression (L2 Regularization)*.
* **Famille :** Apprentissage supervisé — Régression avec régularisation.
* **Objectif :** Prédire une variable continue $y$ en pénalisant les grands coefficients pour éviter le surapprentissage et traiter la multicolinéarité.
* **Intuition & Fonctionnement :**
  L'algorithme modifie les Moindres Carrés Ordinaires en ajoutant une pénalité sur la somme des carrés des coefficients (norme $L_2$).
  1. *Entrée :* Données $X$ (préalablement standardisées) et cible $y$.
  2. *Calcul/Apprentissage :* Minimisation de la perte MCO + pénalité $L_2$. Les coefficients sont contractés (*shrinkage*) vers zéro sans jamais devenir exactement nuls.
  3. *Sortie :* Vecteur de prédictions $\hat{y}$ plus stable et moins sensible au bruit des données d'entraînement.
* **Formule mathématique :**
  $$\mathcal{L}_{Ridge}(\beta) = \frac{1}{2n} \|y - X\beta\|^2_2 + \alpha \|\beta\|^2_2 = \frac{1}{2n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 + \alpha \sum_{j=1}^{p} \beta_j^2$$
  - **Définition des symboles :** $y_i$ est la valeur réelle, $\hat{y}_i$ la prédiction, $\beta_j$ les coefficients, $n$ le nombre d'échantillons, $\alpha$ le coefficient de pénalité.
  - **Paramètres (appris) :** Coefficients $\beta$.
  - **Hyperparamètres :** $\alpha \ge 0$ (intensité de la régularisation).
* **Avantages :**
  1. Très performant en présence de forte multicolinéarité entre variables.
  2. Réduit efficacement la variance du modèle sans augmenter excessivement le biais.
* **Limites :**
  1. Ne réalise pas de sélection de variables (aucun coefficient n'est annulé).
  2. Nécessite obligatoirement une standardisation préalable des variables.

---

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le graphique montre l'effet de contraction (*shrinkage*) : à mesure que l'hyperparamètre $\alpha$ augmente, la magnitude des coefficients $\beta$ diminue vers zéro, lissant la fonction de prédiction.

---

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Économétrie) :** Modélisation des rendements boursiers où de nombreux indicateurs macroéconomiques sont fortement corrélés entre eux.
* **Cas d'usage 2 :** Estimation de prix immobiliers avec des dizaines de caractéristiques corrélées (surface, nombre de pièces, année).
* **Nature des données & Prérequis :** Variables numériques obligatoirement standardisées (`StandardScaler`).
* **Cas peu approprié :** Jeux de données avec un très grand nombre de variables non pertinentes où une sélection de variables est nécessaire.
* **Métriques adaptées :** MAE, RMSE, $R^2$.

---

### 4. Analyse de l'expérience Python
* **Entrées du code :** `Xr_ent` et `yr_ent` passés dans un `Pipeline` combinant `StandardScaler()` et `Ridge(alpha=1.0)`.
* **Hyperparamètres définis :** `alpha=1.0`.
* **Sortie obtenue :** Affiche la MAE calculée sur `Xr_test`.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Valide l'intégration de la standardisation et du calcul du score sur les données de test synthétiques.
  - *Ne permet PAS de conclure :* Ne permet pas de savoir si `alpha=1.0` est la valeur optimale (nécessiterait une validation croisée `RidgeCV`).
2. Bloc Code Python pour le schéma (à copier dans une cellule Code)
Python
import matplotlib.pyplot as plt
import numpy as np

alphas = np.logspace(-2, 3, 50)
coefs = []

# Génération de données synthétiques corrélées
np.random.seed(42)
X_dummy = np.random.randn(50, 5)
y_dummy = X_dummy @ np.array([3, -2, 1.5, 0, 0.5]) + np.random.randn(50)

from sklearn.linear_model import Ridge
from sklearn.preprocessing import StandardScaler

X_scaled = StandardScaler().fit_transform(X_dummy)
for a in alphas:
    ridge = Ridge(alpha=a)
    ridge.fit(X_scaled, y_dummy)
    coefs.append(ridge.coef_)

plt.figure(figsize=(8, 4.5))
plt.plot(alphas, coefs)
plt.xscale('log')
plt.xlabel('Hyperparamètre alpha (Intensité de régularisation L2)')
plt.ylabel('Coefficients beta')
plt.title('Schéma explicatif : Contraction des coefficients par la régression Ridge')
plt.grid(True, which="both", ls="--")
plt.tight_layout()
plt.show()
3. Lasso
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Lasso / *Lasso Regression (L1 Regularization - Least Absolute Shrinkage and Selection Operator)*.
* **Famille :** Apprentissage supervisé — Régression avec sélection de variables.
* **Objectif :** Prédire une variable continue tout en sélectionnant automatiquement les variables les plus pertinentes en annulant les coefficients inutiles.
* **Intuition & Fonctionnement :**
  Lasso ajoute une pénalité sur la somme des valeurs absolues des coefficients (norme $L_1$).
  1. *Entrée :* Données $X$ standardisées et cible $y$.
  2. *Calcul/Apprentissage :* Minimisation de la perte MCO + pénalité $L_1$. En raison de la géométrie de la norme $L_1$ (coins pointus), certains coefficients deviennent exactement égaux à zéro.
  3. *Sortie :* Modèle parcimonieux (*sparse*) utilisant uniquement un sous-ensemble des variables d'origine.
* **Formule mathématique :**
  $$\mathcal{L}_{Lasso}(\beta) = \frac{1}{2n} \|y - X\beta\|^2_2 + \alpha \|\beta\|_1 = \frac{1}{2n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 + \alpha \sum_{j=1}^{p} |\beta_j|$$
  - **Définition des symboles :** $y_i$ valeur réelle, $\hat{y}_i$ prédiction, $\beta_j$ coefficients, $\alpha$ intensité de la pénalité $L_1$.
  - **Paramètres (appris) :** Coefficients $\beta$.
  - **Hyperparamètres :** $\alpha \ge 0$, `max_iter` (nombre maximal d'itérations du solveur).
* **Avantages :**
  1. Réalise une sélection automatique de caractéristiques (*feature selection*).
  2. Rend le modèle final très lisible et réduit le bruit en éliminant les variables non pertinentes.
* **Limites :**
  1. Si deux variables sont très corrélées, Lasso en choisit une arbitrairement et met l'autre à zéro.
  2. Si $p > n$ (plus de variables que d'observations), Lasso ne peut sélectionner au maximum que $n$ variables.

---

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le graphique illustre l'annulation des coefficients : lorsque $\alpha$ augmente, les trajectoires de plusieurs coefficients atteignent la ligne zéro (sélection automatique des variables).

---

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Marketing / Scoring) :** Sélection des variables prédictives parmi des centaines de métriques de comportement client en ligne pour modéliser le panier moyen.
* **Cas d'usage 2 (Génomique / Bioinformatique) :** Identification des quelques gènes responsables d'une maladie parmi des milliers de gènes analysés.
* **Nature des données & Prérequis :** Variables numériques obligatoirement standardisées (`StandardScaler`).
* **Cas peu approprié :** Situations où toutes les variables apportent une petite contribution utile et corrélée (préérer Ridge).
* **Métriques adaptées :** $R^2$, MAE, RMSE.

---

### 4. Analyse de l'expérience Python
* **Entrées du code :** `Xr_ent` et `yr_ent` dans un `Pipeline` avec `StandardScaler()` et `Lasso(alpha=.1, max_iter=10000)`.
* **Hyperparamètres définis :** `alpha=0.1`, `max_iter=10000`.
* **Sortie obtenue :** Affiche le coefficient de détermination ($R^2$).
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Mesure la proportion de variance expliquée par le modèle Lasso sur les données de test.
  - *Ne permet PAS de conclure :* Ne montre pas explicitement quelles variables ont été éliminées (coefficients mis à zéro).

4. Régression polynomiale
1. Bloc Markdown (à copier dans me cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Régression polynomiale / *Polynomial Regression*.
* **Famille :** Apprentissage supervisé — Régression non linéaire.
* **Objectif :** Capter des relations non linéaires entre les variables explicatives et la cible en générant des termes polynomiaux et d'interaction.
* **Intuition & Fonctionnement :**
  L'algorithme transforme l'espace des données de départ en créant des combinaisons de puissances ($x^2, x^3, ...$) et de produits croisés ($x_1 x_2$), puis applique une régression linéaire sur cet espace élargi.
  1. *Entrée :* Variables $X$.
  2. *Calcul/Apprentissage :* Transformation via `PolynomialFeatures(degree=d)` suivie de l'ajustement d'un modèle linéaire (souvent régularisé avec Ridge/Lasso).
  3. *Sortie :* Courbe ou surface d'ajustement non linéaire.
* **Formule mathématique :**
  $$\hat{y} = \beta_0 + \beta_1 x + \beta_2 x^2 + \dots + \beta_d x^d$$
  - **Définition des symboles :** $d$ est le degré du polynôme, $\beta_k$ sont les coefficients associés aux puissances $x^k$.
  - **Paramètres (appris) :** Coefficients $\beta$.
  - **Hyperparamètres :** Degré $d$ (`degree`), inclusion du terme de biais (`include_bias`).
* **Avantages :**
  1. Permet de modéliser des courbes et des tendances complexes sans changer de framework linéaire.
  2. Conserve une résolution analytique directe via les MCO/Ridge.
* **Limites :**
  1. Explosion combinatoire du nombre de caractéristiques quand le nombre de variables et le degré augmentent.
  2. Risque élevé de surapprentissage (*overfitting*) sur les bords du domaine de données.

---

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le graphique compare l'ajustement linéaire classique (droit) et l'ajustement polynomial (courbe), montrant comment le polynôme capture la courbure naturelle des données.

---

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Modélisation de rendements) :** Modélisation de la courbe des taux d'intérêt ou prévision du coût marginal de production selon le volume (économies d'échelle puis déséconomies d'échelle).
* **Cas d'usage 2 :** Estimation de la trajectoire de croissance des ventes d'un nouveau produit après son lancement (courbe en S ou parabole).
* **Nature des données & Prérequis :** Variables numériques. La régularisation (Ridge/Lasso) est fortement recommandée.
* **Cas peu approprié :** Problèmes comportant un très grand nombre de variables d'entrée (ex: 100 variables à un degré 3 créent des centaines de milliers de colonnes).
* **Métriques adaptées :** MAE, RMSE, $R^2$.

---

### 4. Analyse de l'expérience Python
* **Entrées du code :** `Xr_ent` et `yr_ent` dans un pipeline composé de `StandardScaler()`, `PolynomialFeatures(degree=2, include_bias=False)` et `Ridge(alpha=10)`.
* **Hyperparamètres définis :** `degree=2`, `alpha=10`.
* **Sortie obtenue :** Affiche la MAE sur le jeu de test.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Montre le score de l'intégration polynomiale de degré 2 régularisée sur le jeu de test.
  - *Ne permet PAS de conclure :* Ne montre pas si un degré plus élevé (ex: 3 ou 4) améliorerait le résultat ou provoquerait du surapprentissage.

Python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures

np.random.seed(42)
x_poly = np.linspace(-3, 3, 40)
y_poly = 0.5 * x_poly**2 - x_poly + 1 + np.random.normal(0, 1, 40)

pipe_poly = make_pipeline(
    PolynomialFeatures(degree=2), Ridge(alpha=0.1)
)
pipe_poly.fit(x_poly.reshape(-1, 1), y_poly)

plt.figure(figsize=(8, 4.5))
plt.scatter(x_poly, y_poly, color='blue', label='Données réelles')
plt.plot(
    x_poly,
    pipe_poly.predict(x_poly.reshape(-1, 1)),
    color='red',
    linewidth=2,
    label='Ajustement Polynomial (Degré 2)',
)
plt.title('Schéma explicatif : Régression Polynomiale de degré 2')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
plt.tight_layout()
plt.show()
5. Arbre de régression
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Arbre de décision pour la régression / *Decision Tree Regressor*.
* **Famille :** Apprentissage supervisé — Arbres de décision / Non paramétrique.
* **Objectif :** Prédire une cible continue en découpant l'espace des variables en régions rectangulaires disjointes.
* **Intuition & Fonctionnement :**
  L'algorithme sépare récursivement les données selon des seuils sur les variables qui réduisent au maximum la variance de la cible dans chaque sous-groupe (feuille).
  1. *Entrée :* Données $X$ et cible $y$.
  2. *Calcul/Apprentissage :* À chaque nœud, recherche de la meilleure variable $j$ et du meilleur seuil $t$ pour minimiser la somme des erreurs quadratiques des deux sous-groupes créés.
  3. *Sortie :* La prédiction pour une feuille est la moyenne des valeurs cibles des observations qui s'y trouvent.
* **Formule mathématique :**
  Critère de séparation (minimisation de la variance) :
  $$\text{Impureté}(R) = \frac{1}{|R|} \sum_{i \in R} (y_i - \bar{y}_R)^2$$
  - **Définition des symboles :** $R$ est la région (nœud), $|R|$ le nombre d'échantillons dans la région, $\bar{y}_R$ la moyenne des valeurs $y$ dans cette région.
  - **Paramètres (appris) :** Structure de l'arbre, variables de coupure, seuils et valeurs des feuilles $\bar{y}_R$.
  - **Hyperparamètres :** `max_depth` (profondeur max), `min_samples_split`, `min_samples_leaf`.
* **Avantages :**
  1. Modèle intuitif et visuellement interprétable sous forme de diagramme décisionnel.
  2. Ne nécessite aucune mise à l'échelle des données et gère les relations non linéaires complexes.
* **Limites :**
  1. Risque très élevé de surapprentissage si la profondeur n'est pas restreinte.
  2. Prédictions constantes par morceaux (incapable d'extrapoler en dehors de la plage d'entraînement).

---

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le graphique montre comment un arbre de régression crée une fonction en escalier (prédictions constantes par morceaux dans chaque feuille).

---

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Gestion / Immobilier) :** Estimation de la valeur d'un actif immobilier avec des règles simples et explicables aux clients (ex: "Si surface > 80m² et pièces > 3 alors valeur = X").
* **Cas d'usage 2 (Logistique) :** Estimation du temps de livraison en fonction du secteur et du créneau horaire.
* **Nature des données & Prérequis :** Variables numériques ou catégorielles encodées. Aucune standardisation requise.
* **Cas peu approprié :** Données nécessitant une extrapolation continue précise en dehors des bornes d'entraînement.
* **Métriques adaptées :** MAE, RMSE, $R^2$.

---

### 4. Analyse de l'expérience Python
* **Entrées du code :** `DecisionTreeRegressor(max_depth=5, random_state=42)` entraîné sur `Xr_ent` et `yr_ent`.
* **Hyperparamètres définis :** `max_depth=5`, `random_state=GRAINE`.
* **Sortie obtenue :** Affiche la MAE sur l'échantillon de test.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Mesure l'erreur d'un arbre limité à une profondeur de 5.
  - *Ne permet PAS de conclure :* Ne permet pas d'évaluer le surapprentissage sans comparer ce résultat au score d'entraînement (`Xr_ent`).

6. Forêt aléatoire — régression
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Forêt aléatoire pour la régression / *Random Forest Regressor*.
* **Famille :** Apprentissage supervisé — Méthodes d'ensemble (*Bagging* / Agrégation).
* **Objectif :** Améliorer la précision et réduire la variance de la prédiction en combinant un grand nombre d'arbres de décision indépendants.
* **Intuition & Fonctionnement :**
  L'algorithme construit un ensemble d'arbres de décision entraînés chacun sur un sous-échantillon bootstrap des données et un sous-ensemble aléatoire des variables.
  1. *Entrée :* Données $X$ et cible $y$.
  2. *Calcul/Apprentissage :* Entraînement en parallèle de $B$ arbres de régression sur des tirages avec remise et sous-sélection aléatoire de variables à chaque nœud.
  3. *Sortie :* La prédiction finale est la moyenne exacte des prédictions de tous les arbres individuels.
* **Formule mathématique :**
  $$\hat{y}_{RF} = \frac{1}{B} \sum_{b=1}^{B} T_b(x)$$
  - **Définition des symboles :** $B$ est le nombre total d'arbres (`n_estimators`), $T_b(x)$ est la prédiction issue du $b$-ième arbre de décision.
  - **Paramètres (appris) :** L'ensemble des structures et seuils de tous les $B$ arbres.
  - **Hyperparamètres :** `n_estimators`, `max_depth`, `max_features`, `min_samples_split`.
* **Avantages :**
  1. Très robuste au surapprentissage par rapport à un arbre individuel.
  2. Offre des mesures fiables d'importance des variables.
* **Limites :**
  1. Coût en calcul et mémoire plus élevé qu'un simple arbre.
  2. Modèle en "boîte noire" difficile à interpréter visuellement par rapport à un seul arbre.

---

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le schéma représente le principe du Bagging : tirage de sous-échantillons bootstrap $\rightarrow$ entraînement parallèle d'arbres $\rightarrow$ moyenne des prédictions pour l'obtention du résultat final.

---

### 3. Contextes d'utilisation

7. Gradient Boosting — régression
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Gradient Boosting pour la régression / *Gradient Boosting Regressor (HistGradientBoostingRegressor)*.
* **Famille :** Apprentissage supervisé — Méthodes d'ensemble séquentiel (*Boosting*).
* **Objectif :** Obtenir un modèle de prédiction très performant en construisant séquentiellement des arbres où chaque nouvel arbre corrige les erreurs (résidus) des précédents.
* **Intuition & Fonctionnement :**
  Contrairement à la Forêt Aléatoire qui entraîne les arbres en parallèle, le Gradient Boosting construit les arbres un par un.
  1. *Entrée :* Données $X$ et cible $y$.
  2. *Calcul/Apprentissage :* Initialisation avec une prédiction constante. À chaque étape, calcul des résidus (erreurs), puis entraînement d'un nouvel arbre pour prédire ces résidus.
  3. *Sortie :* Somme pondérée de tous les arbres entraînés.
* **Formule mathématique :**
  $$\hat{y}^{(m)} = \hat{y}^{(m-1)} + \nu \cdot h_m(x)$$
  - **Définition des symboles :** $\hat{y}^{(m)}$ est la prédiction à l'étape $m$, $h_m(x)$ est l'arbre entraîné sur les résidus, $\nu$ est le taux d'apprentissage (*learning rate*).
  - **Paramètres (appris) :** Les structures des arbres séquentiels et leurs poids.
  - **Hyperparamètres :** `max_iter` (nombre d'arbres), `learning_rate`, `max_depth`, `l2_regularization`.
* **Avantages :**
  1. Souvent l'un des algorithmes les plus performants et précis sur données tabulaires.
  2. Gestion native efficace des grandes masses de données (`HistGradientBoosting`).
* **Limites :**
  1. Très sensible au surapprentissage si le nombre d'itérations est trop grand ou le taux d'apprentissage mal réglé.
  2. Entraînement séquentiel (plus difficile à paralléliser qu'une forêt aléatoire).

8. Régression logistique
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Régression logistique / *Logistic Regression*.
* **Famille :** Apprentissage supervisé — Classification (binaire et multiclasse).
* **Objectif :** Estimer la probabilité qu'une observation appartienne à une classe spécifique à l'aide d'une fonction sigmoïde.
* **Intuition & Fonctionnement :**
  L'algorithme combine linéairement les variables explicatives puis transforme ce résultat à l'aide de la fonction logistique pour obtenir une valeur entre 0 et 1.
  1. *Entrée :* Matrice de caractéristiques $X$ et étiquettes catégorielles $y \in \{0, 1\}$.
  2. *Calcul/Apprentissage :* Minimisation de la perte d'entropie croisée (*log-loss*) par descente de gradient.
  3. *Sortie :* Probabilité d'appartenance à la classe positive $\hat{p}$, convertie en classe 0 ou 1 selon un seuil (ex: 0.5).
* **Formule mathématique :**
  $$\hat{p} = \sigma(z) = \frac{1}{1 + e^{-(X\beta + \beta_0)}}$$
  Perte entropie croisée binaire :
  $$\mathcal{L}(\beta) = -\frac{1}{n} \sum_{i=1}^{n} \left[ y_i \log(\hat{p}_i) + (1 - y_i) \log(1 - \hat{p}_i) \right]$$
  - **Définition des symboles :** $\sigma(z)$ est la fonction sigmoïde, $y_i$ la vraie classe, $\hat{p}_i$ la probabilité prédite, $\beta$ les coefficients.
  - **Paramètres (appris) :** Coefficients $\beta$ et intercept $\beta_0$.
  - **Hyperparamètres :** `C` (inverse de la pénalité), `penalty` (l1, l2, elasticnet), `solver`.
* **Avantages :**
  1. Modèle très rapide et facilement interprétable via les ratios de chances (*odds ratios*).
  2. Fournit des probabilités bien étalonnées en plus des prédictions de classes.
* **Limites :**
  1. Suppose une frontière de décision linéaire entre les classes.
  2. Sensible à la multicolinéarité et aux valeurs extrêmes.

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le graphique présente la courbe en S de la fonction sigmoïde qui compresse n'importe quelle valeur réelle dans l'intervalle [0, 1] avec le seuil de décision fixé à 0.5.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Credit Scoring) :** Évaluation du risque d'accord de prêt (prédire si un emprunteur sera en défaut de paiement ou non).
* **Cas d'usage 2 (Marketing) :** Prédiction de l'attrition client (*churn*) dans les services d'abonnement.
* **Nature des données & Prérequis :** Variables numériques (mise à l'échelle vivement recommandée via `StandardScaler`).
* **Cas peu approprié :** Données avec des frontières de séparation hautement complexes et non linéaires (ex: reconnaissance d'images).
* **Métriques adaptées :** F1-Score, AUC-ROC, Précision, Rappel, Matrice de confusion.

### 4. Analyse de l'expérience Python

* **Hyperparamètres définis :** `max_iter=1000`, `random_state=GRAINE`.
* **Sortie obtenue :** Affiche le F1-score sur l'échantillon de test.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Donne la mesure du compromis précision/rappel sur le jeu de test binaire synthétique.
  - *Ne permet PAS de conclure :* Ne montre pas si ajuster le seuil de décision (actuellement 0.5) optimiserait le rappel ou la précision selon les contraintes métier.

9. KNN
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** K plus proches voisins / *K-Nearest Neighbors (KNN)*.
* **Famille :** Apprentissage supervisé — Classification / Méthode basée sur les instances (*Lazy Learning*).
* **Objectif :** Classer une nouvelle observation en se basant sur la classe majoritaire de ses $k$ voisins les plus proches dans l'espace des variables.
* **Intuition & Fonctionnement :**
  L'algorithme ne construit pas de modèle explicite pendant l'entraînement ; il stocke simplement les données.
  1. *Entrée :* Nouvelle donnée $x$ et ensemble d'entraînement stocké.
  2. *Calcul/Apprentissage :* Calcul de la distance (ex: euclidienne) entre $x$ et toutes les données stockées pour identifier les $k$ voisins les plus proches.
  3. *Sortie :* Vote majoritaire parmi les $k$ voisins pour déterminer la classe attribuée.
* **Formule mathématique :**
  Distance euclidienne entre deux points $p$ et $q$ :
  $$d(p, q) = \sqrt{\sum_{j=1}^{d} (p_j - q_j)^2}$$
  - **Définition des symboles :** $d$ est le nombre de dimensions (caractéristiques), $p_j$ et $q_j$ les coordonnées respectives des points.
  - **Paramètres (appris) :** Aucun (l'algorithme stocke simplement l'ensemble des données d'entraînement).
  - **Hyperparamètres :** `n_neighbors` ($k$), `metric` (euclidean, manhattan, etc.), `weights` (uniform, distance).
* **Avantages :**
  1. Très simple à comprendre, sans hypothèse sur la distribution des données.
  2. S'adapte naturellement aux problèmes multiclasses et non linéaires.
* **Limites :**
  1. Prédictions très lentes sur de grands jeux de données (doit parcourir toutes les données à chaque prédiction).
  2. Extrêmement sensible au "fléau de la dimension" et à l'absence d'échelle uniforme entre variables.

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le schéma illustre comment le choix de $k$ modifie la décision : pour $k=3$, la classe A l'emporte, tandis que pour un $k$ plus grand, la classe B peut devenir majoritaire autour du point cible.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Risk Management) :** Détection de profils de clients à haut risque par comparaison directe avec les profils historiques similaires.
* **Cas d'usage 2 (Recommandation) :** Systèmes de recommandation basés sur le filtrage collaboratif (trouver les $k$ utilisateurs ayant des goûts similaires).
* **Nature des données & Prérequis :** Variables numériques impérativement standardisées (`StandardScaler`).
* **Cas peu approprié :** Jeux de données volumineux avec des centaines de colonnes (la distance perd de son sens en haute dimension).
* **Métriques adaptées :** F1-Score, Précision, Rappel, Accuracy.

### 4. Analyse de l'expérience Python
* **Entrées du code :** `Xc_ent` et `yc_ent` dans un pipeline avec `StandardScaler()` et `KNeighborsClassifier(n_neighbors=5)`.

* **Sortie obtenue :** Affiche le F1-score sur l'échantillon de test.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Mesure la capacité de classification de 5 voisins sur les données standardisées.
  - *Ne permet PAS de conclure :* Ne montre pas si $k=5$ est la valeur optimale ou si $k=3$ ou $k=7$ donnerait de meilleurs résultats.

10. Naive Bayes
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Naïve Bayes / *Naive Bayes (Gaussian Naive Bayes)*.
* **Famille :** Apprentissage supervisé — Classification probabiliste.
* **Objectif :** Calcule la probabilité a posteriori d'appartenir à chaque classe en utilisant le théorème de Bayes sous l'hypothèse de l'indépendance conditionnelle des caractéristiques.
* **Intuition & Fonctionnement :**
  L'algorithme combine la probabilité a priori des classes avec les probabilités conditionnelles de chaque variable (supposée gaussienne pour des données continues).
  1. *Entrée :* Données $X$ et étiquettes $y$.
  2. *Calcul/Apprentissage :* Estimation de la moyenne et de la variance de chaque caractéristique pour chaque classe.
  3. *Sortie :* Classe qui maximise la probabilité a posteriori (règle du Maximum A Posteriori - MAP).
* **Formule mathématique :**
  Théorème de Bayes appliqué avec l'hypothèse "naïve" d'indépendance :
  $$P(Y = c \mid X_1, \dots, X_d) \propto P(Y = c) \prod_{j=1}^{d} P(X_j \mid Y = c)$$
  - **Définition des symboles :** $P(Y=c)$ est la probabilité a priori de la classe $c$, $P(X_j \mid Y=c)$ est la densité de probabilité gaussienne de la variable $j$ conditionnellement à la classe $c$.
  - **Paramètres (appris) :** Moyennes $\mu_{c,j}$ et variances $\sigma_{c,j}^2$ par classe et par variable.
  - **Hyperparamètres :** `var_smoothing` (stabilité numérique).
* **Avantages :**
  1. Extrêmement rapide à entraîner et peu gourmand en puissance de calcul.
  2. Fonctionne de façon remarquable sur la classification de textes/emails malgré l'hypothèse simplificatrice d'indépendance.
* **Limites :**
  1. Supposer l'indépendance totale des variables est rarement vrai en pratique.
  2. Sensible au problème de "fréquence nulle" (géré par le lissage de Laplace).

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Marketing / Communication) :** Filtrage des spams et classification automatique des emails entrants des clients.
* **Cas d'usage 2 (Finance / Sentiment Analysis) :** Analyse du sentiment (positif/négatif/neutre) dans les rapports financiers ou actualités boursières.
* **Nature des données & Prérequis :** Variables continues (GaussianNB) ou comptages/fréquences textuelles (MultinomialNB).
* **Cas peu approprié :** Données comportant de très fortes corrélations dépendantes entre variables explicatives.
* **Métriques adaptées :** F1-Score, Précision, Rappel, ROC-AUC.

  - *Permet de conclure :* Évalue l'efficacité du classifieur bayésien gaussien rapide sur le jeu de classification.
  - *Ne permet PAS de conclure :* Ne valide pas si l'hypothèse de normalité gaussienne des variables d'entrée est réellement respectée sur ces données.

11. Arbre de décision
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Arbre de décision pour la classification / *Decision Tree Classifier*.
* **Famille :** Apprentissage supervisé — Arbres de décision.
* **Objectif :** Séparer les données en sous-groupes de plus en plus homogènes (*purs*) par une série de règles SI-ALORS sur les caractéristiques.
* **Intuition & Fonctionnement :**
  L'algorithme choisit à chaque étape la variable et le seuil qui maximisent le gain d'information ou réduisent au mieux l'impureté de Gini.
  1. *Entrée :* Données $X$ et étiquettes catégorielles $y$.
  2. *Calcul/Apprentissage :* Évaluation de l'impureté de Gini ou de l'entropie à chaque nœud pour déterminer la séparation optimale.
  3. *Sortie :* Classe majoritaire présente dans la feuille d'arrivée.
* **Formule mathématique :**
  Indice d'impureté de Gini :
  $$Gini(N) = 1 - \sum_{i=1}^{K} p_i^2$$
  - **Définition des symboles :** $K$ est le nombre de classes, $p_i$ est la proportion d'éléments appartenant à la classe $i$ dans le nœud $N$.
  - **Paramètres (appris) :** Variables sélectionnées à chaque nœud, seuils de séparation et affectation des classes aux feuilles.
  - **Hyperparamètres :** `max_depth`, `min_samples_split`, `criterion` (gini, entropy).
* **Avantages :**
  1. Entièrement explicable et visualisable sous forme de diagramme décisionnel simple.
  2. Gère nativement les variables numériques et catégorielles sans mise à l'échelle.
* **Limites :**
  1. Très enclin au surapprentissage si sa profondeur n'est pas limitée.
  2. Instable : une petite modification des données d'entraînement peut changer radicalement la structure de l'arbre.

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le schéma représente la séparation de l'espace des données par des droites orthogonales correspondant aux règles SI-ALORS générées par les nœuds de l'arbre.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Octroi de crédit) :** Arbre d'approbation automatique des prêts bancaires avec règles explicites destinées aux régulateurs.
* **Cas d'usage 2 (Gestion RH) :** Modélisation du risque de démission d'un employé selon l'ancienneté et le salaire.
* **Nature des données & Prérequis :** Variables numériques ou catégorielles. Aucune normalisation préalable nécessaire.
* **Cas peu approprié :** Relations fortement diagonales ou linéaires continues nécessitant une grande quantité de marches d'escalier pour être approchées.
* **Métriques adaptées :** F1-Score, Précision, Rappel, Matrice de confusion.

12. Forêt aléatoire — classification
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Forêt aléatoire pour la classification / *Random Forest Classifier*.
* **Famille :** Apprentissage supervisé — Méthodes d'ensemble (*Bagging*).
* **Objectif :** Combiner la prédiction d'un grand nombre d'arbres de décision indépendants pour améliorer la généralisation et réduire la variance.
* **Intuition & Fonctionnement :**
  L'algorithme introduit une double aléatorisation (tirage des observations avec remise + sous-sélection aléatoire de variables à chaque nœud).
  1. *Entrée :* Données $X$ et étiquettes $y$.
  2. *Calcul/Apprentissage :* Entraînement en parallèle de $B$ arbres de décision sur des échantillons bootstrap distincts.
  3. *Sortie :* Vote majoritaire (ou moyenne des probabilités) des arbres de la forêt.
* **Formule mathématique :**
  Vote majoritaire :
  $$\hat{y}_{RF} = \text{mode} \left( \{ T_1(x), T_2(x), \dots, T_B(x) \} \right)$$
  - **Définition des symboles :** $B$ est le nombre d'arbres (`n_estimators`), $T_b(x)$ est la prédiction du $b$-ième arbre de classification.
  - **Paramètres (appris) :** L'ensemble des structures de tous les arbres individuels de la forêt.
  - **Hyperparamètres :** `n_estimators`, `max_depth`, `max_features`, `criterion`.
* **Avantages :**
  1. Excellente performance globale sans nécessiter un réglage poussé des hyperparamètres.
  2. Évite le surapprentissage grâce au principe d'agrégation d'arbres diversifiés.
* **Limites :**
  1. Perte d'interprétabilité directe par rapport à un arbre individuel.
  2. Consommation en mémoire et temps de calcul plus importants sur les très grands jeux de données.

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le schéma illustre l'obtention du résultat final grâce au vote majoritaire compilant les réponses issues des différents arbres indépendants.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Détection de fraude) :** Détection de transactions frauduleuses par carte de crédit sur des données historiques complexes.
* **Cas d'usage 2 (Marketing / Upselling) :** Ciblage des clients ayant la plus forte propension à acheter un produit complémentaire.
* **Nature des données & Prérequis :** Données numériques ou catégorielles encodées. Aucune mise à l'échelle requise.
* **Cas peu approprié :** Systèmes embarrassés par une forte contrainte de temps de latence en temps réel extrême (où un modèle plus léger est exigé).
* **Métriques adaptées :** F1-Score, AUC-ROC, Précision, Rappel.

13. SVM
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Machines à vecteurs de support / *Support Vector Machines (SVM / SVC)*.
* **Famille :** Apprentissage supervisé — Classification par séparateur à marge maximale.
* **Objectif :** Trouver l'hyperplan qui sépare au mieux les classes en maximisant la distance (marge) entre l'hyperplan et les points les plus proches (vecteurs de support).
* **Intuition & Fonctionnement :**
  L'algorithme s'appuie sur la marge maximale et utilise le "truc du noyau" (*kernel trick*) pour projeter les données dans un espace de plus grande dimension où elles deviennent linéairement séparables.
  1. *Entrée :* Données $X$ et étiquettes $y \in \{-1, 1\}$.
  2. *Calcul/Apprentissage :* Résolution d'un problème d'optimisation quadratique sous contraintes pour trouver les vecteurs de support.
  3. *Sortie :* Classification d'un point selon le côté de l'hyperplan séparateur où il se situe.
* **Formule mathématique :**
  Problème d'optimisation (marge adoucie) :
  $$\min_{w, b, \xi} \frac{1}{2} \|w\|^2 + C \sum_{i=1}^{n} \xi_i \quad \text{sujet à} \quad y_i(w^T \phi(x_i) + b) \ge 1 - \xi_i$$
  - **Définition des symboles :** $w$ est le vecteur normal à l'hyperplan, $C$ contrôle le compromis marge/erreur, $\xi_i$ sont les variables d'ajustement (*slack variables*), $\phi(x)$ la projection par le noyau.
  - **Paramètres (appris) :** Poids $w$, biais $b$ et choix des vecteurs de support.
  - **Hyperparamètres :** `C`, `kernel` (rbf, linear, poly), `gamma`.
* **Avantages :**
  1. Trè efficace en haute dimension (même lorsqu'il y a plus de variables que d'observations).
  2. Robuste au surapprentissage grâce à l'optimisation axée uniquement sur les vecteurs de support.
* **Limites :**
  1. Temps de calcul quadratique à cubique avec le nombre d'observations (lent sur très grands datasets).
  2. Nécessite obligatoirement un réglage méticuleux des hyperparamètres (`C`, `gamma`) et la mise à l'échelle.

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le schéma illustre l'hyperplan séparateur, la largeur de la marge maximale et la localisation des vecteurs de support qui définissent la frontière.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Trading) :** Classification de mouvements de marché à court terme basés sur de multiples indicateurs techniques complexes.
* **Cas d'usage 2 (Biologie / Santé) :** Diagnostic médical à partir de données d'imagerie ou d'analyses génomiques en haute dimension.
* **Nature des données & Prérequis :** Variables numériques obligatoirement standardisées (`StandardScaler`).
* **Cas peu approprié :** Très grands jeux de données dépassant plusieurs centaines de milliers d'observations.
* **Métriques adaptées :** F1-Score, Précision, Rappel, ROC-AUC.

### 4. Analyse de l'expérience Python
* **Entrées du code :** `Xc_ent` et `yc_ent` dans un pipeline avec `StandardScaler()` et `SVC(kernel='rbf', C=1.0, random_state=42)`.
* **Hyperparamètres définis :** `kernel='rbf'`, `C=1.0`, `random_state=GRAINE`.
* **Sortie obtenue :** Affiche le F1-score sur l'échantillon de test `Xc_test`.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Mesure la qualité de classification avec un noyau RBF gaussien à marge adoucie standard (`C=1.0`).
  - *Ne permet PAS de conclure :* Ne montre pas l'influence de la valeur de `gamma` ou d'un noyau alternatif sur le résultat.

14. Gradient Boosting — classification
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Gradient Boosting pour la classification / *Gradient Boosting Classifier (HistGradientBoostingClassifier)*.
* **Famille :** Apprentissage supervisé — Méthodes d'ensemble séquentiel (*Boosting*).
* **Objectif :** Construire séquentiellement une suite d'arbres de décision simples où chaque nouvel arbre cherche à corriger les erreurs de l'ensemble précédent.
* **Intuition & Fonctionnement :**
  L'algorithme optimise une fonction de perte globale en appliquant la descente de gradient dans l'espace des fonctions (les arbres).
  1. *Entrée :* Données $X$ et étiquettes $y$.
  2. *Calcul/Apprentissage :* Calcul du gradient négatif de la fonction de perte logistique à chaque étape, puis entraînement d'un arbre sur ce gradient.
  3. *Sortie :* Probabilité finale obtenue par combinaison pondérée de tous les arbres.
* **Formule mathématique :**
  Mise à jour séquentielle du modèle :
  $$F_m(x) = F_{m-1}(x) + \gamma_m h_m(x)$$
  - **Définition des symboles :** $F_m(x)$ est le modèle à l'étape $m$, $h_m(x)$ l'arbre entraîné sur les pseudo-résidus (gradients), $\gamma_m$ le poids affecté à cet arbre.
  - **Paramètres (appris) :** Ensemble des arbres successifs et leurs coefficients d'apprentissage.
  - **Hyperparamètres :** `max_iter`, `learning_rate`, `max_depth`, `l2_regularization`.
* **Avantages :**
  1. Performances de prédiction exceptionnelles sur les données tabulaires.
  2. Gère efficacement les valeurs manquantes et les jeux de données volumineux via la discrétisation par histogrammes.
* **Limites :**
  1. Risque d'overfitting si le nombre d'itérations est trop élevé sans régularisation adéquate.
  2. Construction séquentielle plus longue à entraîner qu'un modèle basé sur le bagging parallèle.

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le schéma montre comment l'erreur globale de classification diminue au fil des itérations successives d'ajout d'arbres.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / scoring bancaire) :** Détection de l'insolvabilité des clients pour la validation automatique des crédits.
* **Cas d'usage 2 (Marketing) :** Prédiction de la conversion d'un prospect à la suite d'une campagne publicitaire.
* **Nature des données & Prérequis :** Données tabulaires (numériques/catégorielles).
* **Cas peu approprié :** Traitement de données brutes d'images ou de signal textuel non structuré.
* **Métriques adaptées :** F1-Score, AUC-ROC, PR-AUC, Précision, Rappel.

### 4. Analyse de l'expérience Python

15. AdaBoost
1. Bloc Markdown (à copier dans me cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** AdaBoost / *Adaptive Boosting (AdaBoost Classifier)*.
* **Famille :** Apprentissage supervisé — Méthodes d'ensemble séquentiel (*Boosting*).
* **Objectif :** Combiner plusieurs classifieurs faibles (souvent des arbres très courts d'un seul niveau, appelés *stumps*) en augmentant le poids des observations mal classées à chaque étape.
* **Intuition & Fonctionnement :**
  Les modèles sont entraînés les uns après les autres. À chaque tour, les données mal classées par l'arbre précédent reçoivent un poids plus élevé pour contraindre le suivant à se focaliser sur elles.
  1. *Entrée :* Données $X$ et étiquettes $y \in \{-1, 1\}$, poids d'observations initialisés uniformément à $1/n$.
  2. *Calcul/Apprentissage :* Entraînement séquentiel de classifieurs faibles, calcul de leur erreur, et mise à jour exponentielle des poids des exemples.
  3. *Sortie :* Vote pondéré de l'ensemble des classifieurs faibles.
* **Formule mathématique :**
  Mise à jour des poids d'observations :
  $$w_i^{(m+1)} = w_i^{(m)} \exp \left( \alpha_m y_i h_m(x_i) \right)$$
  Poids du classifieur : $\alpha_m = \frac{1}{2} \ln \left( \frac{1 - \epsilon_m}{\epsilon_m} \right)$
  - **Définition des symboles :** $w_i^{(m)}$ est le poids de l'exemple $i$ à l'étape $m$, $\epsilon_m$ est le taux d'erreur du classifieur faible $h_m$, $\alpha_m$ est l'importance accordée à ce classifieur.
  - **Paramètres (appris) :** Poids $\alpha_m$ des classifieurs et règles des arbres courts.
  - **Hyperparamètres :** `n_estimators`, `learning_rate`.
* **Avantages :**
  1. Algorithme historiquement élégant et simple à mettre en œuvre.
  2. Peu enclin au surapprentissage sur des jeux de données propres sans trop de bruit.
* **Limites :**
  1. Extrêmement sensible au bruit et aux valeurs atypiques (*outliers*), qui reçoivent des poids démesurés.
  2. Moins performant que le Gradient Boosting sur les jeux de données complexes et bruités.

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le schéma illustre l'augmentation de la taille (poids) des points mal classés qui contraint le classifieur suivant à ajuster sa frontière.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Vision par ordinateur) :** Détection de visages historiques (algorithme de Viola-Jones basé sur AdaBoost).
* **Cas d'usage 2 (Marketing / Churn) :** Identification rapide de clients insatisfaits sur des critères comportementaux basiques.
* **Nature des données & Prérequis :** Variables numériques ou catégorielles. Sensible au bruit dans les étiquettes.
* **Cas peu approprié :** Jeux de données fortement bruités comportant des erreurs d'étiquetage.
* **Métriques adaptées :** F1-Score, Précision, Rappel, Accuracy.

### 4. Analyse de l'expérience Python

* **Hyperparamètres définis :** `n_estimators=70`, `random_state=GRAINE`.

  - *Permet de conclure :* Mesure la performance d'une combinaison de 70 classifieurs faibles réadaptés.
  - *Ne permet PAS de conclure :* Ne montre pas l'évolution des poids attribués aux observations au cours des itérations.

16. XGBoost
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** XGBoost / *Extreme Gradient Boosting (XGBClassifier)*.
* **Famille :** Apprentissage supervisé — Méthodes d'ensemble séquentiel (*Boosting*).
* **Objectif :** Fournir une implémentation ultra-optimisée, rapide et régularisée du Gradient Boosting pour maximiser les performances de prédiction.
* **Intuition & Fonctionnement :**
  XGBoost utilise un développement de Taylor de second ordre de la fonction de perte et intègre une régularisation ($L_1$ et $L_2$) de la structure des arbres pour contrôler la complexité.
  1. *Entrée :* Données $X$ et étiquettes $y$.
  2. *Calcul/Apprentissage :* Minimisation de la fonction d'objectif avec calcul des dérivées premières (gradients) et secondes (hessiens).
  3. *Sortie :* Prédiction finale issue de la somme pondérée de tous les arbres de régression.
* **Formule mathématique :**
  Fonction d'objectif à l'étape $t$ :
  $$\mathcal{L}^{(t)} \approx \sum_{i=1}^{n} \left[ g_i f_t(x_i) + \frac{1}{2} h_i f_t^2(x_i) \right] + \Omega(f_t)$$
  Terme de régularisation : $\Omega(f) = \gamma T + \frac{1}{2} \lambda \sum_{j=1}^{T} w_j^2$
  - **Définition des symboles :** $g_i$ et $h_i$ sont les gradients du premier et second ordre, $T$ est le nombre de feuilles, $w_j$ les poids des feuilles, $\gamma$ et $\lambda$ les coefficients de régularisation.
  - **Paramètres (appris) :** Structure des arbres et valeurs des feuilles.
  - **Hyperparamètres :** `n_estimators`, `max_depth`, `learning_rate`, `subsample`, `colsample_bytree`, `reg_lambda`, `reg_alpha`.
* **Avantages :**
  1. Vitesse d'exécution exceptionnelle grâce à la parallélisation du calcul des séparations.
  2. Très haute précision, ayant dominé pendant des années les compétitions de science des données (Kaggle).
* **Limites :**
  1. Grand nombre d'hyperparamètres complexes à régler.
  2. Modèle boîte noire plus difficile à interpréter et à déployer sans outils d'explicabilité (SHAP).

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le schéma présente la régularisation de la structure des arbres combinée aux gradients du premier et second ordre qui optimisent la convergence.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Évaluation du risque) :** Systèmes de notation de crédit et détection de fraudes complexes en temps réel.
* **Cas d'usage 2 (E-commerce / Conversion) :** Prédiction du taux de clics (CTR) et personnalisation des recommandations en ligne.
* **Nature des données & Prérequis :** Variables numériques (gère nativement les valeurs manquantes et l'encodage).
* **Cas peu approprié :** Petits jeux de données simples où des modèles plus légers suffisent largement.
* **Métriques adaptées :** F1-Score, AUC-ROC, Log-Loss, PR-AUC.

### 4. Analyse de l'expérience Python
* **Entrées du code :** `XGBClassifier(n_estimators=80, max_depth=3, random_state=42, eval_metric='logloss')` ajusté sur `Xc_ent` et `yc_ent`.
* **Hyperparamètres définis :** `n_estimators=80`, `max_depth=3`, `eval_metric='logloss'`.
* **Sortie obtenue :** Affiche le F1-score calculé sur le jeu de test `Xc_test`.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Évalue les performances brutes du classifieur XGBoost régularisé sur le jeu de test.
  - *Ne permet PAS de conclure :* Ne montre pas l'influence du paramètre `subsample` ou des régularisations `lambda`/`alpha` sur le score final.

17. K-means
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** K-moyennes / *K-Means Clustering*.
* **Famille :** Apprentissage non supervisé — Regroupement (*Clustering* géométrique par partitions).
* **Objectif :** Partitionner un ensemble d'observations en $K$ groupes (clusters) distincts en minimisant la variance au sein de chaque groupe (inertie intra-classe).
* **Intuition & Fonctionnement :**
  L'algorithme positionne aléatoirement $K$ centroïdes, affecte chaque observation au centroïde le plus proche, puis réexécute l'actualisation des centroïdes en calculant la moyenne des points affectés jusqu'à convergence.
  1. *Entrée :* Données $X$ sans étiquettes et nombre préfixé de clusters $K$.
  2. *Calcul/Apprentissage :* Iteration alternée (Assignation des points aux centroïdes ➔ Mise à jour des positions des centroïdes).
  3. *Sortie :* Étiquette du cluster affecté à chaque observation et coordonnées finales des $K$ centroïdes.
* **Formule mathématique :**
  Minimisation de l'inertie intra-cluster ($WCSS$) :
  $$\min_{S} \sum_{k=1}^{K} \sum_{x \in S_k} \|x - \mu_k\|^2$$
  - **Définition des symboles :** $S_k$ représente l'ensemble des points attribués au cluster $k$, $\mu_k$ est le centroïde (moyenne) du cluster $k$.
  - **Paramètres (appris) :** Emplacements des $K$ centroïdes $\mu_1, \dots, \mu_K$.
  - **Hyperparamètres :** `n_clusters` ($K$), `init` (kmeans++, random), `n_init`, `max_iter`.
* **Avantages :**
  1. Algorithme très rapide, scalable et simple à implémenter sur de grandes bases de données.
  2. Convergence garantie en un nombre fini d'itérations.
* **Limites :**
  1. Nécessite de fixer le nombre de clusters $K$ à priori.
  2. Sensible aux initialisations, aux valeurs extrêmes et suppose des clusters de formes sphériques de tailles comparables.

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le schéma présente les $K$ centroïdes calculés et les frontières de Voronoi séparant les différentes régions d'attraction des clusters.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Marketing / Segmentation) :** Segmentation de la base clients en groupes homogènes d'achats pour cibler des offres marketing spécifiques.
* **Cas d'usage 2 (Finance / Portefeuilles) :** Regroupement d'actifs financiers aux comportements de rendements similaires pour la diversification.
* **Nature des données & Prérequis :** Variables numériques impérativement standardisées (`StandardScaler`).
* **Cas peu approprié :** Données comportant des clusters de formes arbitraires non sphériques (ex: croissants de lune) ou de densités très différentes.
* **Métriques adaptées :** Coefficient de Silhouette, Score Davies-Bouldin, Inertie (méthode du coude / *Elbow method*).

### 4. Analyse de l'expérience Python
* **Entrées du code :** `KMeans(n_clusters=3, n_init=10, random_state=42)` appliqué sur les données synthétiques `X_clu`.
* **Hyperparamètres définis :** `n_clusters=3`, `n_init=10`, `random_state=GRAINE`.
* **Sortie obtenue :** Affiche le score de Silhouette et affiche le graphique de dispersion colorié par clusters.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Mesure la séparation globale et l'homogénéité des 3 clusters trouvés sur ce nuage 2D.
  - *Ne permet PAS de conclure :* Ne garantit pas que $K=3$ soit la valeur optimale sur le plan métier sans analyser l'évolution du score de Silhouette pour d'autres valeurs de $K$.

18. Classification hiérarchique
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Classification hiérarchique / *Hierarchical Agglomerative Clustering (HAC)*.
* **Famille :** Apprentissage non supervisé — Regroupement hiérarchique (approche ascendante).
* **Objectif :** Construire une hiérarchie d'emboîtement des données sous forme d'arbre (dendrogramme) sans fixer le nombre de groupes au départ.
* **Intuition & Fonctionnement :**
  Chaque point commence dans son propre cluster. À chaque étape, l'algorithme fusionne les deux clusters les plus proches selon un critère de liaison (*linkage*) prédéfini jusqu'à ce que toutes les données forment un seul groupe.
  1. *Entrée :* Données $X$ et métrique de distance.
  2. *Calcul/Apprentissage :* Calcul séquentiel de la matrice des distances et fusion progressive des paires de clusters les plus proches.
  3. *Sortie :* Arbre hiérarchique (dendrogramme) découpé au niveau souhaité pour extraire $K$ clusters.
* **Formule mathématique :**
  Distance de Ward (minimisation de la variance cumulée) :
  $$\Delta(A, B) = \frac{n_A n_B}{n_A + n_B} \|\mu_A - \mu_B\|^2$$
  - **Définition des symboles :** $n_A, n_B$ représentent les effectifs des clusters $A$ et $B$, $\mu_A, \mu_B$ sont leurs centroïdes respectifs.
  - **Paramètres (appris) :** Matrice de liaisons hiérarchiques et regroupements.
  - **Hyperparamètres :** `n_clusters`, `linkage` (ward, complete, average, single), `metric`.
* **Avantages :**
  1. Fournit une visualisation hiérarchique très intuitive de la structure des données (dendrogramme).
  2. Ne nécessite pas d'imposer $K$ dès le départ (on peut couper l'arbre à n'importe quelle hauteur).
* **Limites :**
  1. Complexité temporelle et spatiale élevée en $\mathcal{O}(n^3)$ ou $\mathcal{O}(n^2)$, inutilisable sur de grands jeux de données.
  2. Une décision de fusion à une étape donnée ne peut jamais être annulée ultérieurement.

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le dendrogramme ci-dessous illustre les fusions ascendantes des observations et la ligne de coupe horizontale déterminant le nombre final de sous-groupes.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Gestion RH / Comportement) :** Établissement d'une typologie d'emplois ou de compétences en construisant des arbres de similarité fonctionnelle.
* **Cas d'usage 2 (Biologie / Logistique) :** Structuration de taxonomies de produits ou de zones géographiques de distribution.
* **Nature des données & Prérequis :** Variables numériques obligatoirement standardisées (`StandardScaler`).
* **Cas peu approprié :** Très volumineux jeux de données comprenant des dizaines de milliers de lignes.
* **Métriques adaptées :** Coefficient de Silhouette, Cofficient Cophenetique (fidélité de l'arbre).

* **Hyperparamètres définis :** `n_clusters=3` (critère de liaison par défaut : Ward).
* **Sortie obtenue :** Affiche la valeur de la Silhouette pour la partition obtenue.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Évalue la qualité des 3 sous-groupes fusionnés par le critère de Ward.
  - *Ne permet PAS de conclure :* Ne montre pas l'arbre d'assemblage hiérarchique (dendrogramme) qui permet d'identifier la hauteur de coupe optimale.

19. DBSCAN
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** DBSCAN / *Density-Based Spatial Clustering of Applications with Noise*.
* **Famille :** Apprentissage non supervisé — Regroupement fondé sur la densité.
* **Objectif :** Découvrir des clusters de formes arbitraires basés sur la densité locale de points tout en isolant automatiquement le bruit (valeurs atypiques).
* **Intuition & Fonctionnement :**
  L'algorithme parcourt les données et définit un cluster comme une zone continue de haute densité séparée par des zones de faible densité. Il classifie chaque point en point cœur, point frontière ou point de bruit.
  1. *Entrée :* Données $X$, rayon $\epsilon$ (*eps*) et nombre minimal de points *min_samples*.
  2. *Calcul/Apprentissage :* Recherche des voisinages de taille $\epsilon$. Si un point possède $\ge \text{min\_samples}$ voisins, un nouveau cluster se forme.
  3. *Sortie :* Attributions des clusters (avec l'étiquette -1 réservée au bruit).
* **Formule mathématique :**
  Voisinage $\epsilon$ d'un point $p$ :
  $$N_{\epsilon}(p) = \{q \in D \mid d(p, q) \le \epsilon\}$$
  Condition de point cœur : $|N_{\epsilon}(p)| \ge \text{min\_samples}$
  - **Définition des symboles :** $d(p, q)$ est la distance entre points, $\epsilon$ le rayon maximal, $\text{min\_samples}$ le seuil minimal de densité.
  - **Paramètres (appris) :** Groupes de densité et détection des points de bruit (-1).
  - **Hyperparamètres :** `eps` ($\epsilon$), `min_samples`, `metric`.
* **Avantages :**
  1. Ne nécessite pas de spécifier le nombre de clusters à l'avance.
  2. Capable de détecter des clusters de formes géométriques arbitraires et isole nativement les *outliers*.
* **Limites :**
  1. Sensible au choix des hyperparamètres `eps` et `min_samples`.
  2. Performant avec peine si les clusters du jeu de données présentent des densités très différentes.

### 2. Schéma explicatif

````python
from sklearn.cluster import DBSCAN

db_demo = DBSCAN(eps=0.8, min_samples=5).fit(X_clu)
labels = db_demo.labels_

plt.figure(figsize=(7, 4.5))
plt.scatter(X_clu[labels != -1, 0], X_clu[labels != -1, 1], c=labels[labels != -1], cmap='tab10', s=20, label='Clusters denses')
plt.scatter(X_clu[labels == -1, 0], X_clu[labels == -1, 1], c='black', marker='x', s=50, label='Bruit (Outliers: -1)')

plt.title('Schéma explicatif : Identification des clusters denses et du bruit avec DBSCAN')
plt.xlabel('X1')
plt.ylabel('X2')
plt.legend()
plt.grid(True, ls='--')
plt.tight_layout()
plt.show()
````

**Commentaire du schéma :** Le schéma illustre la notion de densité : les points cœurs (fort encadrement), les points frontières et les points isolés rejetés sous l'étiquette de bruit (-1).

### 3. Contextes d me utilisation
* **Cas d'usage 1 (Geomarketing / Logistique) :** Détection automatique de zones de forte concentration géographique de clients ou d'accidents.
* **Cas d'usage 2 (Finance / Traitement d'anomalies) :** Regroupement de transactions normales et isolement de comportements hors normes (bruit).
* **Nature des données & Prérequis :** Variables numériques obligatoirement standardisées (`StandardScaler`).
* **Cas peu approprié :** Données en haute dimension où la notion de distance dense s'estompe.
* **Métriques adaptées :** Coefficient de Silhouette (calculé hors points de bruit), Nombre de clusters identifiés, Taux de bruit.

### 4. Analyse de l'expérience Python
* **Entrées du code :** `DBSCAN(eps=.8, min_samples=5)` ajusté sur `X_clu`.
* **Hyperparamètres définis :** `eps=0.8`, `min_samples=5`.
* **Sortie obtenue :** Affiche les étiquettes de groupes identifiées et le nombre total de points considérés comme bruit (`-1`).
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Montre combien de clusters denses ont été construits et combien de points ont été isolés.
  - *Ne permet PAS de conclure :* Ne montre pas si augmenter ou réduire `eps` permettrait de rattacher le bruit sans fusionner des clusters distincts.

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
from sklearn.mixture import GaussianMixture
modele=GaussianMixture(n_components=3,random_state=GRAINE).fit(X_clu)
print('Poids:',modele.weights_,'Silhouette:',silhouette_score(X_clu,modele.predict(X_clu)))
````

### 1. Présentation
* **Nom (FR / EN) :** Modèles de mélanges gaussiens / *Gaussian Mixture Models (GMM)*.
* **Famille :** Apprentissage non supervisé — Clustering probabiliste / Modèles à variables latentes.
* **Objectif :** Représenter la distribution des données comme une superposition de plusieurs distributions gaussiennes multidimensionnelles.
* **Intuition & Fonctionnement :**
  Contrairement à K-means qui effectue une affectation stricte (*hard clustering*), GMM réalise une affectation probabiliste souple (*soft clustering*) via l'algorithme Espérance-Maximisation (EM).
  1. *Entrée :* Données $X$ et nombre de composantes $K$.
  2. *Calcul/Apprentissage :* Algorithme EM (Étape E : calcul des probabilités d'appartenance aux gaussiennes ; Étape M : mise à jour des moyennes, covariances et poids des gaussiennes).
  3. *Sortie :* Probabilités partielles qu'une donnée appartienne à chacune des $K$ composantes.
* **Formule mathématique :**
  Densité du mélange :
  $$p(x) = \sum_{k=1}^{K} \pi_k \mathcal{N}(x \mid \mu_k, \Sigma_k)$$
  - **Définition des symboles :** $\pi_k$ est le poids (proportion) de la $k$-ième gaussienne ($\sum \pi_k = 1$), $\mathcal{N}$ la loi normale multidimensionnelle de moyenne $\mu_k$ et de matrice de covariance $\Sigma_k$.
  - **Paramètres (appris) :** Poids $\pi_k$, moyennes $\mu_k$, matrices de covariance $\Sigma_k$.
  - **Hyperparamètres :** `n_components` ($K$), `covariance_type` (full, tied, diag, spherical).
* **Avantages :**
  1. Offre une grande flexibilité géométrique grâce à l'estimation des matrices de covariance (clusters ellipsoïdaux).
  2. Fournit une mesure de l'incertitude de l'appartenance aux sous-groupes (*soft clustering*).
* **Limites :**
  1. Sensible au choix des conditions initiales et aux maxima locaux de la vraisemblance.
  2. Risque d'instabilité numérique si une matrice de covariance devient singulière.

### 2. Schéma explicatif

````python
from sklearn.mixture import GaussianMixture

gmm_demo = GaussianMixture(n_components=3, random_state=42).fit(X_clu)
probs = gmm_demo.predict_proba(X_clu)

plt.figure(figsize=(7, 4.5))
plt.scatter(X_clu[:, 0], X_clu[:, 1], c=probs[:, 0], cmap='magma', s=20)
plt.colorbar(label='Probabilité d\'appartenance à la Composante 1')
plt.title('Schéma explicatif : Affectation probabiliste souple (Soft Clustering GMM)')
plt.xlabel('X1')
plt.ylabel('X2')
plt.grid(True, ls='--')
plt.tight_layout()
plt.show()
````

* **Commentaire du schéma :** Le schéma présente les ellipsoïdes de niveau de probabilité correspondant aux différentes composantes gaussiennes ajustées sur les données.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Régimes de marché) :** Identification des régimes financiers (ex: régime à faible volatilité vs régime à haute volatilité).
* **Cas d'usage 2 (Traitement du signal / Sécurité) :** Modélisation de distributions complexes de comportements d'utilisateurs.
* **Nature des données & Prérequis :** Variables numériques (mises à l'échelle recommandées).
* **Cas peu approprié :** Jeux de données avec des structures de formes complexes non convexes.
* **Métriques adaptées :** BIC (Critère d'information bayésien), AIC, Coefficient de Silhouette.

### 4. Analyse de l'expérience Python
* **Entrées du code :** `GaussianMixture(n_components=3, random_state=42)` ajusté sur `X_clu`.
* **Hyperparamètres définis :** `n_components=3`, `random_state=GRAINE`.
* **Sortie obtenue :** Affiche les poids appris des 3 gaussiennes ($\pi_k$) et le score de Silhouette des prédictions.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Donne la proportion relative de chaque composante et la qualité du découpage probabiliste.
  - *Ne permet PAS de conclure :* Ne fournit pas la valeur du critère BIC qui permet de choisir scientifiquement le nombre optimal de gaussiennes.

## 21. ACP

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
from sklearn.decomposition import PCA
X_vin=StandardScaler().fit_transform(load_wine().data)
modele=PCA(n_components=2).fit(X_vin)
projection=modele.transform(X_vin)
print('Variance expliquée:',modele.explained_variance_ratio_)
plt.scatter(projection[:,0],projection[:,1],s=10);plt.show()
````

### 1. Présentation
* **Nom (FR / EN) :** Analyse en composantes principales / *Principal Component Analysis (PCA)*.
* **Famille :** Apprentissage non supervisé — Réduction de dimension (méthode linéaire).
* **Objectif :** Réduire la dimension d'un tableau de données en projetant les observations sur un sous-espace linéaire de manière à conserver le maximum de variance (information).
* **Intuition & Fonctionnement :**
  L'ACP effectue une rotation des axes d'origine pour définir de nouvelles variables orthogonales et non corrélées (les composantes principales), ordonnées par quantité de variance expliquée.
  1. *Entrée :* Données $X$ centré-réduites.
  2. *Calcul/Apprentissage :* Calcul de la matrice de covariance, extraction des valeurs propres et des vecteurs propres (ou via SVD).
  3. *Sortie :* Matrice de données projetée dans le nouvel espace de dimension réduite $k < d$.
* **Formule mathématique :**
  Maximisation de la variance projetée sous contrainte d'orthonormalité :
  $$\max_{w} w^T \Sigma w \quad \text{sujet à} \quad w^T w = 1$$
  - **Définition des symboles :** $\Sigma$ est la matrice de covariance des données, $w$ le vecteur propre (axe principal).
  - **Paramètres (appris) :** Composantes principales (vecteurs propres) et variance associée (valeurs propres).
  - **Hyperparamètres :** `n_components` ($k$).
* **Avantages :**
  1. Technique fondamentale d'apprentissage pour éliminer la multicolinéarité et compresser les données.
  2. Permet la visualisation en 2D/3D de jeux de données complexes comportant des dizaines de colonnes.
* **Limites :**
  1. Méthode strictement linéaire : incapable de saisir des structures non linéaires complexes (manifolds).
  2. Les nouvelles composantes projetées sont des combinaisons linéaires parfois complexes à interpréter physiquement.

### 2. Schéma explicatif

````python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_wine

X_vin_scaled = StandardScaler().fit_transform(load_wine().data)
pca_demo = PCA().fit(X_vin_scaled)

plt.figure(figsize=(7, 4.5))
plt.plot(range(1, len(pca_demo.explained_variance_ratio_) + 1),
         np.cumsum(pca_demo.explained_variance_ratio_), marker='o', color='teal')
plt.axhline(0.8, color='r', linestyle='--', label='Seuil de 80% de variance')
plt.title('Schéma explicatif : Variance expliquée cumulée (Éboulis de l\'ACP)')
plt.xlabel('Nombre de composantes retenues')
plt.ylabel('Variance expliquée cumulée')
plt.legend()
plt.grid(True, ls='--')
plt.tight_layout()
plt.show()
````

**Commentaire du schéma :** Le schéma présente l'éboulis des valeurs propres qui illustre le pourcentage de variance expliquée retenu par chaque composante principale.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Analyse factorielle) :** Réduction de dizaines de ratios financiers en quelques facteurs synthétiques représentatifs.
* **Cas d'usage 2 (Traitement d'images / Compression) :** Compression d'images et réduction du bruit avant classification.
* **Nature des données & Prérequis :** Variables numériques obligatoirement centré-réduites (`StandardScaler`).
* **Cas peu approprié :** Données en structure de géométrie fortement non linéaire (ex: structure en spirale 3D).
* **Métriques adaptées :** Ratio de variance expliquée cumulé (*Explained Variance Ratio*).

### 4. Analyse de l'expérience Python
* **Entrées du code :** `PCA(n_components=2)` appliqué sur les données du vin `X_vin` prétraitées par `StandardScaler()`.
* **Hyperparamètres définis :** `n_components=2`.
* **Sortie obtenue :** Affiche le ratio de variance expliquée par les 2 premières composantes et le nuage projeté.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Indique la quantité exacte d'information conservée en passant à 2 dimensions sur le jeu du vin.
  - *Ne permet PAS de conclure :* Ne donne pas le cercle des corrélations (*loadings*) permettant d'identifier la contribution des variables initiales aux axes.

## 22. t-SNE

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
from sklearn.manifold import TSNE
X_vin=StandardScaler().fit_transform(load_wine().data)
projection=TSNE(n_components=2,perplexity=20,random_state=GRAINE).fit_transform(X_vin)
plt.scatter(projection[:,0],projection[:,1],s=10);plt.title('t-SNE');plt.show()
````

### 1. Présentation
* **Nom (FR / EN) :** Plongement voisin stochastique distribué selon une loi t / *t-Distributed Stochastic Neighbor Embedding (t-SNE)*.
* **Famille :** Apprentissage non supervisé — Réduction de dimension non linéaire (visualisation).
* **Objectif :** Projeter des données de haute dimension en 2D ou 3D en conservant prioritairement les relations de voisinage local entre observations.
* **Intuition & Fonctionnement :**
  t-SNE convertit les distances euclidiennes entre points en probabilités de proximité (loi gaussienne en haute dimension, loi de Student t en basse dimension pour éviter le problème d'entassement).
  1. *Entrée :* Données $X$ et paramètre de perplexité.
  2. *Calcul/Apprentissage :* Minimisation de la divergence de Kullback-Leibler (KL) entre les distributions de probabilités de l'espace d'origine et de l'espace réduit.
  3. *Sortie :* Coordonnées en 2D/3D optimisées uniquement pour la visualisation.
* **Formule mathématique :**
  Divergence KL à minimiser :
  $$KL(P \parallel Q) = \sum_{i} \sum_{j} p_{j|i} \log \frac{p_{j|i}}{q_{j|i}}$$
  - **Définition des symboles :** $p_{j|i}$ est la probabilité de proximité dans l'espace d'origine, $q_{j|i}$ la probabilité dans l'espace 2D réduit.
  - **Paramètres (appris) :** Coordonnées des points dans l'espace 2D/3D final.
  - **Hyperparamètres :** `perplexity`, `learning_rate`, `n_iter`, `random_state`.
* **Avantages :**
  1. Capable de révéler des structures visuelles de clusters complexes totalement invisibles avec l'ACP.
  2. Algorithme de visualisation non linéaire de référence.
* **Limites :**
  1. Ne conserve pas les distances globales (seuls les voisinages locaux ont un sens).
  2. Algorithme non déterministe, très coûteux en calcul et inutilisable pour transformer de nouvelles données de test (`transform` inexistant).

### 2. Schéma explicatif

````python
from sklearn.manifold import TSNE

tsne_demo = TSNE(n_components=2, perplexity=20, random_state=42).fit_transform(X_vin_scaled)

plt.figure(figsize=(7, 4.5))
plt.scatter(tsne_demo[:, 0], tsne_demo[:, 1], c=load_wine().target, cmap='Set1', s=25, alpha=0.8)
plt.title('Schéma explicatif : Projection visuelle non linéaire par t-SNE (Jeu du Vin)')
plt.xlabel('Dimension 1 t-SNE')
plt.ylabel('Dimension 2 t-SNE')
plt.grid(True, ls='--')
plt.tight_layout()
plt.show()
````

**Commentaire du schéma :** La figure illustre la projection 2D par t-SNE du jeu du vin, montrant la séparation visuelle nette des sous-groupes.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Détection de comportement) :** Exploration et inspection visuelle de profils de portefeuilles ou de fraudes suspectes.
* **Cas d'usage 2 (Bioinformatique) :** Visualisation de populations de cellules isolées (génomique).

## 23. UMAP

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
import umap
X_vin=StandardScaler().fit_transform(load_wine().data)
projection=umap.UMAP(n_components=2,n_neighbors=15,random_state=GRAINE).fit_transform(X_vin)
plt.scatter(projection[:,0],projection[:,1],s=10);plt.title('UMAP');plt.show()
````

### 1. Présentation
* **Nom (FR / EN) :** UMAP / *Uniform Manifold Approximation and Projection*.
* **Famille :** Apprentissage non supervisé — Réduction de dimension non linéaire (fondée sur la géométrie riemannienne).
* **Objectif :** Projeter des données de haute dimension dans un espace réduit tout en préservant mieux la structure globale de l'espace que t-SNE.
* **Intuition & Fonctionnement :**
  UMAP modélise la variété des données à l'aide de la géométrie floue des complexes simpliciaux, puis trouve une représentation en basse dimension dont le complexe équivalent est le plus similaire possible.
  1. *Entrée :* Données $X$ et hyperparamètre du nombre de voisins.
  2. *Calcul/Apprentissage :* Construction d'un graphe de voisinage flou, puis optimisation par descente de gradient de l'entropie croisée floue.
  3. *Sortie :* Projection en 2D/3D et possibilité de projeter de nouvelles observations (`transform`).
* **Formule mathématique :**
  Entropie croisée floue minimisée :
  $$C_{UMAP} = \sum_{i \neq j} \left[ p_{ij} \log \frac{p_{ij}}{q_{ij}} + (1 - p_{ij}) \log \frac{1 - p_{ij}}{1 - q_{ij}} \right]$$
  - **Définition des symboles :** $p_{ij}$ est la similarité floue dans l'espace d'origine, $q_{ij}$ la similarité dans l'espace de projection basse dimension.
  - **Paramètres (appris) :** Emplacements des points dans l'espace de dimension réduite.
  - **Hyperparamètres :** `n_neighbors`, `min_dist`, `n_components`, `metric`.
* **Avantages :**
  1. Plus rapide que t-SNE et préserve à la fois la structure locale et la structure globale.
  2. Dispose d'une méthode `.transform()` pour appliquer la projection sur de nouvelles observations.
* **Limites :**
  1. Repose sur une théorie mathématique abstraite complexe.
  2. Les hyperparamètres `n_neighbors` et `min_dist` influencent fortement l'aspect visuel des clusters.

### 2. Schéma explicatif

````python
import umap

umap_demo = umap.UMAP(n_components=2, n_neighbors=15, random_state=42).fit_transform(X_vin_scaled)

plt.figure(figsize=(7, 4.5))
plt.scatter(umap_demo[:, 0], umap_demo[:, 1], c=load_wine().target, cmap='Dark2', s=25, alpha=0.8)
plt.title('Schéma explicatif : Projection visuelle et préservation globale avec UMAP')
plt.xlabel('Dimension 1 UMAP')
plt.ylabel('Dimension 2 UMAP')
plt.grid(True, ls='--')
plt.tight_layout()
plt.show()
````

**Commentaire du schéma :** Le schéma montre la projection 2D produite par UMAP du jeu du vin, mettant en évidence la préservation de la séparation globale et locale.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Marketing / Segmentation) :** Visualisation et compression d'espaces d'embeddings d'utilisateurs ou d'articles en e-commerce.
* **Cas d'usage 2 (Traitement du signal / PNL) :** Projection de représentations textuelles (*word embeddings*) pour l'analyse thématique.
* **Nature des données & Prérequis :** Variables numériques obligatoirement standardisées (`StandardScaler`).
* **Cas peu approprié :** Applications financières exigeant une interprétabilité directe et linéaire des coefficients.
* **Métriques adaptées :** Conservation de la structure du graphe de voisinage (*Trustworthiness*).

### 4. Analyse de l'expérience Python
* **Entrées du code :** `umap.UMAP(n_components=2, n_neighbors=15, random_state=42)` appliqué sur `X_vin`.
* **Hyperparamètres définis :** `n_components=2`, `n_neighbors=15`, `random_state=GRAINE`.
* **Sortie obtenue :** Affiche le nuage de points 2D généré par UMAP.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Montre la capacité d'UMAP à compacter les clusters sur des données réelles.
  - *Ne permet PAS de conclure :* Ne montre pas l'effet d'une variation du paramètre `min_dist` sur la compacité interne des sous-groupes visualisés.

## 24. Isolation Forest

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
from sklearn.ensemble import IsolationForest
X_anomal=np.vstack([X_clu,np.array([[9,9],[-8,-8]])])
modele=IsolationForest(contamination=.02,random_state=GRAINE).fit(X_anomal)
print('Observations atypiques:',np.where(modele.predict(X_anomal)==-1)[0])
````

### 1. Présentation
* **Nom (FR / EN) :** Forêt d'isolement / *Isolation Forest*.
* **Famille :** Apprentissage non supervisé — Détection d'anomalies.
* **Objectif :** Isoler et détecter les observations atypiques (*outliers*) en exploitant leur faible densité et leur grande distance par rapport au reste des données.
* **Intuition & Fonctionnement :**
  Plutôt que de modéliser les points normaux, l'algorithme isole directement les anomalies. Les points anormaux nécessitent beaucoup moins de coupures aléatoires pour être isolés dans des feuilles d'arbres.
  1. *Entrée :* Données $X$ et taux de contamination estimé.
  2. *Calcul/Apprentissage :* Entraînement d'un ensemble d'arbres d'isolement aléatoires et mesure de la longueur moyenne du chemin d'accès aux points.
  3. *Sortie :* Score d'anomalie et étiquettes (1 pour normal, -1 pour anomalie).
* **Formule mathématique :**
  Score d'anomalie d'une observation $x$ :
  $$s(x, n) = 2^{-\frac{E(h(x))}{c(n)}}$$
  - **Définition des symboles :** $h(x)$ est la profondeur d'isolement du point dans un arbre, $E(h(x))$ la profondeur moyenne sur la forêt, $c(n)$ la profondeur moyenne théorique d'un arbre binaire de recherche.
  - **Paramètres (appris) :** Coupe des arbres d'isolement aléatoires.
  - **Hyperparamètres :** `contamination`, `n_estimators`, `max_samples`.
* **Avantages :**
  1. Extrêmement rapide et très sobre en mémoire ($\mathcal{O}(n \log n)$).
  2. Très performant sur de très grands jeux de données en haute dimension.
* **Limites :**
  1. Dépend du paramètre `contamination` réglé à la main si aucune donnée n'est étiquetée.
  2. Sensible aux découpages orthogonaux biaisés (résolu par Extended Isolation Forest).

### 2. Schéma explicatif

````python
from sklearn.ensemble import IsolationForest

X_anomal = np.vstack([X_clu, np.array([[9, 9], [-8, -8]])])
iso_demo = IsolationForest(contamination=0.02, random_state=42).fit(X_anomal)
preds_iso = iso_demo.predict(X_anomal)

plt.figure(figsize=(7, 4.5))
plt.scatter(X_anomal[preds_iso == 1, 0], X_anomal[preds_iso == 1, 1], c='blue', s=20, label='Données normales (1)')
plt.scatter(X_anomal[preds_iso == -1, 0], X_anomal[preds_iso == -1, 1], c='red', marker='D', s=80, label='Anomalies isolées (-1)')

plt.title('Schéma explicatif : Détection d\'anomalies par Isolation Forest')
plt.xlabel('X1')
plt.ylabel('X2')
plt.legend()
plt.grid(True, ls='--')
plt.tight_layout()
plt.show()
````

**Commentaire du schéma :** Le schéma montre que les anomalies isolées sur les bords atteignent une feuille très rapidement (chemin court), tandis que les points normaux denses nécessitent de nombreuses coupes (chemin long).

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Détection de fraude) :** Détection d'anomalies de transactions bancaires frauduleuses ou suspectes.
* **Cas d'usage 2 (Industrie / Maintenance) :** Détection de pannes ou dysfonctionnements de capteurs industriels en temps réel.
* **Nature des données & Prérequis :** Variables numériques. Aucune mise à l'échelle impérative requise.
* **Cas peu approprié :** Jeux de données avec une proportion d'anomalies excessivement élevée (> 30%).
* **Métriques adaptées :** Précision, Rappel, F1-Score sur anomalies (si étiquettes), Taux d'alerte, AUC-ROC.

### 4. Analyse de l'expérience Python
* **Entrées du code :** `IsolationForest(contamination=.02, random_state=42)` ajusté sur `X_anomal`.
* **Hyperparamètres définis :** `contamination=0.02`, `random_state=GRAINE`.
* **Sortie obtenue :** Affiche la liste des index d'observations identifiées comme atypiques (`-1`).
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Permet de localiser exactement quelles observations sont considérées comme anormales au seuil de 2%.
  - *Ne permet PAS de conclure :* Ne garantit pas que les points détectés soient de vraies fraudes/erreurs sans validation par un expert métier.

## 25. Bandits manchots

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
generateur=np.random.default_rng(GRAINE)
vraies_probas=np.array([.2,.5,.7]);somme=np.zeros(3);comptes=np.zeros(3)
for t in range(500):
    bras=t%3
    gain=int(generateur.random()<vraies_probas[bras]);somme[bras]+=gain;comptes[bras]+=1
print('Valeurs estimées:',somme/comptes)
````

### 1. Présentation
* **Nom (FR / EN) :** Bandits manchots / *Multi-Armed Bandits (MAB)*.
* **Famille :** Apprentissage par renforcement — Prise de décision sous incertitude.
* **Objectif :** Maximiser le gain cumulé en choisissant séquentiellement parmi plusieurs options (bras) aux récompenses inconnues, en gérant le compromis exploration/exploitation.
* **Intuition & Fonctionnement :**
  L'agent doit équilibrer l'exploration (tester des bras peu connus pour découvrir leur rentabilité) et l'exploitation (tirer le bras historiquement le plus rentable).
  1. *Entrée :* Nombre de bras $K$ et horizon temporel $T$.
  2. *Calcul/Apprentissage :* Sélection séquentielle d'un bras, observation de la récompense et mise à jour de la moyenne empirique du bras tiré.
  3. *Sortie :* Estimation de la valeur moyenne de chaque bras et allocation des tirages.
* **Formule mathématique :**
  Mise à jour empirique de la valeur d'un bras $a$ :
  $$Q(a) \leftarrow Q(a) + \frac{1}{N(a)} \left( R - Q(a) \right)$$
  - **Définition des symboles :** $Q(a)$ est l'estimation de la récompense moyenne du bras $a$, $N(a)$ le nombre de fois où le bras $a$ a été choisi, $R$ la récompense obtenue.
  - **Paramètres (appris) :** Valeurs estimées $Q(a)$ et compteurs $N(a)$.
  - **Hyperparamètres :** Horizon $T$, stratégie d'exploration.
* **Avantages :**
  1. Modèle très simple et efficace pour la prise de décision dynamique en ligne.
  2. Réduit drastiquement le coût d'opportunité par rapport aux tests A/B classiques statiques.
* **Limites :**
  1. Ne prend pas en compte l'état de l'environnement (pas d'état contextuel dans sa forme de base).
  2. Suppose que la distribution des récompenses des bras est stationnaire dans le temps.

### 2. Schéma explicatif

````python
generateur = np.random.default_rng(42)
vraies_probas = np.array([.2, .5, .7])
somme = np.zeros(3)
comptes = np.zeros(3)
gains_cumules = []

for t in range(500):
    bras = t % 3
    gain = int(generateur.random() < vraies_probas[bras])
    somme[bras] += gain
    comptes[bras] += 1
    gains_cumules.append(somme.sum())

plt.figure(figsize=(7, 4))
plt.plot(gains_cumules, color='green', label='Récompense cumulée')
plt.title('Schéma explicatif : Accumulation de gains sur un bandit multi-bras')
plt.xlabel('Tirages (Temps t)')
plt.ylabel('Gain total')
plt.legend()
plt.grid(True, ls='--')
plt.tight_layout()
plt.show()
````

**Commentaire du schéma :** Le schéma montre les distributions des gains associés aux différents bras et la convergence progressive des sélections vers le meilleur bras.

### 3. Contextes d me utilisation
* **Cas d'usage 1 (Marketing / E-commerce) :** Optimisation dynamique des affichages publicitaires ou des prix sur un site marchand (remplacement des tests A/B).
* **Cas d'usage 2 (Finance / Allotement) :** Allocation dynamique d'ordres d'exécution entre plusieurs plateformes de négociation.
* **Nature des données & Prérequis :** Flux de données séquentiel interactif en temps réel.
* **Cas peu approprié :** Problèmes où les actions actuelles modifient l'état futur de l'environnement de manière complexe.
* **Métriques adaptées :** Regret cumulé (*Cumulative Regret*), Récompense cumulée.

### 4. Analyse de l'expérience Python
* **Entrées du code :** Simulation sur 500 étapes avec 3 bras aux probabilités réelles `[0.2, 0.5, 0.7]`.
* **Hyperparamètres définis :** Horizon de 500 itérations, sélection par tour de rôle (Round-Robin simple `t % 3`).
* **Sortie obtenue :** Affiche la valeur moyenne estimée pour chaque bras après 500 tirages.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Montre que la loi des grands nombres permet d'estimer correctement la valeur théorique des 3 bras.
  - *Ne permet PAS de conclure :* Ne constitue pas un algorithme intelligent d'optimisation (le Round-Robin explore indéfiniment sans exploiter le meilleur bras).

## 26. Epsilon-greedy

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
generateur=np.random.default_rng(GRAINE)
vraies_probas=np.array([.2,.5,.7]);somme=np.zeros(3);comptes=np.zeros(3);epsilon=.1
for t in range(1000):
    bras=int(generateur.integers(3)) if generateur.random()<epsilon else int(np.argmax((somme+1)/(comptes+1)))
    gain=int(generateur.random()<vraies_probas[bras]);somme[bras]+=gain;comptes[bras]+=1
print('Nombre de choix:',comptes,'Valeurs estimées:',somme/np.maximum(comptes,1))
````

### 1. Présentation
* **Nom (FR / EN) :** Algorithme Epsilon-avide / *Epsilon-Greedy Algorithm*.
* **Famille :** Apprentissage par renforcement — Stratégie d'exploration pour bandits et RL.
* **Objectif :** Résoudre concrètement le compromis exploration/exploitation en choisissant une action aléatoire avec une probabilité $\epsilon$ et la meilleure action connue avec une probabilité $1-\epsilon$.
* **Intuition & Fonctionnement :**
  L'agent est la plupart du temps "avide" (il choisit le bras qui a le meilleur historique), mais réserve une petite fraction du temps ($\epsilon$) à l'exploration pure pour vérifier si un autre bras ne serait pas supérieur.
  1. *Entrée :* Paramètre d'exploration $\epsilon \in [0, 1]$ et estimations initiales.
  2. *Calcul/Apprentissage :* Tirage d'un nombre aléatoire $u \in [0, 1]$. Si $u < \epsilon$, action aléatoire ; sinon action $\arg\max Q(a)$.
  3. *Sortie :* Action choisie et mise à jour de la valeur estimée $Q(a)$.
* **Formule mathématique :**
  Politique de sélection d'action :
  $$\pi(a) = \begin{cases} 1 - \epsilon + \frac{\epsilon}{|A|} & \text{si } a = \arg\max Q(a) \\ \frac{\epsilon}{|A|} & \text{si } a \neq \arg\max Q(a) \end{cases}$$
  - **Définition des symboles :** $\epsilon$ est le taux d'exploration, $|A|$ le nombre total d'actions possibles, $Q(a)$ la valeur estimée de l'action $a$.
  - **Paramètres (appris) :** Valeurs d'actions $Q(a)$ et compteurs d'exécutions.
  - **Hyperparamètres :** $\epsilon$ (fixe ou décroissant).
* **Avantages :**
  1. Stratégie extrêmement simple, intuitive et d'une grande efficacité pratique.
  2. Assure qu'aucune action n'est totalement ignorée sur le long terme.
* **Limites :**
  1. Lors de l'exploration, l'agent choisit de manière indifférenciée une très mauvaise action ou une action presque optimale.
  2. Nécessite d'ajuster le paramètre $\epsilon$ (ou de programmer sa décroissance).

### 2. Schéma explicatif

````python
epsilon = 0.1
labels = ['Exploitation Avide (1 - ε)', 'Exploration Aléatoire (ε)']
sizes = [1 - epsilon, epsilon]

plt.figure(figsize=(6, 4))
plt.pie(sizes, labels=labels, autopct='%1.0f%%', colors=['skyblue', 'orange'], startangle=140)
plt.title('Schéma explicatif : Répartition du choix d\'action (Epsilon-Greedy avec ε=10%)')
plt.tight_layout()
plt.show()
````

**Commentaire du schéma :** Le diagramme circulaire schématise la répartition Epsilon-Greedy : $(1-\epsilon)$ réservé à l'exploitation avide et $\epsilon$ dédié à l'exploration équirégulière.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Trading) :** Exploration de nouvelles stratégies d'arbitrage tout en exécutant majoritairement la meilleure stratégie connue.
* **Cas d'usage 2 (Jeux / IA) :** Stratégie de contrôle dans le cadre de l'apprentissage d'agents sur des environnements simulés (ex: Gymnasium).
* **Nature des données & Prérequis :** Interactions séquentielles en ligne avec système de récompenses.
* **Cas peu approprié :** Environnements où une action d'exploration aléatoire peut provoquer un dommage critique ou mortel (ex: conduite autonome).
* **Métriques adaptées :** Récompense cumulée, Taux d'utilisation du meilleur bras (*Optimal Action %*).

### 4. Analyse de l'expérience Python
* **Entrées du code :** Simulation de 1000 tirages avec $\epsilon=0.1$ et probabilités réelles `[0.2, 0.5, 0.7]`.
* **Hyperparamètres définis :** `epsilon=0.1`, 1000 itérations.
* **Sortie obtenue :** Affiche le nombre de choix par bras et les valeurs estimées finales.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Montre que le bras le plus rentable (index 2 avec probabilité 0.7) a été sélectionné de manière très largement majoritaire.
  - *Ne permet PAS de conclure :* Ne compare pas cette performance avec une stratégie UCB (*Upper Confidence Bound*) qui explore plus intelligemment.

## 27. Q-learning

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False)
Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode)
    for pas in range(100):
        action=int(generateur.integers(env.action_space.n)) if generateur.random()<max(.05,1-episode/2500) else int(np.argmax(Q[etat]))
        suivant,recompense,termine,tronque,_=env.step(action)
        cible=recompense+(0 if termine else .95*np.max(Q[suivant]))
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat=suivant
        if termine or tronque:break
print('Politique:',Q.argmax(axis=1));env.close()
````

Python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import Ridge
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error

# Configuration du style graphique
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (12, 6)

# ==========================================
# 1. TÉLÉCHARGEMENT DES DONNÉES DEPUIS FRED
# ==========================================
print("Téléchargement des séries temporelles...")
url_cpi = "https://fred.stlouisfed.org/graph/fredgraph.csv?id=CPIAUCSL"      # Indice des prix à la consommation
url_unrate = "https://fred.stlouisfed.org/graph/fredgraph.csv?id=UNRATE"    # Taux de chômage
url_fedfunds = "https://fred.stlouisfed.org/graph/fredgraph.csv?id=FEDFUNDS"# Taux directeur

df_cpi = pd.read_csv(url_cpi, parse_dates=['DATE'], index_col='DATE')
df_unrate = pd.read_csv(url_unrate, parse_dates=['DATE'], index_col='DATE')
df_fedfunds = pd.read_csv(url_fedfunds, parse_dates=['DATE'], index_col='DATE')

# Fusion des données sur la date
df = df_cpi.join([df_unrate, df_fedfunds], how='inner')
df.columns = ['CPI', 'UNRATE', 'FEDFUNDS']

# Conversion en numérique (gestion des éventuelles valeurs 'NaN' ou '.')
df = df.apply(pd.to_numeric, errors='coerce').dropna()

# ==========================================
# 2. INGÉNIERIE DES VARIABLES (FEATURE ENG.)
# ==========================================
# Calcul de l'inflation annuelle : % de variation sur 12 mois (en %)
df['INFLATION'] = df['CPI'].pct_change(12) * 100

# Création des variables retardées (Lags) pour éviter le look-ahead bias
for lag in [1, 2, 3, 6, 12]:
    df[f'INFLATION_lag_{lag}'] = df['INFLATION'].shift(lag)
    df[f'UNRATE_lag_{lag}'] = df['UNRATE'].shift(lag)
    df[f'FEDFUNDS_lag_{lag}'] = df['FEDFUNDS'].shift(lag)

# Définition de la cible (Y) : Inflation dans 12 mois
df['TARGET_12M'] = df['INFLATION'].shift(-12)

# Suppression des lignes contenant des NaN suite aux décalages
df_clean = df.dropna().copy()

# ==========================================
# 3. DÉCOUPAGE CHRONOLOGIQUE (TRAIN / TEST)
# ==========================================
# Variables explicatives (X) et Cible (y)
features = [col for col in df_clean.columns if 'lag' in col]
X = df_clean[features]
y = df_clean['TARGET_12M']

# Découpage temporel : 80% Entraînement, 20% Test (Pas de train_test_split aléatoire !)
split_idx = int(len(df_clean) * 0.8)
X_train, X_test = X.iloc[:split_idx], X.iloc[split_idx:]
y_train, y_test = y.iloc[:split_idx], y.iloc[split_idx:]

print(f"Période d'entraînement : {X_train.index.min().strftime('%Y-%m')} à {X_train.index.max().strftime('%Y-%m')}")
print(f"Période de test         : {X_test.index.min().strftime('%Y-%m')} à {X_test.index.max().strftime('%Y-%m')}")

# ==========================================
# 4. MODÉLISATION ET ÉVALUATION
# ==========================================
# Référence (Naive Baseline) : La valeur actuelle de l'inflation
y_pred_baseline = X_test['INFLATION_lag_1']

# Modèle 1 : Régression Ridge
model_ridge = Ridge(alpha=1.0)
model_ridge.fit(X_train, y_train)
y_pred_ridge = model_ridge.predict(X_test)

# Modèle 2 : Gradient Boosting
model_gb = GradientBoostingRegressor(n_estimators=100, random_state=42)
model_gb.fit(X_train, y_train)
y_pred_gb = model_gb.predict(X_test)

# Calcul des métriques RMSE
rmse_base = np.sqrt(mean_squared_error(y_test, y_pred_baseline))
rmse_ridge = np.sqrt(mean_squared_error(y_test, y_pred_ridge))
rmse_gb = np.sqrt(mean_squared_error(y_test, y_pred_gb))

print("\n--- RÉSULTATS (RMSE) ---")
print(f"Référence (Naïve) : {rmse_base:.3f}")
print(f"Régression Ridge  : {rmse_ridge:.3f}")
print(f"Gradient Boosting : {rmse_gb:.3f}")

# ==========================================
# 5. VISUALISATION
# ==========================================
plt.figure(figsize=(12, 5))
plt.plot(y_test.index, y_test, label='Inflation Réelle (+12 mois)', color='black', linewidth=2)
plt.plot(y_test.index, y_pred_baseline, label='Référence Naïve', linestyle='--', color='gray', alpha=0.7)
plt.plot(y_test.index, y_pred_ridge, label='Prédiction Ridge', color='blue', alpha=0.8)
plt.plot(y_test.index, y_pred_gb, label='Prédiction Gradient Boosting', color='green', alpha=0.8)

plt.title("Prévision de l'Inflation US à un horizon de 12 mois (Jeu de Test)")
plt.xlabel("Date")
plt.ylabel("Taux d'inflation (%)")
plt.legend()
plt.tight_layout()
plt.show()

import matplotlib.pyplot as plt
import numpy as np

# Schéma explicatif de la Régression Linéaire
np.random.seed(42)
x_demo = np.linspace(0, 10, 30)
y_demo = 2 * x_demo + 1 + np.random.normal(0, 2, 30)

plt.figure(figsize=(7, 4))
plt.scatter(
    x_demo, y_demo, color="blue", label="Données réelles (Observations)"
)
plt.plot(
    x_demo, 2 * x_demo + 1, color="red", linewidth=2, label="Droite MCO (Modèle)"
)

# Tracer un exemple de résidu
plt.vlines(
    x_demo[15],
    y_demo[15],
    2 * x_demo[15] + 1,
    color="gray",
    linestyle="--",
    label="Résidu (Écart e_i)",
)

plt.title("Principe de la Régression Linéaire (Moindres Carrés)")
plt.xlabel("Variable explicative X")
plt.ylabel("Cible continue y")
plt.legend()
plt.tight_layout()
plt.show()
Commentaire du schéma : La figure illustre le nuage de points d'observation (en bleu) et la droite de régression optimale (en rouge) qui minimise la distance verticale globale (les résidus en pointillés gris) entre la droite et chaque point.

Markdown
### 1. Présentation
* **Nom (FR / EN) :** Régression linéaire / *Linear Regression*.
* **Famille :** Apprentissage supervisé — Régression.
* **Objectif :** Prédire une variable cible continue $y$ à partir d'une combinaison linéaire de variables explicatives $X$.
* **Intuition & Fonctionnement :**
  L'algorithme cherche à ajuster la meilleure droite (ou hyperplan en multi-dimensions) passant à travers le nuage de points de données.
  1. *Entrée :* Matrice de données $X \in \mathbb{R}^{n \times p}$ et vecteur cible $y \in \mathbb{R}^n$.
  2. *Calcul/Apprentissage :* Minimisation de la somme des carrés des écarts (Moindres Carrés Ordinaires - MCO) entre la cible réelle et la valeur prédite.
  3. *Sortie :* Une fonction de prédiction $f(X) = \hat{y}$ qui attribue une valeur numérique continue à toute nouvelle observation.
* **Formule mathématique :**
  $$\hat{y} = \beta_0 + \sum_{j=1}^{p} \beta_j x_j = X\beta$$
  Fonction de perte (Moindres Carrés Ordinaires) :
  $$\mathcal{L}(\beta) = \frac{1}{2n} \|y - X\beta\|^2_2 = \frac{1}{2n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$
  - **Définition des symboles :** $y_i$ est la vraie valeur, $\hat{y}_i$ la valeur prédite, $x_j$ les variables explicatives, et $n$ le nombre d'observations.
  - **Paramètres (appris) :** Coefficients d'onde/pentes $\beta_j$ et biais/ordonnée à l'origine $\beta_0$.
  - **Hyperparamètres :** Aucun dans la version analytique standard des MCO (solution fermée $\beta = (X^T X)^{-1} X^T y$).
* **Avantages :**
  1. Très rapide à calculer et très sobre en ressources calculatoires.
  2. Modèle entièrement interprétable : chaque coefficient $\beta_j$ indique la variation exacte de $y$ pour une augmentation d'une unité de $x_j$.
* **Limites :**
  1. Hypothèse forte de linéarité entre les variables d'entrée et la cible.
  2. Extrêmement sensible aux valeurs extrêmes (*outliers*) et à la multicolinéarité entre les variables $X$.

---

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** La figure illustre le nuage de points d'observation (en bleu) et la droite de régression optimale (en rouge) qui minimise la distance verticale globale (les résidus en pointillés gris) entre la droite et chaque point d'observation.

---

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Logistique) :** Prévision du chiffre d'affaires mensuel d'une entreprise en fonction des dépenses publicitaires ou estimation des coûts de transport en fonction du poids et de la distance.
* **Cas d'usage 2 :** Estimation de la consommation d'énergie d'un bâtiment selon la température extérieure.
* **Nature des données & Prérequis :** Variables explicatives numériques continues ou catégorielles encodées. Nécessite une absence de multicolinéarité parfaite.
* **Cas peu approprié :** Prédiction de séries temporelles très non linéaires (ex: cours d'une crypto-monnaie) ou classification (la cible est une catégorie et non une quantité).
* **Métriques adaptées :** MAE (Erreur Absolue Moyenne), RMSE (Racine de l'Erreur Quadratique Moyenne) et $R^2$ (Coefficient de détermination).

---

2. Ridge
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Régression Ridge / *Ridge Regression (L2 Regularization)*.
* **Famille :** Apprentissage supervisé — Régression avec régularisation.
* **Objectif :** Prédire une variable continue $y$ en pénalisant les grands coefficients pour éviter le surapprentissage et traiter la multicolinéarité.
* **Intuition & Fonctionnement :**
  L'algorithme modifie les Moindres Carrés Ordinaires en ajoutant une pénalité sur la somme des carrés des coefficients (norme $L_2$).
  1. *Entrée :* Données $X$ (préalablement standardisées) et cible $y$.
  2. *Calcul/Apprentissage :* Minimisation de la perte MCO + pénalité $L_2$. Les coefficients sont contractés (*shrinkage*) vers zéro sans jamais devenir exactement nuls.
  3. *Sortie :* Vecteur de prédictions $\hat{y}$ plus stable et moins sensible au bruit des données d'entraînement.
* **Formule mathématique :**
  $$\mathcal{L}_{Ridge}(\beta) = \frac{1}{2n} \|y - X\beta\|^2_2 + \alpha \|\beta\|^2_2 = \frac{1}{2n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 + \alpha \sum_{j=1}^{p} \beta_j^2$$
  - **Définition des symboles :** $y_i$ est la valeur réelle, $\hat{y}_i$ la prédiction, $\beta_j$ les coefficients, $n$ le nombre d'échantillons, $\alpha$ le coefficient de pénalité.
  - **Paramètres (appris) :** Coefficients $\beta$.
  - **Hyperparamètres :** $\alpha \ge 0$ (intensité de la régularisation).
* **Avantages :**
  1. Très performant en présence de forte multicolinéarité entre variables.
  2. Réduit efficacement la variance du modèle sans augmenter excessivement le biais.
* **Limites :**
  1. Ne réalise pas de sélection de variables (aucun coefficient n'est annulé).
  2. Nécessite obligatoirement une standardisation préalable des variables.

---

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le graphique montre l'effet de contraction (*shrinkage*) : à mesure que l'hyperparamètre $\alpha$ augmente, la magnitude des coefficients $\beta$ diminue vers zéro, lissant la fonction de prédiction.

---

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Économétrie) :** Modélisation des rendements boursiers où de nombreux indicateurs macroéconomiques sont fortement corrélés entre eux.
* **Cas d'usage 2 :** Estimation de prix immobiliers avec des dizaines de caractéristiques corrélées (surface, nombre de pièces, année).
* **Nature des données & Prérequis :** Variables numériques obligatoirement standardisées (`StandardScaler`).
* **Cas peu approprié :** Jeux de données avec un très grand nombre de variables non pertinentes où une sélection de variables est nécessaire.
* **Métriques adaptées :** MAE, RMSE, $R^2$.

---

### 4. Analyse de l'expérience Python
* **Entrées du code :** `Xr_ent` et `yr_ent` passés dans un `Pipeline` combinant `StandardScaler()` et `Ridge(alpha=1.0)`.
* **Hyperparamètres définis :** `alpha=1.0`.
* **Sortie obtenue :** Affiche la MAE calculée sur `Xr_test`.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Valide l'intégration de la standardisation et du calcul du score sur les données de test synthétiques.
  - *Ne permet PAS de conclure :* Ne permet pas de savoir si `alpha=1.0` est la valeur optimale (nécessiterait une validation croisée `RidgeCV`).
2. Bloc Code Python pour le schéma (à copier dans une cellule Code)
Python
import matplotlib.pyplot as plt
import numpy as np

alphas = np.logspace(-2, 3, 50)
coefs = []

# Génération de données synthétiques corrélées
np.random.seed(42)
X_dummy = np.random.randn(50, 5)
y_dummy = X_dummy @ np.array([3, -2, 1.5, 0, 0.5]) + np.random.randn(50)

from sklearn.linear_model import Ridge
from sklearn.preprocessing import StandardScaler

X_scaled = StandardScaler().fit_transform(X_dummy)
for a in alphas:
    ridge = Ridge(alpha=a)
    ridge.fit(X_scaled, y_dummy)
    coefs.append(ridge.coef_)

plt.figure(figsize=(8, 4.5))
plt.plot(alphas, coefs)
plt.xscale('log')
plt.xlabel('Hyperparamètre alpha (Intensité de régularisation L2)')
plt.ylabel('Coefficients beta')
plt.title('Schéma explicatif : Contraction des coefficients par la régression Ridge')
plt.grid(True, which="both", ls="--")
plt.tight_layout()
plt.show()
3. Lasso
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Lasso / *Lasso Regression (L1 Regularization - Least Absolute Shrinkage and Selection Operator)*.
* **Famille :** Apprentissage supervisé — Régression avec sélection de variables.
* **Objectif :** Prédire une variable continue tout en sélectionnant automatiquement les variables les plus pertinentes en annulant les coefficients inutiles.
* **Intuition & Fonctionnement :**
  Lasso ajoute une pénalité sur la somme des valeurs absolues des coefficients (norme $L_1$).
  1. *Entrée :* Données $X$ standardisées et cible $y$.
  2. *Calcul/Apprentissage :* Minimisation de la perte MCO + pénalité $L_1$. En raison de la géométrie de la norme $L_1$ (coins pointus), certains coefficients deviennent exactement égaux à zéro.
  3. *Sortie :* Modèle parcimonieux (*sparse*) utilisant uniquement un sous-ensemble des variables d'origine.
* **Formule mathématique :**
  $$\mathcal{L}_{Lasso}(\beta) = \frac{1}{2n} \|y - X\beta\|^2_2 + \alpha \|\beta\|_1 = \frac{1}{2n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 + \alpha \sum_{j=1}^{p} |\beta_j|$$
  - **Définition des symboles :** $y_i$ valeur réelle, $\hat{y}_i$ prédiction, $\beta_j$ coefficients, $\alpha$ intensité de la pénalité $L_1$.
  - **Paramètres (appris) :** Coefficients $\beta$.
  - **Hyperparamètres :** $\alpha \ge 0$, `max_iter` (nombre maximal d'itérations du solveur).
* **Avantages :**
  1. Réalise une sélection automatique de caractéristiques (*feature selection*).
  2. Rend le modèle final très lisible et réduit le bruit en éliminant les variables non pertinentes.
* **Limites :**
  1. Si deux variables sont très corrélées, Lasso en choisit une arbitrairement et met l'autre à zéro.
  2. Si $p > n$ (plus de variables que d'observations), Lasso ne peut sélectionner au maximum que $n$ variables.

---

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le graphique illustre l'annulation des coefficients : lorsque $\alpha$ augmente, les trajectoires de plusieurs coefficients atteignent la ligne zéro (sélection automatique des variables).

---

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Marketing / Scoring) :** Sélection des variables prédictives parmi des centaines de métriques de comportement client en ligne pour modéliser le panier moyen.
* **Cas d'usage 2 (Génomique / Bioinformatique) :** Identification des quelques gènes responsables d'une maladie parmi des milliers de gènes analysés.
* **Nature des données & Prérequis :** Variables numériques obligatoirement standardisées (`StandardScaler`).
* **Cas peu approprié :** Situations où toutes les variables apportent une petite contribution utile et corrélée (préérer Ridge).
* **Métriques adaptées :** $R^2$, MAE, RMSE.

---

### 4. Analyse de l'expérience Python
* **Entrées du code :** `Xr_ent` et `yr_ent` dans un `Pipeline` avec `StandardScaler()` et `Lasso(alpha=.1, max_iter=10000)`.
* **Hyperparamètres définis :** `alpha=0.1`, `max_iter=10000`.
* **Sortie obtenue :** Affiche le coefficient de détermination ($R^2$).
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Mesure la proportion de variance expliquée par le modèle Lasso sur les données de test.
  - *Ne permet PAS de conclure :* Ne montre pas explicitement quelles variables ont été éliminées (coefficients mis à zéro).

4. Régression polynomiale
1. Bloc Markdown (à copier dans me cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Régression polynomiale / *Polynomial Regression*.
* **Famille :** Apprentissage supervisé — Régression non linéaire.
* **Objectif :** Capter des relations non linéaires entre les variables explicatives et la cible en générant des termes polynomiaux et d'interaction.
* **Intuition & Fonctionnement :**
  L'algorithme transforme l'espace des données de départ en créant des combinaisons de puissances ($x^2, x^3, ...$) et de produits croisés ($x_1 x_2$), puis applique une régression linéaire sur cet espace élargi.
  1. *Entrée :* Variables $X$.
  2. *Calcul/Apprentissage :* Transformation via `PolynomialFeatures(degree=d)` suivie de l'ajustement d'un modèle linéaire (souvent régularisé avec Ridge/Lasso).
  3. *Sortie :* Courbe ou surface d'ajustement non linéaire.
* **Formule mathématique :**
  $$\hat{y} = \beta_0 + \beta_1 x + \beta_2 x^2 + \dots + \beta_d x^d$$
  - **Définition des symboles :** $d$ est le degré du polynôme, $\beta_k$ sont les coefficients associés aux puissances $x^k$.
  - **Paramètres (appris) :** Coefficients $\beta$.
  - **Hyperparamètres :** Degré $d$ (`degree`), inclusion du terme de biais (`include_bias`).
* **Avantages :**
  1. Permet de modéliser des courbes et des tendances complexes sans changer de framework linéaire.
  2. Conserve une résolution analytique directe via les MCO/Ridge.
* **Limites :**
  1. Explosion combinatoire du nombre de caractéristiques quand le nombre de variables et le degré augmentent.
  2. Risque élevé de surapprentissage (*overfitting*) sur les bords du domaine de données.

---

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le graphique compare l'ajustement linéaire classique (droit) et l'ajustement polynomial (courbe), montrant comment le polynôme capture la courbure naturelle des données.

---

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Modélisation de rendements) :** Modélisation de la courbe des taux d'intérêt ou prévision du coût marginal de production selon le volume (économies d'échelle puis déséconomies d'échelle).
* **Cas d'usage 2 :** Estimation de la trajectoire de croissance des ventes d'un nouveau produit après son lancement (courbe en S ou parabole).
* **Nature des données & Prérequis :** Variables numériques. La régularisation (Ridge/Lasso) est fortement recommandée.
* **Cas peu approprié :** Problèmes comportant un très grand nombre de variables d'entrée (ex: 100 variables à un degré 3 créent des centaines de milliers de colonnes).
* **Métriques adaptées :** MAE, RMSE, $R^2$.

---

### 4. Analyse de l'expérience Python
* **Entrées du code :** `Xr_ent` et `yr_ent` dans un pipeline composé de `StandardScaler()`, `PolynomialFeatures(degree=2, include_bias=False)` et `Ridge(alpha=10)`.
* **Hyperparamètres définis :** `degree=2`, `alpha=10`.
* **Sortie obtenue :** Affiche la MAE sur le jeu de test.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Montre le score de l'intégration polynomiale de degré 2 régularisée sur le jeu de test.
  - *Ne permet PAS de conclure :* Ne montre pas si un degré plus élevé (ex: 3 ou 4) améliorerait le résultat ou provoquerait du surapprentissage.
2. Bloc Code Python pour le schéma (à copier dans une cellule Code)
Python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures

np.random.seed(42)
x_poly = np.linspace(-3, 3, 40)
y_poly = 0.5 * x_poly**2 - x_poly + 1 + np.random.normal(0, 1, 40)

pipe_poly = make_pipeline(
    PolynomialFeatures(degree=2), Ridge(alpha=0.1)
)
pipe_poly.fit(x_poly.reshape(-1, 1), y_poly)

plt.figure(figsize=(8, 4.5))
plt.scatter(x_poly, y_poly, color='blue', label='Données réelles')
plt.plot(
    x_poly,
    pipe_poly.predict(x_poly.reshape(-1, 1)),
    color='red',
    linewidth=2,
    label='Ajustement Polynomial (Degré 2)',
)
plt.title('Schéma explicatif : Régression Polynomiale de degré 2')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
plt.tight_layout()
plt.show()
5. Arbre de régression
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Arbre de décision pour la régression / *Decision Tree Regressor*.
* **Famille :** Apprentissage supervisé — Arbres de décision / Non paramétrique.
* **Objectif :** Prédire une cible continue en découpant l'espace des variables en régions rectangulaires disjointes.
* **Intuition & Fonctionnement :**
  L'algorithme sépare récursivement les données selon des seuils sur les variables qui réduisent au maximum la variance de la cible dans chaque sous-groupe (feuille).
  1. *Entrée :* Données $X$ et cible $y$.
  2. *Calcul/Apprentissage :* À chaque nœud, recherche de la meilleure variable $j$ et du meilleur seuil $t$ pour minimiser la somme des erreurs quadratiques des deux sous-groupes créés.
  3. *Sortie :* La prédiction pour une feuille est la moyenne des valeurs cibles des observations qui s'y trouvent.
* **Formule mathématique :**
  Critère de séparation (minimisation de la variance) :
  $$\text{Impureté}(R) = \frac{1}{|R|} \sum_{i \in R} (y_i - \bar{y}_R)^2$$
  - **Définition des symboles :** $R$ est la région (nœud), $|R|$ le nombre d'échantillons dans la région, $\bar{y}_R$ la moyenne des valeurs $y$ dans cette région.
  - **Paramètres (appris) :** Structure de l'arbre, variables de coupure, seuils et valeurs des feuilles $\bar{y}_R$.
  - **Hyperparamètres :** `max_depth` (profondeur max), `min_samples_split`, `min_samples_leaf`.
* **Avantages :**
  1. Modèle intuitif et visuellement interprétable sous forme de diagramme décisionnel.
  2. Ne nécessite aucune mise à l'échelle des données et gère les relations non linéaires complexes.
* **Limites :**
  1. Risque très élevé de surapprentissage si la profondeur n'est pas restreinte.
  2. Prédictions constantes par morceaux (incapable d'extrapoler en dehors de la plage d'entraînement).

---

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le graphique montre comment un arbre de régression crée une fonction en escalier (prédictions constantes par morceaux dans chaque feuille).

---

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Gestion / Immobilier) :** Estimation de la valeur d'un actif immobilier avec des règles simples et explicables aux clients (ex: "Si surface > 80m² et pièces > 3 alors valeur = X").
* **Cas d'usage 2 (Logistique) :** Estimation du temps de livraison en fonction du secteur et du créneau horaire.
* **Nature des données & Prérequis :** Variables numériques ou catégorielles encodées. Aucune standardisation requise.
* **Cas peu approprié :** Données nécessitant une extrapolation continue précise en dehors des bornes d'entraînement.
* **Métriques adaptées :** MAE, RMSE, $R^2$.

---

### 4. Analyse de l'expérience Python
* **Entrées du code :** `DecisionTreeRegressor(max_depth=5, random_state=42)` entraîné sur `Xr_ent` et `yr_ent`.
* **Hyperparamètres définis :** `max_depth=5`, `random_state=GRAINE`.
* **Sortie obtenue :** Affiche la MAE sur l'échantillon de test.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Mesure l'erreur d'un arbre limité à une profondeur de 5.
  - *Ne permet PAS de conclure :* Ne permet pas d'évaluer le surapprentissage sans comparer ce résultat au score d'entraînement (`Xr_ent`).

6. Forêt aléatoire — régression
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Forêt aléatoire pour la régression / *Random Forest Regressor*.
* **Famille :** Apprentissage supervisé — Méthodes d'ensemble (*Bagging* / Agrégation).
* **Objectif :** Améliorer la précision et réduire la variance de la prédiction en combinant un grand nombre d'arbres de décision indépendants.
* **Intuition & Fonctionnement :**
  L'algorithme construit un ensemble d'arbres de décision entraînés chacun sur un sous-échantillon bootstrap des données et un sous-ensemble aléatoire des variables.
  1. *Entrée :* Données $X$ et cible $y$.
  2. *Calcul/Apprentissage :* Entraînement en parallèle de $B$ arbres de régression sur des tirages avec remise et sous-sélection aléatoire de variables à chaque nœud.
  3. *Sortie :* La prédiction finale est la moyenne exacte des prédictions de tous les arbres individuels.
* **Formule mathématique :**
  $$\hat{y}_{RF} = \frac{1}{B} \sum_{b=1}^{B} T_b(x)$$
  - **Définition des symboles :** $B$ est le nombre total d'arbres (`n_estimators`), $T_b(x)$ est la prédiction issue du $b$-ième arbre de décision.
  - **Paramètres (appris) :** L'ensemble des structures et seuils de tous les $B$ arbres.
  - **Hyperparamètres :** `n_estimators`, `max_depth`, `max_features`, `min_samples_split`.
* **Avantages :**
  1. Très robuste au surapprentissage par rapport à un arbre individuel.
  2. Offre des mesures fiables d'importance des variables.
* **Limites :**
  1. Coût en calcul et mémoire plus élevé qu'un simple arbre.
  2. Modèle en "boîte noire" difficile à interpréter visuellement par rapport à un seul arbre.

---

* **Cas d'usage 2 (Marketing) :** Prédiction de la valeur à vie d'un client (*Customer Lifetime Value - CLV*).
* **Nature des données & Prérequis :** Accepte des données mixtes (continues et catégorielles encodées). Aucun prétraitement de mise à l'échelle nécessaire.
* **Cas peu approprié :** Prédiction avec extrapolation forte sur des séries temporelles (comme tous les modèles à base d'arbres).
* **Métriques adaptées :** MAE, RMSE, $R^2$.

7. Gradient Boosting — régression
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Gradient Boosting pour la régression / *Gradient Boosting Regressor (HistGradientBoostingRegressor)*.
* **Famille :** Apprentissage supervisé — Méthodes d'ensemble séquentiel (*Boosting*).
* **Objectif :** Obtenir un modèle de prédiction très performant en construisant séquentiellement des arbres où chaque nouvel arbre corrige les erreurs (résidus) des précédents.
* **Intuition & Fonctionnement :**
  Contrairement à la Forêt Aléatoire qui entraîne les arbres en parallèle, le Gradient Boosting construit les arbres un par un.
  1. *Entrée :* Données $X$ et cible $y$.
  2. *Calcul/Apprentissage :* Initialisation avec une prédiction constante. À chaque étape, calcul des résidus (erreurs), puis entraînement d'un nouvel arbre pour prédire ces résidus.
  3. *Sortie :* Somme pondérée de tous les arbres entraînés.
* **Formule mathématique :**
  $$\hat{y}^{(m)} = \hat{y}^{(m-1)} + \nu \cdot h_m(x)$$
  - **Définition des symboles :** $\hat{y}^{(m)}$ est la prédiction à l'étape $m$, $h_m(x)$ est l'arbre entraîné sur les résidus, $\nu$ est le taux d'apprentissage (*learning rate*).
  - **Paramètres (appris) :** Les structures des arbres séquentiels et leurs poids.
  - **Hyperparamètres :** `max_iter` (nombre d'arbres), `learning_rate`, `max_depth`, `l2_regularization`.
* **Avantages :**
  1. Souvent l'un des algorithmes les plus performants et précis sur données tabulaires.
  2. Gestion native efficace des grandes masses de données (`HistGradientBoosting`).
* **Limites :**
  1. Très sensible au surapprentissage si le nombre d'itérations est trop grand ou le taux d'apprentissage mal réglé.
  2. Entraînement séquentiel (plus difficile à paralléliser qu'une forêt aléatoire).

---

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le schéma montre l'amélioration progressive : à chaque étape de boosting, l'ajout d'un nouvel arbre réduit le résidu global d'entraînement.

---

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Actuariat) :** Tarification des risques d'assurance et modélisation précise du coût moyen des sinistres.
* **Cas d'usage 2 (Énergie / Logistique) :** Prévision ultra-précise de la demande électrique à court terme selon la météo et le jour de la semaine.
* **Nature des données & Prérequis :** Données tabulaires (numériques/catégorielles). `HistGradientBoosting` gère nativement les valeurs manquantes.
* **Cas peu approprié :** Données non structurées (images, audio, texte brut).
* **Métriques adaptées :** MAE, RMSE, $R^2$.

  - *Ne permet PAS de conclure :* Ne donne pas accès à la courbe de perte au fil des 80 itérations (qui permettrait de repérer le début du surapprentissage).

8. Régression logistique
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Régression logistique / *Logistic Regression*.
* **Famille :** Apprentissage supervisé — Classification (binaire et multiclasse).
* **Objectif :** Estimer la probabilité qu'une observation appartienne à une classe spécifique à l'aide d'une fonction sigmoïde.
* **Intuition & Fonctionnement :**
  L'algorithme combine linéairement les variables explicatives puis transforme ce résultat à l'aide de la fonction logistique pour obtenir une valeur entre 0 et 1.
  1. *Entrée :* Matrice de caractéristiques $X$ et étiquettes catégorielles $y \in \{0, 1\}$.
  2. *Calcul/Apprentissage :* Minimisation de la perte d'entropie croisée (*log-loss*) par descente de gradient.
  3. *Sortie :* Probabilité d'appartenance à la classe positive $\hat{p}$, convertie en classe 0 ou 1 selon un seuil (ex: 0.5).
* **Formule mathématique :**
  $$\hat{p} = \sigma(z) = \frac{1}{1 + e^{-(X\beta + \beta_0)}}$$
  Perte entropie croisée binaire :
  $$\mathcal{L}(\beta) = -\frac{1}{n} \sum_{i=1}^{n} \left[ y_i \log(\hat{p}_i) + (1 - y_i) \log(1 - \hat{p}_i) \right]$$
  - **Définition des symboles :** $\sigma(z)$ est la fonction sigmoïde, $y_i$ la vraie classe, $\hat{p}_i$ la probabilité prédite, $\beta$ les coefficients.
  - **Paramètres (appris) :** Coefficients $\beta$ et intercept $\beta_0$.
  - **Hyperparamètres :** `C` (inverse de la pénalité), `penalty` (l1, l2, elasticnet), `solver`.
* **Avantages :**
  1. Modèle très rapide et facilement interprétable via les ratios de chances (*odds ratios*).
  2. Fournit des probabilités bien étalonnées en plus des prédictions de classes.
* **Limites :**
  1. Suppose une frontière de décision linéaire entre les classes.
  2. Sensible à la multicolinéarité et aux valeurs extrêmes.

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le graphique présente la courbe en S de la fonction sigmoïde qui compresse n'importe quelle valeur réelle dans l'intervalle [0, 1] avec le seuil de décision fixé à 0.5.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Credit Scoring) :** Évaluation du risque d'accord de prêt (prédire si un emprunteur sera en défaut de paiement ou non).
* **Cas d'usage 2 (Marketing) :** Prédiction de l'attrition client (*churn*) dans les services d'abonnement.
* **Nature des données & Prérequis :** Variables numériques (mise à l'échelle vivement recommandée via `StandardScaler`).
* **Cas peu approprié :** Données avec des frontières de séparation hautement complexes et non linéaires (ex: reconnaissance d'images).
* **Métriques adaptées :** F1-Score, AUC-ROC, Précision, Rappel, Matrice de confusion.

### 4. Analyse de l'expérience Python
* **Entrées du code :** `Xc_ent` et `yc_ent` dans un pipeline avec `StandardScaler()` et `LogisticRegression(max_iter=1000, random_state=42)`.
* **Hyperparamètres définis :** `max_iter=1000`, `random_state=GRAINE`.
* **Sortie obtenue :** Affiche le F1-score sur l'échantillon de test.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Donne la mesure du compromis précision/rappel sur le jeu de test binaire synthétique.
  - *Ne permet PAS de conclure :* Ne montre pas si ajuster le seuil de décision (actuellement 0.5) optimiserait le rappel ou la précision selon les contraintes métier.

9. KNN
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** K plus proches voisins / *K-Nearest Neighbors (KNN)*.
* **Famille :** Apprentissage supervisé — Classification / Méthode basée sur les instances (*Lazy Learning*).
* **Objectif :** Classer une nouvelle observation en se basant sur la classe majoritaire de ses $k$ voisins les plus proches dans l'espace des variables.
* **Intuition & Fonctionnement :**
  L'algorithme ne construit pas de modèle explicite pendant l'entraînement ; il stocke simplement les données.
  1. *Entrée :* Nouvelle donnée $x$ et ensemble d'entraînement stocké.
  2. *Calcul/Apprentissage :* Calcul de la distance (ex: euclidienne) entre $x$ et toutes les données stockées pour identifier les $k$ voisins les plus proches.
  3. *Sortie :* Vote majoritaire parmi les $k$ voisins pour déterminer la classe attribuée.
* **Formule mathématique :**
  Distance euclidienne entre deux points $p$ et $q$ :
  $$d(p, q) = \sqrt{\sum_{j=1}^{d} (p_j - q_j)^2}$$
  - **Définition des symboles :** $d$ est le nombre de dimensions (caractéristiques), $p_j$ et $q_j$ les coordonnées respectives des points.
  - **Paramètres (appris) :** Aucun (l'algorithme stocke simplement l'ensemble des données d'entraînement).
  - **Hyperparamètres :** `n_neighbors` ($k$), `metric` (euclidean, manhattan, etc.), `weights` (uniform, distance).
* **Avantages :**
  1. Très simple à comprendre, sans hypothèse sur la distribution des données.
  2. S'adapte naturellement aux problèmes multiclasses et non linéaires.
* **Limites :**
  1. Prédictions très lentes sur de grands jeux de données (doit parcourir toutes les données à chaque prédiction).
  2. Extrêmement sensible au "fléau de la dimension" et à l'absence d'échelle uniforme entre variables.

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le schéma illustre comment le choix de $k$ modifie la décision : pour $k=3$, la classe A l'emporte, tandis que pour un $k$ plus grand, la classe B peut devenir majoritaire autour du point cible.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Risk Management) :** Détection de profils de clients à haut risque par comparaison directe avec les profils historiques similaires.
* **Cas d'usage 2 (Recommandation) :** Systèmes de recommandation basés sur le filtrage collaboratif (trouver les $k$ utilisateurs ayant des goûts similaires).
* **Nature des données & Prérequis :** Variables numériques impérativement standardisées (`StandardScaler`).
* **Cas peu approprié :** Jeux de données volumineux avec des centaines de colonnes (la distance perd de son sens en haute dimension).
* **Métriques adaptées :** F1-Score, Précision, Rappel, Accuracy.

### 4. Analyse de l'expérience Python
* **Entrées du code :** `Xc_ent` et `yc_ent` dans un pipeline avec `StandardScaler()` et `KNeighborsClassifier(n_neighbors=5)`.
* **Hyperparamètres définis :** `n_neighbors=5`.
* **Sortie obtenue :** Affiche le F1-score sur l'échantillon de test.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Mesure la capacité de classification de 5 voisins sur les données standardisées.
  - *Ne permet PAS de conclure :* Ne montre pas si $k=5$ est la valeur optimale ou si $k=3$ ou $k=7$ donnerait de meilleurs résultats.

10. Naive Bayes
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Naïve Bayes / *Naive Bayes (Gaussian Naive Bayes)*.
* **Famille :** Apprentissage supervisé — Classification probabiliste.
* **Objectif :** Calcule la probabilité a posteriori d'appartenir à chaque classe en utilisant le théorème de Bayes sous l'hypothèse de l'indépendance conditionnelle des caractéristiques.
* **Intuition & Fonctionnement :**
  L'algorithme combine la probabilité a priori des classes avec les probabilités conditionnelles de chaque variable (supposée gaussienne pour des données continues).
  1. *Entrée :* Données $X$ et étiquettes $y$.
  2. *Calcul/Apprentissage :* Estimation de la moyenne et de la variance de chaque caractéristique pour chaque classe.
  3. *Sortie :* Classe qui maximise la probabilité a posteriori (règle du Maximum A Posteriori - MAP).
* **Formule mathématique :**
  Théorème de Bayes appliqué avec l'hypothèse "naïve" d'indépendance :
  $$P(Y = c \mid X_1, \dots, X_d) \propto P(Y = c) \prod_{j=1}^{d} P(X_j \mid Y = c)$$
  - **Définition des symboles :** $P(Y=c)$ est la probabilité a priori de la classe $c$, $P(X_j \mid Y=c)$ est la densité de probabilité gaussienne de la variable $j$ conditionnellement à la classe $c$.
  - **Paramètres (appris) :** Moyennes $\mu_{c,j}$ et variances $\sigma_{c,j}^2$ par classe et par variable.
  - **Hyperparamètres :** `var_smoothing` (stabilité numérique).
* **Avantages :**
  1. Extrêmement rapide à entraîner et peu gourmand en puissance de calcul.
  2. Fonctionne de façon remarquable sur la classification de textes/emails malgré l'hypothèse simplificatrice d'indépendance.
* **Limites :**
  1. Supposer l'indépendance totale des variables est rarement vrai en pratique.
  2. Sensible au problème de "fréquence nulle" (géré par le lissage de Laplace).

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le schéma présente les distributions gaussiennes de chaque classe pour une variable et la frontière théorique où les probabilités a posteriori s'équilibrent.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Marketing / Communication) :** Filtrage des spams et classification automatique des emails entrants des clients.
* **Cas d'usage 2 (Finance / Sentiment Analysis) :** Analyse du sentiment (positif/négatif/neutre) dans les rapports financiers ou actualités boursières.
* **Nature des données & Prérequis :** Variables continues (GaussianNB) ou comptages/fréquences textuelles (MultinomialNB).
* **Cas peu approprié :** Données comportant de très fortes corrélations dépendantes entre variables explicatives.
* **Métriques adaptées :** F1-Score, Précision, Rappel, ROC-AUC.

### 4. Analyse de l'expérience Python
* **Entrées du code :** `GaussianNB()` entraîné sur `Xc_ent` et `yc_ent`.
* **Hyperparamètres définis :** Aucun (configuration gaussienne standard).
* **Sortie obtenue :** Affiche le F1-score sur `Xc_test`.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Évalue l'efficacité du classifieur bayésien gaussien rapide sur le jeu de classification.
  - *Ne permet PAS de conclure :* Ne valide pas si l'hypothèse de normalité gaussienne des variables d'entrée est réellement respectée sur ces données.

11. Arbre de décision
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Arbre de décision pour la classification / *Decision Tree Classifier*.
* **Famille :** Apprentissage supervisé — Arbres de décision.
* **Objectif :** Séparer les données en sous-groupes de plus en plus homogènes (*purs*) par une série de règles SI-ALORS sur les caractéristiques.
* **Intuition & Fonctionnement :**
  L'algorithme choisit à chaque étape la variable et le seuil qui maximisent le gain d'information ou réduisent au mieux l'impureté de Gini.
  1. *Entrée :* Données $X$ et étiquettes catégorielles $y$.
  2. *Calcul/Apprentissage :* Évaluation de l'impureté de Gini ou de l'entropie à chaque nœud pour déterminer la séparation optimale.
  3. *Sortie :* Classe majoritaire présente dans la feuille d'arrivée.
* **Formule mathématique :**
  Indice d'impureté de Gini :
  $$Gini(N) = 1 - \sum_{i=1}^{K} p_i^2$$
  - **Définition des symboles :** $K$ est le nombre de classes, $p_i$ est la proportion d'éléments appartenant à la classe $i$ dans le nœud $N$.
  - **Paramètres (appris) :** Variables sélectionnées à chaque nœud, seuils de séparation et affectation des classes aux feuilles.
  - **Hyperparamètres :** `max_depth`, `min_samples_split`, `criterion` (gini, entropy).
* **Avantages :**
  1. Entièrement explicable et visualisable sous forme de diagramme décisionnel simple.
  2. Gère nativement les variables numériques et catégorielles sans mise à l'échelle.
* **Limites :**
  1. Très enclin au surapprentissage si sa profondeur n'est pas limitée.
  2. Instable : une petite modification des données d'entraînement peut changer radicalement la structure de l'arbre.

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le schéma représente la séparation de l'espace des données par des droites orthogonales correspondant aux règles SI-ALORS générées par les nœuds de l'arbre.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Octroi de crédit) :** Arbre d'approbation automatique des prêts bancaires avec règles explicites destinées aux régulateurs.

12. Forêt aléatoire — classification
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Forêt aléatoire pour la classification / *Random Forest Classifier*.
* **Famille :** Apprentissage supervisé — Méthodes d'ensemble (*Bagging*).
* **Objectif :** Combiner la prédiction d'un grand nombre d'arbres de décision indépendants pour améliorer la généralisation et réduire la variance.
* **Intuition & Fonctionnement :**
  L'algorithme introduit une double aléatorisation (tirage des observations avec remise + sous-sélection aléatoire de variables à chaque nœud).
  1. *Entrée :* Données $X$ et étiquettes $y$.
  2. *Calcul/Apprentissage :* Entraînement en parallèle de $B$ arbres de décision sur des échantillons bootstrap distincts.
  3. *Sortie :* Vote majoritaire (ou moyenne des probabilités) des arbres de la forêt.
* **Formule mathématique :**
  Vote majoritaire :
  $$\hat{y}_{RF} = \text{mode} \left( \{ T_1(x), T_2(x), \dots, T_B(x) \} \right)$$
  - **Définition des symboles :** $B$ est le nombre d'arbres (`n_estimators`), $T_b(x)$ est la prédiction du $b$-ième arbre de classification.
  - **Paramètres (appris) :** L'ensemble des structures de tous les arbres individuels de la forêt.
  - **Hyperparamètres :** `n_estimators`, `max_depth`, `max_features`, `criterion`.
* **Avantages :**
  1. Excellente performance globale sans nécessiter un réglage poussé des hyperparamètres.
  2. Évite le surapprentissage grâce au principe d'agrégation d'arbres diversifiés.
* **Limites :**
  1. Perte d'interprétabilité directe par rapport à un arbre individuel.
  2. Consommation en mémoire et temps de calcul plus importants sur les très grands jeux de données.

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le schéma illustre l'obtention du résultat final grâce au vote majoritaire compilant les réponses issues des différents arbres indépendants.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Détection de fraude) :** Détection de transactions frauduleuses par carte de crédit sur des données historiques complexes.
* **Cas d'usage 2 (Marketing / Upselling) :** Ciblage des clients ayant la plus forte propension à acheter un produit complémentaire.
* **Nature des données & Prérequis :** Données numériques ou catégorielles encodées. Aucune mise à l'échelle requise.
* **Cas peu approprié :** Systèmes embarrassés par une forte contrainte de temps de latence en temps réel extrême (où un modèle plus léger est exigé).
* **Métriques adaptées :** F1-Score, AUC-ROC, Précision, Rappel.

13. SVM
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Machines à vecteurs de support / *Support Vector Machines (SVM / SVC)*.
* **Famille :** Apprentissage supervisé — Classification par séparateur à marge maximale.
* **Objectif :** Trouver l'hyperplan qui sépare au mieux les classes en maximisant la distance (marge) entre l'hyperplan et les points les plus proches (vecteurs de support).
* **Intuition & Fonctionnement :**
  L'algorithme s'appuie sur la marge maximale et utilise le "truc du noyau" (*kernel trick*) pour projeter les données dans un espace de plus grande dimension où elles deviennent linéairement séparables.
  1. *Entrée :* Données $X$ et étiquettes $y \in \{-1, 1\}$.
  2. *Calcul/Apprentissage :* Résolution d'un problème d'optimisation quadratique sous contraintes pour trouver les vecteurs de support.
  3. *Sortie :* Classification d'un point selon le côté de l'hyperplan séparateur où il se situe.
* **Formule mathématique :**
  Problème d'optimisation (marge adoucie) :
  $$\min_{w, b, \xi} \frac{1}{2} \|w\|^2 + C \sum_{i=1}^{n} \xi_i \quad \text{sujet à} \quad y_i(w^T \phi(x_i) + b) \ge 1 - \xi_i$$
  - **Définition des symboles :** $w$ est le vecteur normal à l'hyperplan, $C$ contrôle le compromis marge/erreur, $\xi_i$ sont les variables d'ajustement (*slack variables*), $\phi(x)$ la projection par le noyau.
  - **Paramètres (appris) :** Poids $w$, biais $b$ et choix des vecteurs de support.
  - **Hyperparamètres :** `C`, `kernel` (rbf, linear, poly), `gamma`.
* **Avantages :**
  1. Trè efficace en haute dimension (même lorsqu'il y a plus de variables que d'observations).
  2. Robuste au surapprentissage grâce à l'optimisation axée uniquement sur les vecteurs de support.
* **Limites :**
  1. Temps de calcul quadratique à cubique avec le nombre d'observations (lent sur très grands datasets).
  2. Nécessite obligatoirement un réglage méticuleux des hyperparamètres (`C`, `gamma`) et la mise à l'échelle.

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le schéma illustre l'hyperplan séparateur, la largeur de la marge maximale et la localisation des vecteurs de support qui définissent la frontière.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Trading) :** Classification de mouvements de marché à court terme basés sur de multiples indicateurs techniques complexes.
* **Cas d'usage 2 (Biologie / Santé) :** Diagnostic médical à partir de données d'imagerie ou d'analyses génomiques en haute dimension.
* **Nature des données & Prérequis :** Variables numériques obligatoirement standardisées (`StandardScaler`).
* **Cas peu approprié :** Très grands jeux de données dépassant plusieurs centaines de milliers d'observations.
* **Métriques adaptées :** F1-Score, Précision, Rappel, ROC-AUC.

### 4. Analyse de l'expérience Python

* **Hyperparamètres définis :** `kernel='rbf'`, `C=1.0`, `random_state=GRAINE`.
* **Sortie obtenue :** Affiche le F1-score sur l'échantillon de test `Xc_test`.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Mesure la qualité de classification avec un noyau RBF gaussien à marge adoucie standard (`C=1.0`).
  - *Ne permet PAS de conclure :* Ne montre pas l'influence de la valeur de `gamma` ou d'un noyau alternatif sur le résultat.

14. Gradient Boosting — classification
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Gradient Boosting pour la classification / *Gradient Boosting Classifier (HistGradientBoostingClassifier)*.
* **Famille :** Apprentissage supervisé — Méthodes d'ensemble séquentiel (*Boosting*).
* **Objectif :** Construire séquentiellement une suite d'arbres de décision simples où chaque nouvel arbre cherche à corriger les erreurs de l'ensemble précédent.
* **Intuition & Fonctionnement :**
  L'algorithme optimise une fonction de perte globale en appliquant la descente de gradient dans l'espace des fonctions (les arbres).
  1. *Entrée :* Données $X$ et étiquettes $y$.
  2. *Calcul/Apprentissage :* Calcul du gradient négatif de la fonction de perte logistique à chaque étape, puis entraînement d'un arbre sur ce gradient.
  3. *Sortie :* Probabilité finale obtenue par combinaison pondérée de tous les arbres.
* **Formule mathématique :**
  Mise à jour séquentielle du modèle :
  $$F_m(x) = F_{m-1}(x) + \gamma_m h_m(x)$$
  - **Définition des symboles :** $F_m(x)$ est le modèle à l'étape $m$, $h_m(x)$ l'arbre entraîné sur les pseudo-résidus (gradients), $\gamma_m$ le poids affecté à cet arbre.
  - **Paramètres (appris) :** Ensemble des arbres successifs et leurs coefficients d'apprentissage.
  - **Hyperparamètres :** `max_iter`, `learning_rate`, `max_depth`, `l2_regularization`.
* **Avantages :**
  1. Performances de prédiction exceptionnelles sur les données tabulaires.
  2. Gère efficacement les valeurs manquantes et les jeux de données volumineux via la discrétisation par histogrammes.
* **Limites :**
  1. Risque d'overfitting si le nombre d'itérations est trop élevé sans régularisation adéquate.
  2. Construction séquentielle plus longue à entraîner qu'un modèle basé sur le bagging parallèle.

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / scoring bancaire) :** Détection de l'insolvabilité des clients pour la validation automatique des crédits.
* **Cas d'usage 2 (Marketing) :** Prédiction de la conversion d'un prospect à la suite d'une campagne publicitaire.
* **Nature des données & Prérequis :** Données tabulaires (numériques/catégorielles).
* **Cas peu approprié :** Traitement de données brutes d'images ou de signal textuel non structuré.
* **Métriques adaptées :** F1-Score, AUC-ROC, PR-AUC, Précision, Rappel.

### 4. Analyse de l'expérience Python

15. AdaBoost
1. Bloc Markdown (à copier dans me cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** AdaBoost / *Adaptive Boosting (AdaBoost Classifier)*.
* **Famille :** Apprentissage supervisé — Méthodes d'ensemble séquentiel (*Boosting*).
* **Objectif :** Combiner plusieurs classifieurs faibles (souvent des arbres très courts d'un seul niveau, appelés *stumps*) en augmentant le poids des observations mal classées à chaque étape.
* **Intuition & Fonctionnement :**
  Les modèles sont entraînés les uns après les autres. À chaque tour, les données mal classées par l'arbre précédent reçoivent un poids plus élevé pour contraindre le suivant à se focaliser sur elles.
  1. *Entrée :* Données $X$ et étiquettes $y \in \{-1, 1\}$, poids d'observations initialisés uniformément à $1/n$.
  2. *Calcul/Apprentissage :* Entraînement séquentiel de classifieurs faibles, calcul de leur erreur, et mise à jour exponentielle des poids des exemples.
  3. *Sortie :* Vote pondéré de l'ensemble des classifieurs faibles.
* **Formule mathématique :**
  Mise à jour des poids d'observations :
  $$w_i^{(m+1)} = w_i^{(m)} \exp \left( \alpha_m y_i h_m(x_i) \right)$$
  Poids du classifieur : $\alpha_m = \frac{1}{2} \ln \left( \frac{1 - \epsilon_m}{\epsilon_m} \right)$
  - **Définition des symboles :** $w_i^{(m)}$ est le poids de l'exemple $i$ à l'étape $m$, $\epsilon_m$ est le taux d'erreur du classifieur faible $h_m$, $\alpha_m$ est l'importance accordée à ce classifieur.
  - **Paramètres (appris) :** Poids $\alpha_m$ des classifieurs et règles des arbres courts.
  - **Hyperparamètres :** `n_estimators`, `learning_rate`.
* **Avantages :**
  1. Algorithme historiquement élégant et simple à mettre en œuvre.
  2. Peu enclin au surapprentissage sur des jeux de données propres sans trop de bruit.
* **Limites :**
  1. Extrêmement sensible au bruit et aux valeurs atypiques (*outliers*), qui reçoivent des poids démesurés.
  2. Moins performant que le Gradient Boosting sur les jeux de données complexes et bruités.

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le schéma illustre l'augmentation de la taille (poids) des points mal classés qui contraint le classifieur suivant à ajuster sa frontière.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Vision par ordinateur) :** Détection de visages historiques (algorithme de Viola-Jones basé sur AdaBoost).
* **Cas d'usage 2 (Marketing / Churn) :** Identification rapide de clients insatisfaits sur des critères comportementaux basiques.
* **Nature des données & Prérequis :** Variables numériques ou catégorielles. Sensible au bruit dans les étiquettes.
* **Cas peu approprié :** Jeux de données fortement bruités comportant des erreurs d'étiquetage.
* **Métriques adaptées :** F1-Score, Précision, Rappel, Accuracy.

### 4. Analyse de l'expérience Python

* **Hyperparamètres définis :** `n_estimators=70`, `random_state=GRAINE`.

  - *Permet de conclure :* Mesure la performance d'une combinaison de 70 classifieurs faibles réadaptés.
  - *Ne permet PAS de conclure :* Ne montre pas l'évolution des poids attribués aux observations au cours des itérations.

16. XGBoost
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** XGBoost / *Extreme Gradient Boosting (XGBClassifier)*.
* **Famille :** Apprentissage supervisé — Méthodes d'ensemble séquentiel (*Boosting*).
* **Objectif :** Fournir une implémentation ultra-optimisée, rapide et régularisée du Gradient Boosting pour maximiser les performances de prédiction.
* **Intuition & Fonctionnement :**
  XGBoost utilise un développement de Taylor de second ordre de la fonction de perte et intègre une régularisation ($L_1$ et $L_2$) de la structure des arbres pour contrôler la complexité.
  1. *Entrée :* Données $X$ et étiquettes $y$.
  2. *Calcul/Apprentissage :* Minimisation de la fonction d'objectif avec calcul des dérivées premières (gradients) et secondes (hessiens).
  3. *Sortie :* Prédiction finale issue de la somme pondérée de tous les arbres de régression.
* **Formule mathématique :**
  Fonction d'objectif à l'étape $t$ :
  $$\mathcal{L}^{(t)} \approx \sum_{i=1}^{n} \left[ g_i f_t(x_i) + \frac{1}{2} h_i f_t^2(x_i) \right] + \Omega(f_t)$$
  Terme de régularisation : $\Omega(f) = \gamma T + \frac{1}{2} \lambda \sum_{j=1}^{T} w_j^2$
  - **Définition des symboles :** $g_i$ et $h_i$ sont les gradients du premier et second ordre, $T$ est le nombre de feuilles, $w_j$ les poids des feuilles, $\gamma$ et $\lambda$ les coefficients de régularisation.
  - **Paramètres (appris) :** Structure des arbres et valeurs des feuilles.
  - **Hyperparamètres :** `n_estimators`, `max_depth`, `learning_rate`, `subsample`, `colsample_bytree`, `reg_lambda`, `reg_alpha`.
* **Avantages :**
  1. Vitesse d'exécution exceptionnelle grâce à la parallélisation du calcul des séparations.
  2. Très haute précision, ayant dominé pendant des années les compétitions de science des données (Kaggle).
* **Limites :**
  1. Grand nombre d'hyperparamètres complexes à régler.
  2. Modèle boîte noire plus difficile à interpréter et à déployer sans outils d'explicabilité (SHAP).

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le schéma présente la régularisation de la structure des arbres combinée aux gradients du premier et second ordre qui optimisent la convergence.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Évaluation du risque) :** Systèmes de notation de crédit et détection de fraudes complexes en temps réel.
* **Cas d'usage 2 (E-commerce / Conversion) :** Prédiction du taux de clics (CTR) et personnalisation des recommandations en ligne.
* **Nature des données & Prérequis :** Variables numériques (gère nativement les valeurs manquantes et l'encodage).
* **Cas peu approprié :** Petits jeux de données simples où des modèles plus légers suffisent largement.
* **Métriques adaptées :** F1-Score, AUC-ROC, Log-Loss, PR-AUC.

### 4. Analyse de l'expérience Python
* **Entrées du code :** `XGBClassifier(n_estimators=80, max_depth=3, random_state=42, eval_metric='logloss')` ajusté sur `Xc_ent` et `yc_ent`.
* **Hyperparamètres définis :** `n_estimators=80`, `max_depth=3`, `eval_metric='logloss'`.
* **Sortie obtenue :** Affiche le F1-score calculé sur le jeu de test `Xc_test`.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Évalue les performances brutes du classifieur XGBoost régularisé sur le jeu de test.
  - *Ne permet PAS de conclure :* Ne montre pas l'influence du paramètre `subsample` ou des régularisations `lambda`/`alpha` sur le score final.
2. Bloc Code Python pour le schéma (à copier dans une cellule Code)

Markdown
### 1. Présentation
* **Nom (FR / EN) :** K-moyennes / *K-Means Clustering*.
* **Famille :** Apprentissage non supervisé — Regroupement (*Clustering* géométrique par partitions).
* **Objectif :** Partitionner un ensemble d'observations en $K$ groupes (clusters) distincts en minimisant la variance au sein de chaque groupe (inertie intra-classe).
* **Intuition & Fonctionnement :**
  L'algorithme positionne aléatoirement $K$ centroïdes, affecte chaque observation au centroïde le plus proche, puis réexécute l'actualisation des centroïdes en calculant la moyenne des points affectés jusqu'à convergence.
  1. *Entrée :* Données $X$ sans étiquettes et nombre préfixé de clusters $K$.
  2. *Calcul/Apprentissage :* Iteration alternée (Assignation des points aux centroïdes ➔ Mise à jour des positions des centroïdes).
  3. *Sortie :* Étiquette du cluster affecté à chaque observation et coordonnées finales des $K$ centroïdes.
* **Formule mathématique :**
  Minimisation de l'inertie intra-cluster ($WCSS$) :
  $$\min_{S} \sum_{k=1}^{K} \sum_{x \in S_k} \|x - \mu_k\|^2$$
  - **Définition des symboles :** $S_k$ représente l'ensemble des points attribués au cluster $k$, $\mu_k$ est le centroïde (moyenne) du cluster $k$.
  - **Paramètres (appris) :** Emplacements des $K$ centroïdes $\mu_1, \dots, \mu_K$.
  - **Hyperparamètres :** `n_clusters` ($K$), `init` (kmeans++, random), `n_init`, `max_iter`.
* **Avantages :**
  1. Algorithme très rapide, scalable et simple à implémenter sur de grandes bases de données.
  2. Convergence garantie en un nombre fini d'itérations.
* **Limites :**
  1. Nécessite de fixer le nombre de clusters $K$ à priori.
  2. Sensible aux initialisations, aux valeurs extrêmes et suppose des clusters de formes sphériques de tailles comparables.

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le schéma présente les $K$ centroïdes calculés et les frontières de Voronoi séparant les différentes régions d'attraction des clusters.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Marketing / Segmentation) :** Segmentation de la base clients en groupes homogènes d'achats pour cibler des offres marketing spécifiques.
* **Cas d'usage 2 (Finance / Portefeuilles) :** Regroupement d'actifs financiers aux comportements de rendements similaires pour la diversification.
* **Nature des données & Prérequis :** Variables numériques impérativement standardisées (`StandardScaler`).
* **Cas peu approprié :** Données comportant des clusters de formes arbitraires non sphériques (ex: croissants de lune) ou de densités très différentes.
* **Métriques adaptées :** Coefficient de Silhouette, Score Davies-Bouldin, Inertie (méthode du coude / *Elbow method*).

### 4. Analyse de l'expérience Python
* **Entrées du code :** `KMeans(n_clusters=3, n_init=10, random_state=42)` appliqué sur les données synthétiques `X_clu`.
* **Hyperparamètres définis :** `n_clusters=3`, `n_init=10`, `random_state=GRAINE`.
* **Sortie obtenue :** Affiche le score de Silhouette et affiche le graphique de dispersion colorié par clusters.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Mesure la séparation globale et l'homogénéité des 3 clusters trouvés sur ce nuage 2D.
  - *Ne permet PAS de conclure :* Ne garantit pas que $K=3$ soit la valeur optimale sur le plan métier sans analyser l'évolution du score de Silhouette pour d'autres valeurs de $K$.

18. Classification hiérarchique
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Classification hiérarchique / *Hierarchical Agglomerative Clustering (HAC)*.
* **Famille :** Apprentissage non supervisé — Regroupement hiérarchique (approche ascendante).
* **Objectif :** Construire une hiérarchie d'emboîtement des données sous forme d'arbre (dendrogramme) sans fixer le nombre de groupes au départ.
* **Intuition & Fonctionnement :**
  Chaque point commence dans son propre cluster. À chaque étape, l'algorithme fusionne les deux clusters les plus proches selon un critère de liaison (*linkage*) prédéfini jusqu'à ce que toutes les données forment un seul groupe.
  1. *Entrée :* Données $X$ et métrique de distance.
  2. *Calcul/Apprentissage :* Calcul séquentiel de la matrice des distances et fusion progressive des paires de clusters les plus proches.
  3. *Sortie :* Arbre hiérarchique (dendrogramme) découpé au niveau souhaité pour extraire $K$ clusters.
* **Formule mathématique :**
  Distance de Ward (minimisation de la variance cumulée) :
  $$\Delta(A, B) = \frac{n_A n_B}{n_A + n_B} \|\mu_A - \mu_B\|^2$$
  - **Définition des symboles :** $n_A, n_B$ représentent les effectifs des clusters $A$ et $B$, $\mu_A, \mu_B$ sont leurs centroïdes respectifs.
  - **Paramètres (appris) :** Matrice de liaisons hiérarchiques et regroupements.
  - **Hyperparamètres :** `n_clusters`, `linkage` (ward, complete, average, single), `metric`.
* **Avantages :**
  1. Fournit une visualisation hiérarchique très intuitive de la structure des données (dendrogramme).
  2. Ne nécessite pas d'imposer $K$ dès le départ (on peut couper l'arbre à n'importe quelle hauteur).
* **Limites :**
  1. Complexité temporelle et spatiale élevée en $\mathcal{O}(n^3)$ ou $\mathcal{O}(n^2)$, inutilisable sur de grands jeux de données.
  2. Une décision de fusion à une étape donnée ne peut jamais être annulée ultérieurement.

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le dendrogramme ci-dessous illustre les fusions ascendantes des observations et la ligne de coupe horizontale déterminant le nombre final de sous-groupes.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Gestion RH / Comportement) :** Établissement d'une typologie d'emplois ou de compétences en construisant des arbres de similarité fonctionnelle.
* **Cas d'usage 2 (Biologie / Logistique) :** Structuration de taxonomies de produits ou de zones géographiques de distribution.
* **Nature des données & Prérequis :** Variables numériques obligatoirement standardisées (`StandardScaler`).
* **Cas peu approprié :** Très volumineux jeux de données comprenant des dizaines de milliers de lignes.
* **Métriques adaptées :** Coefficient de Silhouette, Cofficient Cophenetique (fidélité de l'arbre).

### 4. Analyse de l'expérience Python
* **Entrées du code :** `AgglomerativeClustering(n_clusters=3)` appliqué directement sur `X_clu`.
* **Hyperparamètres définis :** `n_clusters=3` (critère de liaison par défaut : Ward).
* **Sortie obtenue :** Affiche la valeur de la Silhouette pour la partition obtenue.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Évalue la qualité des 3 sous-groupes fusionnés par le critère de Ward.
  - *Ne permet PAS de conclure :* Ne montre pas l'arbre d'assemblage hiérarchique (dendrogramme) qui permet d'identifier la hauteur de coupe optimale.

19. DBSCAN
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** DBSCAN / *Density-Based Spatial Clustering of Applications with Noise*.
* **Famille :** Apprentissage non supervisé — Regroupement fondé sur la densité.
* **Objectif :** Découvrir des clusters de formes arbitraires basés sur la densité locale de points tout en isolant automatiquement le bruit (valeurs atypiques).
* **Intuition & Fonctionnement :**
  L'algorithme parcourt les données et définit un cluster comme une zone continue de haute densité séparée par des zones de faible densité. Il classifie chaque point en point cœur, point frontière ou point de bruit.
  1. *Entrée :* Données $X$, rayon $\epsilon$ (*eps*) et nombre minimal de points *min_samples*.
  2. *Calcul/Apprentissage :* Recherche des voisinages de taille $\epsilon$. Si un point possède $\ge \text{min\_samples}$ voisins, un nouveau cluster se forme.
  3. *Sortie :* Attributions des clusters (avec l'étiquette -1 réservée au bruit).
* **Formule mathématique :**
  Voisinage $\epsilon$ d'un point $p$ :
  $$N_{\epsilon}(p) = \{q \in D \mid d(p, q) \le \epsilon\}$$
  Condition de point cœur : $|N_{\epsilon}(p)| \ge \text{min\_samples}$
  - **Définition des symboles :** $d(p, q)$ est la distance entre points, $\epsilon$ le rayon maximal, $\text{min\_samples}$ le seuil minimal de densité.
  - **Paramètres (appris) :** Groupes de densité et détection des points de bruit (-1).
  - **Hyperparamètres :** `eps` ($\epsilon$), `min_samples`, `metric`.
* **Avantages :**
  1. Ne nécessite pas de spécifier le nombre de clusters à l'avance.
  2. Capable de détecter des clusters de formes géométriques arbitraires et isole nativement les *outliers*.
* **Limites :**
  1. Sensible au choix des hyperparamètres `eps` et `min_samples`.
  2. Performant avec peine si les clusters du jeu de données présentent des densités très différentes.

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le schéma illustre la notion de densité : les points cœurs (fort encadrement), les points frontières et les points isolés rejetés sous l'étiquette de bruit (-1).

### 3. Contextes d me utilisation
* **Cas d'usage 1 (Geomarketing / Logistique) :** Détection automatique de zones de forte concentration géographique de clients ou d'accidents.
* **Cas d'usage 2 (Finance / Traitement d'anomalies) :** Regroupement de transactions normales et isolement de comportements hors normes (bruit).
* **Nature des données & Prérequis :** Variables numériques obligatoirement standardisées (`StandardScaler`).
* **Cas peu approprié :** Données en haute dimension où la notion de distance dense s'estompe.
* **Métriques adaptées :** Coefficient de Silhouette (calculé hors points de bruit), Nombre de clusters identifiés, Taux de bruit.

### 4. Analyse de l'expérience Python
* **Entrées du code :** `DBSCAN(eps=.8, min_samples=5)` ajusté sur `X_clu`.

* **Sortie obtenue :** Affiche les étiquettes de groupes identifiées et le nombre total de points considérés comme bruit (`-1`).
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Montre combien de clusters denses ont été construits et combien de points ont été isolés.
  - *Ne permet PAS de conclure :* Ne montre pas si augmenter ou réduire `eps` permettrait de rattacher le bruit sans fusionner des clusters distincts.

20. Mélanges gaussiens
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Modèles de mélanges gaussiens / *Gaussian Mixture Models (GMM)*.
* **Famille :** Apprentissage non supervisé — Clustering probabiliste / Modèles à variables latentes.
* **Objectif :** Représenter la distribution des données comme une superposition de plusieurs distributions gaussiennes multidimensionnelles.
* **Intuition & Fonctionnement :**
  Contrairement à K-means qui effectue une affectation stricte (*hard clustering*), GMM réalise une affectation probabiliste souple (*soft clustering*) via l'algorithme Espérance-Maximisation (EM).
  1. *Entrée :* Données $X$ et nombre de composantes $K$.
  2. *Calcul/Apprentissage :* Algorithme EM (Étape E : calcul des probabilités d'appartenance aux gaussiennes ; Étape M : mise à jour des moyennes, covariances et poids des gaussiennes).
  3. *Sortie :* Probabilités partielles qu'une donnée appartienne à chacune des $K$ composantes.
* **Formule mathématique :**
  Densité du mélange :
  $$p(x) = \sum_{k=1}^{K} \pi_k \mathcal{N}(x \mid \mu_k, \Sigma_k)$$
  - **Définition des symboles :** $\pi_k$ est le poids (proportion) de la $k$-ième gaussienne ($\sum \pi_k = 1$), $\mathcal{N}$ la loi normale multidimensionnelle de moyenne $\mu_k$ et de matrice de covariance $\Sigma_k$.
  - **Paramètres (appris) :** Poids $\pi_k$, moyennes $\mu_k$, matrices de covariance $\Sigma_k$.
  - **Hyperparamètres :** `n_components` ($K$), `covariance_type` (full, tied, diag, spherical).
* **Avantages :**
  1. Offre une grande flexibilité géométrique grâce à l'estimation des matrices de covariance (clusters ellipsoïdaux).
  2. Fournit une mesure de l'incertitude de l'appartenance aux sous-groupes (*soft clustering*).
* **Limites :**
  1. Sensible au choix des conditions initiales et aux maxima locaux de la vraisemblance.
  2. Risque d'instabilité numérique si une matrice de covariance devient singulière.

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le schéma présente les ellipsoïdes de niveau de probabilité correspondant aux différentes composantes gaussiennes ajustées sur les données.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Régimes de marché) :** Identification des régimes financiers (ex: régime à faible volatilité vs régime à haute volatilité).
* **Cas d'usage 2 (Traitement du signal / Sécurité) :** Modélisation de distributions complexes de comportements d'utilisateurs.
* **Nature des données & Prérequis :** Variables numériques (mises à l'échelle recommandées).
* **Cas peu approprié :** Jeux de données avec des structures de formes complexes non convexes.
* **Métriques adaptées :** BIC (Critère d'information bayésien), AIC, Coefficient de Silhouette.

### 4. Analyse de l'expérience Python

* **Hyperparamètres définis :** `n_components=3`, `random_state=GRAINE`.
* **Sortie obtenue :** Affiche les poids appris des 3 gaussiennes ($\pi_k$) et le score de Silhouette des prédictions.

  - *Permet de conclure :* Donne la proportion relative de chaque composante et la qualité du découpage probabiliste.
  - *Ne permet PAS de conclure :* Ne fournit pas la valeur du critère BIC qui permet de choisir scientifiquement le nombre optimal de gaussiennes.

21. ACP
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Analyse en composantes principales / *Principal Component Analysis (PCA)*.
* **Famille :** Apprentissage non supervisé — Réduction de dimension (méthode linéaire).
* **Objectif :** Réduire la dimension d'un tableau de données en projetant les observations sur un sous-espace linéaire de manière à conserver le maximum de variance (information).
* **Intuition & Fonctionnement :**
  L'ACP effectue une rotation des axes d'origine pour définir de nouvelles variables orthogonales et non corrélées (les composantes principales), ordonnées par quantité de variance expliquée.
  1. *Entrée :* Données $X$ centré-réduites.
  2. *Calcul/Apprentissage :* Calcul de la matrice de covariance, extraction des valeurs propres et des vecteurs propres (ou via SVD).
  3. *Sortie :* Matrice de données projetée dans le nouvel espace de dimension réduite $k < d$.
* **Formule mathématique :**
  Maximisation de la variance projetée sous contrainte d'orthonormalité :
  $$\max_{w} w^T \Sigma w \quad \text{sujet à} \quad w^T w = 1$$
  - **Définition des symboles :** $\Sigma$ est la matrice de covariance des données, $w$ le vecteur propre (axe principal).
  - **Paramètres (appris) :** Composantes principales (vecteurs propres) et variance associée (valeurs propres).
  - **Hyperparamètres :** `n_components` ($k$).
* **Avantages :**
  1. Technique fondamentale d'apprentissage pour éliminer la multicolinéarité et compresser les données.
  2. Permet la visualisation en 2D/3D de jeux de données complexes comportant des dizaines de colonnes.
* **Limites :**
  1. Méthode strictement linéaire : incapable de saisir des structures non linéaires complexes (manifolds).
  2. Les nouvelles composantes projetées sont des combinaisons linéaires parfois complexes à interpréter physiquement.

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le schéma présente l'éboulis des valeurs propres qui illustre le pourcentage de variance expliquée retenu par chaque composante principale.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Analyse factorielle) :** Réduction de dizaines de ratios financiers en quelques facteurs synthétiques représentatifs.
* **Cas d'usage 2 (Traitement d'images / Compression) :** Compression d'images et réduction du bruit avant classification.
* **Nature des données & Prérequis :** Variables numériques obligatoirement centré-réduites (`StandardScaler`).
* **Cas peu approprié :** Données en structure de géométrie fortement non linéaire (ex: structure en spirale 3D).
* **Métriques adaptées :** Ratio de variance expliquée cumulé (*Explained Variance Ratio*).

### 4. Analyse de l'expérience Python
* **Entrées du code :** `PCA(n_components=2)` appliqué sur les données du vin `X_vin` prétraitées par `StandardScaler()`.
* **Hyperparamètres définis :** `n_components=2`.
* **Sortie obtenue :** Affiche le ratio de variance expliquée par les 2 premières composantes et le nuage projeté.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Indique la quantité exacte d'information conservée en passant à 2 dimensions sur le jeu du vin.
  - *Ne permet PAS de conclure :* Ne donne pas le cercle des corrélations (*loadings*) permettant d'identifier la contribution des variables initiales aux axes.

22. t-SNE
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Plongement voisin stochastique distribué selon une loi t / *t-Distributed Stochastic Neighbor Embedding (t-SNE)*.
* **Famille :** Apprentissage non supervisé — Réduction de dimension non linéaire (visualisation).
* **Objectif :** Projeter des données de haute dimension en 2D ou 3D en conservant prioritairement les relations de voisinage local entre observations.
* **Intuition & Fonctionnement :**
  t-SNE convertit les distances euclidiennes entre points en probabilités de proximité (loi gaussienne en haute dimension, loi de Student t en basse dimension pour éviter le problème d'entassement).
  1. *Entrée :* Données $X$ et paramètre de perplexité.
  2. *Calcul/Apprentissage :* Minimisation de la divergence de Kullback-Leibler (KL) entre les distributions de probabilités de l'espace d'origine et de l'espace réduit.
  3. *Sortie :* Coordonnées en 2D/3D optimisées uniquement pour la visualisation.
* **Formule mathématique :**
  Divergence KL à minimiser :
  $$KL(P \parallel Q) = \sum_{i} \sum_{j} p_{j|i} \log \frac{p_{j|i}}{q_{j|i}}$$
  - **Définition des symboles :** $p_{j|i}$ est la probabilité de proximité dans l'espace d'origine, $q_{j|i}$ la probabilité dans l'espace 2D réduit.
  - **Paramètres (appris) :** Coordonnées des points dans l'espace 2D/3D final.
  - **Hyperparamètres :** `perplexity`, `learning_rate`, `n_iter`, `random_state`.
* **Avantages :**
  1. Capable de révéler des structures visuelles de clusters complexes totalement invisibles avec l'ACP.
  2. Algorithme de visualisation non linéaire de référence.
* **Limites :**
  1. Ne conserve pas les distances globales (seuls les voisinages locaux ont un sens).
  2. Algorithme non déterministe, très coûteux en calcul et inutilisable pour transformer de nouvelles données de test (`transform` inexistant).

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** La figure illustre la projection 2D par t-SNE du jeu du vin, montrant la séparation visuelle nette des sous-groupes.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Détection de comportement) :** Exploration et inspection visuelle de profils de portefeuilles ou de fraudes suspectes.
* **Cas d'usage 2 (Bioinformatique) :** Visualisation de populations de cellules isolées (génomique).
* **Nature des données & Prérequis :** Variables numériques obligatoirement standardisées (`StandardScaler`).

23. UMAP
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** UMAP / *Uniform Manifold Approximation and Projection*.
* **Famille :** Apprentissage non supervisé — Réduction de dimension non linéaire (fondée sur la géométrie riemannienne).
* **Objectif :** Projeter des données de haute dimension dans un espace réduit tout en préservant mieux la structure globale de l'espace que t-SNE.
* **Intuition & Fonctionnement :**
  UMAP modélise la variété des données à l'aide de la géométrie floue des complexes simpliciaux, puis trouve une représentation en basse dimension dont le complexe équivalent est le plus similaire possible.
  1. *Entrée :* Données $X$ et hyperparamètre du nombre de voisins.
  2. *Calcul/Apprentissage :* Construction d'un graphe de voisinage flou, puis optimisation par descente de gradient de l'entropie croisée floue.
  3. *Sortie :* Projection en 2D/3D et possibilité de projeter de nouvelles observations (`transform`).
* **Formule mathématique :**
  Entropie croisée floue minimisée :
  $$C_{UMAP} = \sum_{i \neq j} \left[ p_{ij} \log \frac{p_{ij}}{q_{ij}} + (1 - p_{ij}) \log \frac{1 - p_{ij}}{1 - q_{ij}} \right]$$
  - **Définition des symboles :** $p_{ij}$ est la similarité floue dans l'espace d'origine, $q_{ij}$ la similarité dans l'espace de projection basse dimension.
  - **Paramètres (appris) :** Emplacements des points dans l'espace de dimension réduite.
  - **Hyperparamètres :** `n_neighbors`, `min_dist`, `n_components`, `metric`.
* **Avantages :**
  1. Plus rapide que t-SNE et préserve à la fois la structure locale et la structure globale.
  2. Dispose d'une méthode `.transform()` pour appliquer la projection sur de nouvelles observations.
* **Limites :**
  1. Repose sur une théorie mathématique abstraite complexe.
  2. Les hyperparamètres `n_neighbors` et `min_dist` influencent fortement l'aspect visuel des clusters.

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le schéma montre la projection 2D produite par UMAP du jeu du vin, mettant en évidence la préservation de la séparation globale et locale.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Marketing / Segmentation) :** Visualisation et compression d'espaces d'embeddings d'utilisateurs ou d'articles en e-commerce.
* **Cas d'usage 2 (Traitement du signal / PNL) :** Projection de représentations textuelles (*word embeddings*) pour l'analyse thématique.
* **Nature des données & Prérequis :** Variables numériques obligatoirement standardisées (`StandardScaler`).
* **Cas peu approprié :** Applications financières exigeant une interprétabilité directe et linéaire des coefficients.
* **Métriques adaptées :** Conservation de la structure du graphe de voisinage (*Trustworthiness*).

### 4. Analyse de l'expérience Python

* **Hyperparamètres définis :** `n_components=2`, `n_neighbors=15`, `random_state=GRAINE`.

  - *Permet de conclure :* Montre la capacité d'UMAP à compacter les clusters sur des données réelles.
  - *Ne permet PAS de conclure :* Ne montre pas l'effet d'une variation du paramètre `min_dist` sur la compacité interne des sous-groupes visualisés.

24. Isolation Forest
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Forêt d'isolement / *Isolation Forest*.
* **Famille :** Apprentissage non supervisé — Détection d'anomalies.
* **Objectif :** Isoler et détecter les observations atypiques (*outliers*) en exploitant leur faible densité et leur grande distance par rapport au reste des données.
* **Intuition & Fonctionnement :**
  Plutôt que de modéliser les points normaux, l'algorithme isole directement les anomalies. Les points anormaux nécessitent beaucoup moins de coupures aléatoires pour être isolés dans des feuilles d'arbres.
  1. *Entrée :* Données $X$ et taux de contamination estimé.
  2. *Calcul/Apprentissage :* Entraînement d'un ensemble d'arbres d'isolement aléatoires et mesure de la longueur moyenne du chemin d'accès aux points.
  3. *Sortie :* Score d'anomalie et étiquettes (1 pour normal, -1 pour anomalie).
* **Formule mathématique :**
  Score d'anomalie d'une observation $x$ :
  $$s(x, n) = 2^{-\frac{E(h(x))}{c(n)}}$$
  - **Définition des symboles :** $h(x)$ est la profondeur d'isolement du point dans un arbre, $E(h(x))$ la profondeur moyenne sur la forêt, $c(n)$ la profondeur moyenne théorique d'un arbre binaire de recherche.
  - **Paramètres (appris) :** Coupe des arbres d'isolement aléatoires.
  - **Hyperparamètres :** `contamination`, `n_estimators`, `max_samples`.
* **Avantages :**
  1. Extrêmement rapide et très sobre en mémoire ($\mathcal{O}(n \log n)$).
  2. Très performant sur de très grands jeux de données en haute dimension.
* **Limites :**
  1. Dépend du paramètre `contamination` réglé à la main si aucune donnée n'est étiquetée.
  2. Sensible aux découpages orthogonaux biaisés (résolu par Extended Isolation Forest).

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le schéma montre que les anomalies isolées sur les bords atteignent une feuille très rapidement (chemin court), tandis que les points normaux denses nécessitent de nombreuses coupes (chemin long).

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Détection de fraude) :** Détection d'anomalies de transactions bancaires frauduleuses ou suspectes.
* **Cas d'usage 2 (Industrie / Maintenance) :** Détection de pannes ou dysfonctionnements de capteurs industriels en temps réel.
* **Nature des données & Prérequis :** Variables numériques. Aucune mise à l'échelle impérative requise.
* **Cas peu approprié :** Jeux de données avec une proportion d'anomalies excessivement élevée (> 30%).
* **Métriques adaptées :** Précision, Rappel, F1-Score sur anomalies (si étiquettes), Taux d'alerte, AUC-ROC.

### 4. Analyse de l'expérience Python

  - *Ne permet PAS de conclure :* Ne garantit pas que les points détectés soient de vraies fraudes/erreurs sans validation par un expert métier.

25. Bandits manchots
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Bandits manchots / *Multi-Armed Bandits (MAB)*.
* **Famille :** Apprentissage par renforcement — Prise de décision sous incertitude.
* **Objectif :** Maximiser le gain cumulé en choisissant séquentiellement parmi plusieurs options (bras) aux récompenses inconnues, en gérant le compromis exploration/exploitation.
* **Intuition & Fonctionnement :**
  L'agent doit équilibrer l'exploration (tester des bras peu connus pour découvrir leur rentabilité) et l'exploitation (tirer le bras historiquement le plus rentable).
  1. *Entrée :* Nombre de bras $K$ et horizon temporel $T$.
  2. *Calcul/Apprentissage :* Sélection séquentielle d'un bras, observation de la récompense et mise à jour de la moyenne empirique du bras tiré.
  3. *Sortie :* Estimation de la valeur moyenne de chaque bras et allocation des tirages.
* **Formule mathématique :**
  Mise à jour empirique de la valeur d'un bras $a$ :
  $$Q(a) \leftarrow Q(a) + \frac{1}{N(a)} \left( R - Q(a) \right)$$
  - **Définition des symboles :** $Q(a)$ est l'estimation de la récompense moyenne du bras $a$, $N(a)$ le nombre de fois où le bras $a$ a été choisi, $R$ la récompense obtenue.
  - **Paramètres (appris) :** Valeurs estimées $Q(a)$ et compteurs $N(a)$.
  - **Hyperparamètres :** Horizon $T$, stratégie d'exploration.
* **Avantages :**
  1. Modèle très simple et efficace pour la prise de décision dynamique en ligne.
  2. Réduit drastiquement le coût d'opportunité par rapport aux tests A/B classiques statiques.
* **Limites :**
  1. Ne prend pas en compte l'état de l'environnement (pas d'état contextuel dans sa forme de base).
  2. Suppose que la distribution des récompenses des bras est stationnaire dans le temps.

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le schéma montre les distributions des gains associés aux différents bras et la convergence progressive des sélections vers le meilleur bras.

### 3. Contextes d me utilisation
* **Cas d'usage 1 (Marketing / E-commerce) :** Optimisation dynamique des affichages publicitaires ou des prix sur un site marchand (remplacement des tests A/B).
* **Cas d'usage 2 (Finance / Allotement) :** Allocation dynamique d'ordres d'exécution entre plusieurs plateformes de négociation.
* **Nature des données & Prérequis :** Flux de données séquentiel interactif en temps réel.
* **Cas peu approprié :** Problèmes où les actions actuelles modifient l'état futur de l'environnement de manière complexe.
* **Métriques adaptées :** Regret cumulé (*Cumulative Regret*), Récompense cumulée.

### 4. Analyse de l'expérience Python
* **Entrées du code :** Simulation sur 500 étapes avec 3 bras aux probabilités réelles `[0.2, 0.5, 0.7]`.
* **Hyperparamètres définis :** Horizon de 500 itérations, sélection par tour de rôle (Round-Robin simple `t % 3`).
* **Sortie obtenue :** Affiche la valeur moyenne estimée pour chaque bras après 500 tirages.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Montre que la loi des grands nombres permet d'estimer correctement la valeur théorique des 3 bras.
  - *Ne permet PAS de conclure :* Ne constitue pas un algorithme intelligent d'optimisation (le Round-Robin explore indéfiniment sans exploiter le meilleur bras).
2. Bloc Code Python pour le schéma (à copier dans une cellule Code)
Python
generateur = np.random.default_rng(42)
vraies_probas = np.array([.2, .5, .7])
somme = np.zeros(3)
comptes = np.zeros(3)
gains_cumules = []

for t in range(500):
    bras = t % 3
    gain = int(generateur.random() < vraies_probas[bras])
    somme[bras] += gain
    comptes[bras] += 1
    gains_cumules.append(somme.sum())

plt.figure(figsize=(7, 4))
plt.plot(gains_cumules, color='green', label='Récompense cumulée')
plt.title('Schéma explicatif : Accumulation de gains sur un bandit multi-bras')
plt.xlabel('Tirages (Temps t)')
plt.ylabel('Gain total')
plt.legend()
plt.grid(True, ls='--')
plt.tight_layout()
plt.show()
26. Epsilon-greedy
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Algorithme Epsilon-avide / *Epsilon-Greedy Algorithm*.
* **Famille :** Apprentissage par renforcement — Stratégie d'exploration pour bandits et RL.
* **Objectif :** Résoudre concrètement le compromis exploration/exploitation en choisissant une action aléatoire avec une probabilité $\epsilon$ et la meilleure action connue avec une probabilité $1-\epsilon$.
* **Intuition & Fonctionnement :**
  L'agent est la plupart du temps "avide" (il choisit le bras qui a le meilleur historique), mais réserve une petite fraction du temps ($\epsilon$) à l'exploration pure pour vérifier si un autre bras ne serait pas supérieur.
  1. *Entrée :* Paramètre d'exploration $\epsilon \in [0, 1]$ et estimations initiales.
  2. *Calcul/Apprentissage :* Tirage d'un nombre aléatoire $u \in [0, 1]$. Si $u < \epsilon$, action aléatoire ; sinon action $\arg\max Q(a)$.
  3. *Sortie :* Action choisie et mise à jour de la valeur estimée $Q(a)$.
* **Formule mathématique :**
  Politique de sélection d'action :
  $$\pi(a) = \begin{cases} 1 - \epsilon + \frac{\epsilon}{|A|} & \text{si } a = \arg\max Q(a) \\ \frac{\epsilon}{|A|} & \text{si } a \neq \arg\max Q(a) \end{cases}$$
  - **Définition des symboles :** $\epsilon$ est le taux d'exploration, $|A|$ le nombre total d'actions possibles, $Q(a)$ la valeur estimée de l'action $a$.
  - **Paramètres (appris) :** Valeurs d'actions $Q(a)$ et compteurs d'exécutions.
  - **Hyperparamètres :** $\epsilon$ (fixe ou décroissant).
* **Avantages :**
  1. Stratégie extrêmement simple, intuitive et d'une grande efficacité pratique.
  2. Assure qu'aucune action n'est totalement ignorée sur le long terme.
* **Limites :**
  1. Lors de l'exploration, l'agent choisit de manière indifférenciée une très mauvaise action ou une action presque optimale.
  2. Nécessite d'ajuster le paramètre $\epsilon$ (ou de programmer sa décroissance).

### 2. Schéma explicatif
*(Voir la cellule de code ci-dessous pour la génération du schéma Matplotlib)*

* **Commentaire du schéma :** Le diagramme circulaire schématise la répartition Epsilon-Greedy : $(1-\epsilon)$ réservé à l'exploitation avide et $\epsilon$ dédié à l'exploration équirégulière.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Finance / Trading) :** Exploration de nouvelles stratégies d'arbitrage tout en exécutant majoritairement la meilleure stratégie connue.
* **Cas d'usage 2 (Jeux / IA) :** Stratégie de contrôle dans le cadre de l'apprentissage d'agents sur des environnements simulés (ex: Gymnasium).
* **Nature des données & Prérequis :** Interactions séquentielles en ligne avec système de récompenses.
* **Cas peu approprié :** Environnements où une action d'exploration aléatoire peut provoquer un dommage critique ou mortel (ex: conduite autonome).
* **Métriques adaptées :** Récompense cumulée, Taux d'utilisation du meilleur bras (*Optimal Action %*).

### 4. Analyse de l'expérience Python
* **Entrées du code :** Simulation de 1000 tirages avec $\epsilon=0.1$ et probabilités réelles `[0.2, 0.5, 0.7]`.
* **Hyperparamètres définis :** `epsilon=0.1`, 1000 itérations.
* **Sortie obtenue :** Affiche le nombre de choix par bras et les valeurs estimées finales.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Montre que le bras le plus rentable (index 2 avec probabilité 0.7) a été sélectionné de manière très largement majoritaire.
  - *Ne permet PAS de conclure :* Ne compare pas cette performance avec une stratégie UCB (*Upper Confidence Bound*) qui explore plus intelligemment.

27. Q-learning
1. Bloc Markdown (à copier dans une cellule Texte)
Markdown
### 1. Présentation
* **Nom (FR / EN) :** Q-apprentissage / *Q-Learning*.
* **Famille :** Apprentissage par renforcement — Contrôle hors-politique (*Off-Policy Temporal Difference Learning*).
* **Objectif :** Apprendre la fonction de valeur action-état optimale $Q^*(s, a)$ permettant à un agent de prendre les meilleures décisions dans un Processus de Décision de Markov (MDP).
* **Intuition & Fonctionnement :**
  L'agent interagit avec l'environnement et met à jour sa table $Q$ après chaque action en utilisant la meilleure récompense future estimée, indépendamment de l'action qu'il exécutera réellement à l'étape suivante (hors-politique).
  1. *Entrée :* Espace d'états $\mathcal{S}$, espace d'actions $\mathcal{A}$, taux d'apprentissage $\alpha$ et facteur de remise $\gamma$.
  2. *Calcul/Apprentissage :* Interaction $(s, a, r, s')$ et mise à jour de Bellman de la table $Q(s, a)$.
  3. *Sortie :* Politique optimale $\pi^*(s) = \arg\max_a Q(s, a)$.
* **Formule mathématique :**
  Équation de mise à jour du Q-learning (Bellman) :
  $$Q(s, a) \leftarrow Q(s, a) + \alpha \left[ r + \gamma \max_{a'} Q(s', a') - Q(s, a) \right]$$
  - **Définition des symboles :** $s$ est l'état actuel, $a$ l'action prise, $r$ la récompense perçue, $s'$ l'état suivant, $\alpha$ le taux d'apprentissage, $\gamma$ le facteur de remise (*discount factor*).
  - **Paramètres (appris) :** La table des Q-valeurs $Q(s, a)$ de dimension $|\mathcal{S}| \times |\mathcal{A}|$.
  - **Hyperparamètres :** Taux d'apprentissage $\alpha$, facteur de remise $\gamma$, stratégie $\epsilon$-greedy.
* **Avantages :**
  1. Apprend la politique optimale directement sans nécessiter de modèle de transition de l'environnement (*model-free*).
  2. Algorithme hors-politique (*off-policy*) qui peut apprendre à partir de trajectoires de démonstration passées.
* **Limites :**
  1. Limité aux espaces d'états et d'actions discrets et de petite taille (souffre du fléau de la dimension sous forme tabulaire).
  2. Surestime souvent les Q-valeurs en raison de l'opérateur $\max$.

### 2. Schéma explicatif

````python
import gymnasium as gym

env = gym.make('FrozenLake-v1', is_slippery=False)
Q_demo = np.zeros((env.observation_space.n, env.action_space.n))

plt.figure(figsize=(7, 4.5))
plt.imshow(Q_demo, cmap='Blues', aspect='auto')
plt.colorbar(label='Q-Valeur initiale')
plt.title('Schéma explicatif : Représentation de la Table Q(s,a) [16 états x 4 actions]')
plt.xlabel('Actions (0: Gauche, 1: Bas, 2: Droite, 3: Haut)')
plt.ylabel('États de la grille (0 à 15)')
plt.tight_layout()
plt.show()
env.close()
````

* **Commentaire du schéma :** Le schéma illustre la boucle d'interaction Agent-Environnement et la mise à jour de la table des Q-valeurs basée sur l'équation de Bellman.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Logistique / Robotique) :** Guidage automatique de véhicules autonomes (*AGV*) dans un entrepôt quadrillé.
* **Cas d'usage 2 (Gestion d'énergie / Smart Grids) :** Contrôle optimal de la charge et décharge de batteries réseau selon le tarif horaire.
* **Nature des données & Prérequis :** Environnement de simulation interactif (ex: Gymnasium).
* **Cas peu approprié :** Environnements complexes à états continus comme la vision par ordinateur ou le pilotage d'avions (exige du Deep Q-Learning).
* **Métriques adaptées :** Récompense cumulée par épisode, Taux de succès (*Success Rate*), Convergence des Q-valeurs.

### 4. Analyse de l'expérience Python
* **Entrées du code :** Environnement `FrozenLake-v1` (non glissant), 4000 épisodes, table $Q$ $16 \times 4$.
* **Hyperparamètres définis :** $\alpha=0.3$, $\gamma=0.95$, $\epsilon$ décroissant de 1.0 à 0.05.
* **Sortie obtenue :** Affiche la politique optimale apprise sous forme de tableau d'actions `Q.argmax(axis=1)`.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Confirme que l'agent a trouvé un chemin valide menant au frisbee sur la grille sans tomber dans les trous.
  - *Ne permet PAS de conclure :* Ne garantit pas que cette politique reste valide si le lac devient glissant (`is_slippery=True`).

## 28. SARSA

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));generateur=np.random.default_rng(GRAINE)
def choisir(etat,epsilon):
    return int(generateur.integers(4)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.05,1-episode/2500);action=choisir(etat,epsilon)
    for pas in range(100):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir(suivant,epsilon)
        cible=recompense+(0 if termine else .95*Q[suivant,action_suiv])
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('Politique SARSA:',Q.argmax(axis=1));env.close()
````

### 1. Présentation
* **Nom (FR / EN) :** SARSA / *State-Action-Reward-State-Action*.
* **Famille :** Apprentissage par renforcement — Contrôle sur-politique (*On-Policy Temporal Difference Learning*).
* **Objectif :** Apprendre la fonction de valeur action-état $Q^\pi(s, a)$ tout en évaluant et en améliorant la politique réellement exécutée par l'agent.
* **Intuition & Fonctionnement :**
  Contrairement au Q-Learning, SARSA est un algorithme **sur-politique** (*on-policy*). La mise à jour de l'état $s$ et de l'action $a$ utilise la véritable action suivante $a'$ réellement sélectionnée par la politique courante ($\epsilon$-greedy).
  1. *Entrée :* Espace d'états $\mathcal{S}$, espace d'actions $\mathcal{A}$, hyperparamètres $\alpha, \gamma$.
  2. *Calcul/Apprentissage :* Séquence $(s, a, r, s', a')$, puis mise à jour de la valeur $Q(s, a)$ en intégrant $Q(s', a')$.
  3. *Sortie :* Politique optimale prudente adaptée au comportement réel de l'agent.
* **Formule mathématique :**
  Équation de mise à jour de SARSA :
  $$Q(s, a) \leftarrow Q(s, a) + \alpha \left[ r + \gamma Q(s', a') - Q(s, a) \right]$$
  - **Définition des symboles :** $s$ état actuel, $a$ action actuelle, $r$ récompense, $s'$ état suivant, $a'$ action suivante réellement choisie.
  - **Paramètres (appris) :** La table des Q-valeurs $Q(s, a)$.
  - **Hyperparamètres :** $\alpha, \gamma, \epsilon$.
* **Avantages :**
  1. Plus prudent et sécurisé que le Q-Learning en présence d'exploration risquée ou de pénalités importantes.
  2. Prend en compte le risque d'erreurs d'exploration réelles de l'agent pendant son apprentissage.
* **Limites :**
  1. Peut converger vers une solution sous-optimale (plus prudente) si l'exploration reste élevée.
  2. Nécessite également un espace d'états et d'actions discrets sous sa forme tabulaire.

### 2. Schéma explicatif

````python
fig, ax = plt.subplots(figsize=(7, 3))
steps = ['État S_t', 'Action A_t', 'Récompense R_t+1', 'État S_t+1', 'Action A_t+1']
colors = ['lightblue', 'lightgreen', 'gold', 'lightblue', 'lightgreen']

for i, (step, col) in enumerate(zip(steps, colors)):
    ax.text(i*0.25, 0.5, step, bbox=dict(boxstyle='round', facecolor=col, edgecolor='b'), fontsize=10, ha='center')
    if i < 4:
        ax.annotate('', xy=((i+1)*0.25 - 0.05, 0.5), xytext=(i*0.25 + 0.05, 0.5),
                    arrowprops=dict(arrowstyle="->", lw=1.5))

ax.set_xlim(-0.15, 1.15)
ax.set_ylim(0, 1)
ax.axis('off')
plt.title('Schéma explicatif : Séquence temporelle SARSA (On-Policy)')
plt.tight_layout()
plt.show()
````

**Commentaire du schéma :** Le schéma illustre le quintuplet $(S_t, A_t, R_{t+1}, S_{t+1}, A_{t+1})$ qui donne son nom à l'algorithme SARSA.

### 3. Contextes d'utilisation
* **Cas d'usage 1 (Robotique / Sécurité) :** Apprentissage du déplacement d'un robot en évitant les falaises ou obstacles dangereux lors des phases d'exploration.
* **Cas d'usage 2 (Finance / Gestion de portefeuille) :** Prise de décision sous risque d'exécution avec prise en compte du dérapage des ordres (*slippage*).
* **Nature des données & Prérequis :** Simulation interactive en temps réel (Gymnasium).
* **Cas peu approprié :** Apprentissage purement passif à partir de bases de données historiques enregistrées par d'autres agents.
* **Métriques adaptées :** Récompense cumulée pendant l'apprentissage, Stabilité de la politique.

* **Hyperparamètres définis :** $\alpha=0.3$, $\gamma=0.95$, $\epsilon$ décroissant.
* **Sortie obtenue :** Affiche la politique finale calculée avec SARSA.
* **Ce que la sortie permet / ne permet pas de conclure :**
  - *Permet de conclure :* Démontre la convergence vers une politique valide sur la grille.
  - *Ne permet PAS de conclure :* Ne montre pas la différence de trajectoire "prudente" par rapport au Q-learning sans comparer les deux sur un environnement avec pièges mortels (ex: *CliffWalking*).

## 29. DQN — cible conceptuelle

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

````python
# Calcul de cible DQN, sans entraîner un réseau neuronal complet
recompenses=np.array([0.,1.,-1.]);termine=np.array([False,True,False]);meilleur_Q_suivant=np.array([2.,3.,1.]);gamma=.95
cibles=recompenses+gamma*(~termine)*meilleur_Q_suivant
print('Cibles de Bellman:',cibles)
# Un DQN complet ajoute réseau en ligne, réseau cible et mémoire de rejeu.
````

### 1. Présentation

* **Nom (FR / EN) :** Réseau Q-profond / *Deep Q-Network (DQN)*.
* **Famille :** Apprentissage par renforcement profond — Contrôle par réseaux de neurones.
* **Objectif :** Généraliser le Q-Learning à des espaces d'états continus et de haute dimension (ex: images brutes, capteurs complexes) en remplaçant la table $Q$ tabulaire par un réseau de neurones profond approximateur de fonction.
* **Intuition & Fonctionnement :** Le réseau de neurones prend en entrée le vecteur d'état continu $s$ et prédit en sortie la valeur $Q(s, a)$ pour chaque action possible. L'apprentissage est stabilisé par une mémoire de rejeu (*Replay Buffer*) et un réseau cible (*Target Network*) mis à jour périodiquement.
  1. *Entrée :* Données $X$ (état continu $s \in \mathbb{R}^d$, récompense $r$, état suivant $s'$, indicateur de fin d'épisode).
  2. *Calcul/Apprentissage :* Échantillonnage de mini-lots de transitions passées $(s, a, r, s')$ dans le Replay Buffer, calcul des cibles de Bellman avec le Target Network, et rétropropagation du gradient de la perte MSE.
  3. *Sortie :* Estimation des Q-valeurs $Q(s, a)$ pour choisir l'action optimale $\arg\max_a Q(s, a)$.
* **Formule mathématique :** Calcul de la cible de Bellman et fonction de perte du réseau DQN :
  $$y_i = r + \gamma (1 - \text{terminé}) \max_{a'} Q_{\text{target}}(s', a'; \theta^-)$$
  $$\mathcal{L}(\theta) = \frac{1}{B} \sum_{i=1}^{B} \left( y_i - Q(s_i, a_i; \theta) \right)^2$$
  o **Définition des symboles :** $Q(s, a; \theta)$ est le réseau principal paramétré par $\theta$, $Q_{\text{target}}$ le réseau cible paramétré par $\theta^-$, $r$ la récompense, $\gamma$ le facteur de remise, $B$ la taille du batch.
  o **Paramètres (appris) :** Poids et biais $\theta$ des couches du réseau de neurones.
  o **Hyperparamètres :** `learning_rate`, `buffer_size`, `batch_size`, `target_update_interval`, `gamma`, `epsilon`.
* **Avantages :**
  1. Traite nativement des espaces d'états continus et de haute dimension sans discrétisation manuelle.
  2. Stabilisé efficacement grâce au découplage apporté par la mémoire de rejeu et le réseau cible.
* **Limites :**
  1. Entraînement parfois instable et très sensible au réglage précis des hyperparamètres.
  2. Ne convient nativement qu'aux espaces d'actions discrets (nécessite des architectures comme DDPG ou SAC pour des actions continues).

### 2. Schéma explicatif

````python
import numpy as np
import matplotlib.pyplot as plt

# Simulation vectorisée du calcul des cibles de Bellman
recompenses = np.array([0., 1., -1.])
termine = np.array([False, True, False])
meilleur_Q_suivant = np.array([2., 3., 1.])
gamma = 0.95
cibles = recompenses + gamma * (~termine) * meilleur_Q_suivant

plt.figure(figsize=(8, 4.5))
plt.bar(range(1, 4), cibles, color='crimson', alpha=0.85)
plt.xticks(range(1, 4), ['Transition 1 (En cours)', 'Transition 2 (Terminé)', 'Transition 3 (En cours)'])
plt.ylabel('Valeur Cible de Bellman (y_i)')
plt.title('Schéma explicatif : Calcul des cibles de Bellman dans un DQN')
plt.grid(True, ls='--', alpha=0.6)
plt.tight_layout()
plt.show()
````

Commentaire du schéma : Le schéma illustre le calcul vectorisé des cibles de Bellman $y_i$ sur 3 transitions d'exemple : la valeur future est intégrée avec le facteur d'atténuation $\gamma=0.95$, mais s'annule immédiatement lorsque l'état atteint est terminal (termine = True).3. ## Annexes — implémentations manuelles des trois algorithmes imposés

````python
# Régression linéaire par descente de gradient : données univariées
x=np.linspace(-1,1,100); y=2+3*x
poids=0.;biais=0.;pas=.1
for iteration in range(500):
    prediction=poids*x+biais
    erreur=prediction-y
    poids-=pas*2*np.mean(erreur*x)
    biais-=pas*2*np.mean(erreur)
print('Paramètres appris manuellement:',poids,biais)
from sklearn.linear_model import LinearRegression
print('Référence sklearn:',LinearRegression().fit(x.reshape(-1,1),y).coef_,LinearRegression().fit(x.reshape(-1,1),y).intercept_)
````
